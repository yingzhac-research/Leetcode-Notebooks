## Problem Statement
Given strings `s1`, `s2`, and `s3`, find whether `s3` is formed by an **interleaving** of `s1` and `s2`.

An **interleaving** of two strings `s` and `t` is a configuration where they are divided into **non-empty** substrings such that:

- `s = s1 + s2 + ... + sn`
- `t = t1 + t2 + ... + tm`
- `|n - m| <= 1`
- The interleaving is `s1 + t1 + s2 + t2 + ...` or `t1 + s1 + t2 + s2 + ...`

**Note**: `a + b` is the concatenation of strings `a` and `b`.

**Constraints**:
- `0 <= s1.length, s2.length <= 100`
- `0 <= s3.length <= 200`
- `s1`, `s2`, and `s3` consist of lowercase English letters.

## Progressive Hints
- Hint 1: Check if len(s3) == len(s1) + len(s2).
- Hint 2: Use 2D DP where dp[i][j] = can we form s3[0:i+j] using s1[0:i] and s2[0:j].
- Hint 3: At each position, either take from s1 or s2.
- Hint 4: dp[i][j] = (dp[i-1][j] and s1[i-1]==s3[i+j-1]) or (dp[i][j-1] and s2[j-1]==s3[i+j-1]).
- Hint 5: Base case: dp[0][0] = True.
- Hint 6: Can optimize to 1D DP using rolling array.

## Solution Overview
Use **2D DP**.

**State**: `dp[i][j]` = can form s3[0:i+j] using s1[0:i] and s2[0:j]

**Recurrence**:
```python
dp[i][j] = (dp[i-1][j] and s1[i-1] == s3[i+j-1]) or 
           (dp[i][j-1] and s2[j-1] == s3[i+j-1])
```

## Detailed Explanation

### Problem Understanding

Interleaving: maintain relative order of characters from each string.

Example: s1="ab", s2="cd", s3="acbd" ✓ (a from s1, c from s2, b from s1, d from s2)

### 2D DP Approach

**State meaning**: dp[i][j] means we've used first i chars of s1 and first j chars of s2, forming first i+j chars of s3.

**Transitions**: At position (i,j), we can:
1. Take from s1: if s1[i-1] matches s3[i+j-1] and dp[i-1][j] is true
2. Take from s2: if s2[j-1] matches s3[i+j-1] and dp[i][j-1] is true

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| 2D DP | O(m×n) | O(m×n) | Standard approach |
| 1D DP | O(m×n) | O(n) | Space optimized |
| DFS + memo | O(m×n) | O(m×n) | Top-down version |
| BFS | O(m×n) | O(m×n) | Queue-based |

````python
class Solution:
    def isInterleave(self, s1: str, s2: str, s3: str) -> bool:
        m, n = len(s1), len(s2)
        if m + n != len(s3):
            return False
        
        dp = [[False] * (n + 1) for _ in range(m + 1)]
        dp[0][0] = True
        
        # First row: only use s2
        for j in range(1, n + 1):
            dp[0][j] = dp[0][j-1] and s2[j-1] == s3[j-1]
        
        # First column: only use s1
        for i in range(1, m + 1):
            dp[i][0] = dp[i-1][0] and s1[i-1] == s3[i-1]
        
        # Fill table
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                dp[i][j] = (dp[i-1][j] and s1[i-1] == s3[i+j-1]) or \
                           (dp[i][j-1] and s2[j-1] == s3[i+j-1])
        
        return dp[m][n]
````

````python
# Test cases
tests = [
    ('aabcc', 'dbbca', 'aadbbcbcac', True),
    ('aabcc', 'dbbca', 'aadbbbaccc', False),
    ('', '', '', True),
]

solver = Solution()
for s1, s2, s3, expected in tests:
    result = solver.isInterleave(s1, s2, s3)
    assert result == expected
print('All tests passed.')
````

- **Space**: O(m × n), can optimize to O(n)

## Edge Cases & Pitfalls
- Length mismatch: return False immediately
- Empty strings: handle correctly
- All same characters: need DP, greedy fails

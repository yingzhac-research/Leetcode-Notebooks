## Problem Statement
Given a string `s`, return the number of **palindromic substrings** in it.

A string is a **palindrome** when it reads the same backward as forward.

A **substring** is a contiguous sequence of characters within the string.

**Constraints**:
- 1 <= s.length <= 1000
- s consists of lowercase English letters

## Progressive Hints
- Hint 1: This is similar to LC 5 but counts all palindromic substrings, not just the longest.
- Hint 2: Use expand around center: for each possible center, expand and count palindromes.
- Hint 3: There are 2n-1 possible centers (n single + n-1 double).
- Hint 4: Alternatively, use DP: dp[i][j] = True if s[i:j+1] is palindrome, count all True values.
- Hint 5: Expand around center is more space-efficient: O(1) space vs O(n²).

## Solution Overview
Count all palindromic substrings.

**Approaches**:
1. **Expand Around Center** (O(n²) time, O(1) space):
   - For each center, expand while palindrome
   - Count each valid palindrome

2. **Interval DP** (O(n²) time, O(n²) space):
   - dp[i][j] = True if s[i:j+1] is palindrome
   - Count all True values

3. **Manacher's Algorithm** (O(n) time, O(n) space):
   - Advanced linear-time algorithm
   - Can count palindromes in O(n)

**Best for this problem**: Expand around center (simple and space-efficient)

## Detailed Explanation

### Expand Around Center Approach

**Key Idea**: For each possible center, expand outward and count palindromes.

**Centers**:
- **Odd-length palindromes**: n centers (each character)
- **Even-length palindromes**: n-1 centers (between each pair)
- Total: 2n - 1 centers

**Algorithm**:
```python
count = 0
for center in all_centers:
    left, right = center_positions
    while left >= 0 and right < n and s[left] == s[right]:
        count += 1  # Found a palindrome
        left -= 1
        right += 1
return count
```

**Difference from LC 5**:
- **LC 5**: Find length of longest palindrome
- **LC 647**: Count all palindromes (increment count at each expansion)

---

### Example Walkthrough

**Input**: s = "abc"

**Process**:
1. Center at 'a' (index 0):
   - Odd: "a" ✓ (count = 1)
   - Even: 'a' vs 'b', no match

2. Center at 'b' (index 1):
   - Odd: "b" ✓ (count = 2)
   - Even: 'b' vs 'c', no match

3. Center at 'c' (index 2):
   - Odd: "c" ✓ (count = 3)

**Result**: count = 3 ("a", "b", "c")

---

**Input**: s = "aaa"

**Process**:
1. Center at index 0:
   - Odd: "a" ✓ (count = 1)
   - Even: "aa" ✓ (count = 2)

2. Center at index 1:
   - Odd: "a" ✓, "aaa" ✓ (count = 4)
   - Even: "aa" ✓ (count = 5)

3. Center at index 2:
   - Odd: "a" ✓ (count = 6)

**Result**: count = 6 ("a", "a", "a", "aa", "aa", "aaa")

### Interval DP Approach

**State**: dp[i][j] = True if s[i:j+1] is palindrome

**Recurrence**:
```python
dp[i][j] = (s[i] == s[j]) and (j - i <= 2 or dp[i+1][j-1])
```

**Count**:
```python
count = sum(sum(row) for row in dp)
```

**Space**: O(n²) for DP table

---

### Comparison: Expand vs DP

| Aspect | Expand Around Center | Interval DP |
|--------|---------------------|-------------|
| Time | O(n²) | O(n²) |
| Space | O(1) | O(n²) |
| Simplicity | Very simple | More complex |
| Extensibility | Hard to modify | Easy to extend |

For this problem, **expand around center** is better (simpler and less space).

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Brute force | O(n³) | O(1) | Check all substrings |
| Expand around center | O(n²) | O(1) | Most practical |
| Interval DP | O(n²) | O(n²) | Uses DP table |
| Manacher's algorithm | O(n) | O(n) | Complex but optimal |

````python
class Solution:
    def countSubstrings(self, s: str) -> int:
        """
        Expand around center approach.
        
        Time: O(n²)
        Space: O(1)
        """
        count = 0
        n = len(s)
        
        def expandAroundCenter(left: int, right: int) -> int:
            """Expand around center and count palindromes."""
            palindrome_count = 0
            while left >= 0 and right < n and s[left] == s[right]:
                palindrome_count += 1  # Found a palindrome
                left -= 1
                right += 1
            return palindrome_count
        
        for i in range(n):
            # Odd-length palindromes (single center)
            count += expandAroundCenter(i, i)
            # Even-length palindromes (double center)
            count += expandAroundCenter(i, i + 1)
        
        return count
````

````python
# Test cases
tests = [
    ("abc", 3),      # "a", "b", "c"
    ("aaa", 6),      # "a", "a", "a", "aa", "aa", "aaa"
    ("a", 1),        # "a"
    ("aba", 4),      # "a", "b", "a", "aba"
    ("abba", 6),     # "a", "b", "b", "a", "bb", "abba"
    ("racecar", 10), # Multiple palindromes
    ("", 0),         # Empty (by constraints won't happen)
]

solver = Solution()
for s, expected in tests[:-1]:  # Skip empty string test
    result = solver.countSubstrings(s)
    assert result == expected, f"Failed for s='{s}': got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n²) where n = len(s)
  - Try 2n - 1 centers
  - Each expansion can take up to O(n) time
  - Total: O(n) × O(n) = O(n²)
- **Space**: O(1)
  - Only use constant variables
  - No additional data structures

## Edge Cases & Pitfalls
- **Single character**: Return 1
- **All same characters**: For string of length n with all same chars, count = n(n+1)/2
- **No palindrome > 1**: Count equals n (all single characters)
- **Odd vs even**: Must handle both center types
- **Counting**: Increment count at EACH expansion step, not just once per center
- **Index bounds**: Ensure left >= 0 and right < n
- **Comparison with LC 5**: This counts all, LC 5 finds longest

## Takeaways
- **Count at Each Step**: Unlike LC 5 (track max length), here we count every palindrome found
- **Expand Around Center**: Most practical approach for counting
- **Two Center Types**: Always handle both odd and even-length palindromes
- **Space Efficiency**: O(1) space makes this better than DP for this problem
- **Formula for Same Chars**: n identical characters have n(n+1)/2 palindromic substrings
- **Pattern Reuse**: Same expansion logic as LC 5, just different goal

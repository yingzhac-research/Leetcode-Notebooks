## Problem Statement
Given an integer array `nums`, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.

A **subarray** is a contiguous part of an array.

## Progressive Hints
- Hint 1: At each position, decide whether to extend the current subarray or start a new one.
- Hint 2: If the current sum becomes negative, it's better to start fresh from the next element.
- Hint 3: Keep track of both the current subarray sum and the global maximum sum.
- Hint 4: This is Kadane's Algorithm - a classic O(n) greedy/DP approach.

## Solution Overview
Use **Kadane's Algorithm**: maintain two variables:
- `current_sum`: maximum sum ending at the current position
- `max_sum`: global maximum sum seen so far

At each element, update `current_sum = max(nums[i], current_sum + nums[i])`, which decides whether to extend the current subarray or start anew. Then update `max_sum = max(max_sum, current_sum)`.

Alternative approaches include divide-and-conquer (O(n log n)) and prefix sums, but Kadane's is optimal.

## Detailed Explanation
**Approach: Kadane's Algorithm (DP)**

1. **Core idea**: At each position i, we have two choices:
   - Extend the previous subarray: `current_sum + nums[i]`
   - Start a new subarray: `nums[i]`
   
   Choose the larger option.

2. **DP formulation**:
   ```
   dp[i] = max(nums[i], dp[i-1] + nums[i])
   ```
   where `dp[i]` is the maximum sum ending at index i.

3. **Space optimization**: We only need the previous dp value, so use a single variable `current_sum`.

4. **Algorithm steps**:
   - Initialize `current_sum = nums[0]`, `max_sum = nums[0]`
   - For each element from index 1 to n-1:
     - `current_sum = max(nums[i], current_sum + nums[i])`
     - `max_sum = max(max_sum, current_sum)`
   - Return `max_sum`

**Example Walkthrough** ([-2, 1, -3, 4, -1, 2, 1, -5, 4]):
- i=0: current=−2, max=−2
- i=1: current=max(1, −2+1)=1, max=1
- i=2: current=max(−3, 1−3)=−2, max=1
- i=3: current=max(4, −2+4)=4, max=4
- i=4: current=max(−1, 4−1)=3, max=4
- i=5: current=max(2, 3+2)=5, max=5
- i=6: current=max(1, 5+1)=6, max=6
- i=7: current=max(−5, 6−5)=1, max=6
- i=8: current=max(4, 1+4)=5, max=6
- Result: 6 (subarray [4, -1, 2, 1])

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Brute force | O(n^2) | O(1) | Check all subarrays |
| Kadane's algorithm | O(n) | O(1) | Optimal greedy/DP approach |
| Divide and conquer | O(n log n) | O(log n) | Recursive, educational but slower |
| Prefix sum | O(n) | O(n) | Track min prefix, less elegant |

````python
from typing import List

class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        """
        Kadane's Algorithm for maximum subarray sum.
        
        Time: O(n)
        Space: O(1)
        """
        current_sum = max_sum = nums[0]
        
        for i in range(1, len(nums)):
            # Either extend current subarray or start new
            current_sum = max(nums[i], current_sum + nums[i])
            # Update global maximum
            max_sum = max(max_sum, current_sum)
        
        return max_sum
````

````python
# Test cases
tests = [
    ([-2,1,-3,4,-1,2,1,-5,4], 6),     # [4,-1,2,1]
    ([1], 1),                          # Single element
    ([5,4,-1,7,8], 23),                # All positive except one
    ([-1], -1),                        # Single negative
    ([-2,-1], -1),                     # All negative
    ([1,2,3,4,5], 15),                 # All positive
    ([-1,-2,-3,-4], -1),               # All negative
]

solver = Solution()
for nums, expected in tests:
    result = solver.maxSubArray(nums)
    assert result == expected, f"Failed for {nums}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n) - Single pass through the array.
- **Space**: O(1) - Only two variables used.

## Edge Cases & Pitfalls
- **All negative numbers**: Must return the largest (least negative) single element.
- **Single element**: Return that element.
- **Empty array**: Problem guarantees at least one element.
- **Integer overflow**: For very large arrays or values, consider using long or checking bounds.
- **Finding the actual subarray**: If you need indices, track start and end positions while updating max_sum.
- **Initialization**: Start with nums[0], not 0, to handle all-negative arrays correctly.

## Takeaways
- Kadane's Algorithm is a fundamental technique for subarray optimization problems.
- The key insight is maintaining the maximum sum ending at the current position.
- This greedy approach is actually a form of dynamic programming with O(1) space optimization.
- The pattern extends to many variants including circular arrays, 2D matrices, and product-based problems.
- Understanding when to "reset" the current sum (when it goes negative) is crucial.

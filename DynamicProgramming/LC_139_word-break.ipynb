## Problem Statement
Given a string `s` and a dictionary of strings `wordDict`, return `true` if `s` can be segmented into a space-separated sequence of one or more dictionary words.

**Note** that the same word in the dictionary may be reused multiple times in the segmentation.

## Progressive Hints
- Hint 1: Think about whether a prefix of the string can be segmented.
- Hint 2: Use DP where dp[i] represents whether s[0:i] can be segmented.
- Hint 3: For each position i, check all positions j < i: if dp[j] is true and s[j:i] is in wordDict, then dp[i] is true.
- Hint 4: Use a set for wordDict for O(1) lookup.
- Hint 5: Base case: dp[0] = true (empty string can be segmented).

## Solution Overview
Use dynamic programming where `dp[i]` indicates whether the substring `s[0:i]` can be segmented using words from the dictionary.

**Recurrence**:
```
dp[i] = true if there exists j < i such that:
        dp[j] == true AND s[j:i] in wordDict
```

**Base case**: `dp[0] = true` (empty string)

Convert wordDict to a set for O(1) lookup. The final answer is `dp[n]` where n is the length of s.

## Detailed Explanation
**Approach: Bottom-Up Dynamic Programming**

1. **State Definition**: 
   - `dp[i]` = whether substring `s[0:i]` can be segmented
   - Array size: n+1 (to handle empty string at index 0)

2. **Transition Logic**:
   - For each position i from 1 to n:
     - Check all possible split points j from 0 to i-1
     - If `dp[j]` is true (prefix s[0:j] can be segmented) AND
       `s[j:i]` is in wordDict (current word is valid)
     - Then `dp[i] = true`

3. **Optimization**:
   - Convert wordDict to set for O(1) lookup
   - Can break early if dp[i] is already true
   - Can optimize inner loop by checking only lengths present in wordDict

4. **Algorithm Steps**:
   ```
   dp = [False] * (n + 1)
   dp[0] = True
   word_set = set(wordDict)
   
   for i in range(1, n + 1):
       for j in range(i):
           if dp[j] and s[j:i] in word_set:
               dp[i] = True
               break
   
   return dp[n]
   ```

**Example Walkthrough** (s="leetcode", wordDict=["leet","code"]):
```
i=0: dp[0]=True (base case)
i=1: s[0:1]="l" not in dict → dp[1]=False
i=2: s[0:2]="le" not in dict → dp[2]=False
i=3: s[0:3]="lee" not in dict → dp[3]=False
i=4: s[0:4]="leet" in dict, dp[0]=True → dp[4]=True
i=5: s[4:5]="c", dp[4]=True but "c" not in dict → dp[5]=False
i=6: s[4:6]="co", dp[4]=True but "co" not in dict → dp[6]=False
i=7: s[4:7]="cod", dp[4]=True but "cod" not in dict → dp[7]=False
i=8: s[4:8]="code", dp[4]=True and "code" in dict → dp[8]=True
Result: True
```

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Brute force (recursion) | O(2^n) | O(n) | Try all segmentations |
| DP (bottom-up) | O(n^2 * m) | O(n) | m = avg word length for substring |
| DP with optimization | O(n^2) | O(n) | Using set lookup as O(1) |
| Trie-based DP | O(n^2) | O(total chars) | Faster word lookup |

````python
from typing import List

class Solution:
    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        """
        DP solution for word break.
        
        Time: O(n^2 * m) where m is average word length (substring operation)
              In practice, O(n^2) with set hashing
        Space: O(n) for dp array
        """
        n = len(s)
        word_set = set(wordDict)
        
        # dp[i] means s[0:i] can be segmented
        dp = [False] * (n + 1)
        dp[0] = True  # Empty string
        
        for i in range(1, n + 1):
            for j in range(i):
                # If prefix s[0:j] can be segmented and s[j:i] is a word
                if dp[j] and s[j:i] in word_set:
                    dp[i] = True
                    break  # Found one valid segmentation, move to next i
        
        return dp[n]
````

````python
# Test cases
tests = [
    ("leetcode", ["leet","code"], True),
    ("applepenapple", ["apple","pen"], True),
    ("catsandog", ["cats","dog","sand","and","cat"], False),
    ("a", ["a"], True),
    ("", ["a"], True),  # Empty string
    ("aaaaaaa", ["aaaa","aaa"], True),  # Word reuse
    ("bb", ["a","b","bbb","bbbb"], True),  # "b" + "b"
]

solver = Solution()
for s, wordDict, expected in tests:
    result = solver.wordBreak(s, wordDict)
    assert result == expected, f"Failed for s={s}, wordDict={wordDict}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n^2 * m) where n is length of s, m is average word length
  - Outer loop: O(n)
  - Inner loop: O(n)
  - Substring + hash lookup: O(m) for substring creation, O(1) average for set lookup
  - In practice with Python string hashing: closer to O(n^2)
- **Space**: O(n) for dp array + O(k*m) for word_set where k is number of words

## Edge Cases & Pitfalls
- **Empty string**: Should return true (base case)
- **Empty dictionary**: Return false unless s is empty
- **Word reuse**: Words can be used multiple times (e.g., "aaaa" with ["aa"])
- **Overlapping words**: Dictionary may have words that are prefixes of each other
- **Single character**: Handle correctly
- **No valid segmentation**: All dp values except dp[0] remain false
- **Substring creation**: Be careful with indices (s[j:i] not s[j:i-1])

## Follow-up Variants
- **Return all possible segmentations**: Use backtracking with DP (LC 140 - Word Break II)
- **Minimum number of words**: Find the segmentation with fewest words
- **With costs**: Each word has a cost, find minimum cost segmentation
- **Circular string**: String wraps around, find if any rotation can be segmented
- **Trie optimization**: Build a Trie from wordDict for faster prefix matching
- **Concatenated words**: Find words in dictionary that are concatenations of other words (LC 472)

## Takeaways
- Word Break is a classic DP problem demonstrating string partitioning.
- The key is recognizing that we can build solutions for longer strings from solutions for shorter prefixes.
- Using a set for wordDict is crucial for efficient lookup.
- The dp[0] = True base case represents the empty string, which is essential for the recurrence.
- This pattern of "can we partition/segment up to position i" appears in many string problems.
- Breaking early when dp[i] becomes true is a simple but effective optimization.

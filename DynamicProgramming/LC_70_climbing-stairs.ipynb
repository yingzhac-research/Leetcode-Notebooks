## Problem Statement 原题描述
You are climbing a staircase. It takes `n` steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

## Progressive Hints
- Hint 1: To reach step n, you must have been at step n-1 or n-2.
- Hint 2: The number of ways to reach step n equals the sum of ways to reach n-1 and n-2.
- Hint 3: This is the Fibonacci sequence! You can compute it iteratively with O(1) space.

## Solution Overview
This problem follows a simple recurrence relation: `dp[i] = dp[i-1] + dp[i-2]`, where `dp[i]` represents the number of ways to reach step `i`. Base cases are `dp[0] = 1` (one way to stay at ground) and `dp[1] = 1` (one way to reach first step). We can implement this using:

1. **Top-down DP (Memoization)**: Recursion with caching
2. **Bottom-up DP (Tabulation)**: Iterative with array
3. **Space-optimized**: Keep only last two values

## Detailed Explanation
**Approach: Space-Optimized Bottom-Up DP**

1. **Base Cases**: There is 1 way to reach step 0 (stay at ground) and 1 way to reach step 1 (take one step).

2. **Recurrence Relation**: For any step `i >= 2`, you can arrive from step `i-1` (by taking 1 step) or from step `i-2` (by taking 2 steps). Therefore:
   ```
   ways(i) = ways(i-1) + ways(i-2)
   ```

3. **Iteration**: Start from step 2 and compute up to step n. At each step, we only need the previous two values, so we can use two variables instead of an array.

4. **Result**: After iterating to step n, the answer is in our "current" variable.

**Example Walkthrough** (n = 5):
- Step 0: 1 way
- Step 1: 1 way
- Step 2: 1+1 = 2 ways
- Step 3: 2+1 = 3 ways
- Step 4: 3+2 = 5 ways
- Step 5: 5+3 = 8 ways

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Recursion (naive) | O(2^n) | O(n) | Exponential time, not practical |
| Top-down DP (memoization) | O(n) | O(n) | Recursion stack + memo array |
| Bottom-up DP (array) | O(n) | O(n) | Iterative with full dp array |
| Space-optimized | O(n) | O(1) | Optimal; only two variables needed |

````python
class Solution:
    def climbStairs(self, n: int) -> int:
        """
        Space-optimized DP solution.
        
        Time: O(n)
        Space: O(1)
        """
        if n <= 1:
            return 1
        
        # prev2 = dp[i-2], prev1 = dp[i-1]
        prev2, prev1 = 1, 1
        
        for i in range(2, n + 1):
            current = prev1 + prev2
            prev2 = prev1
            prev1 = current
        
        return prev1
````

````python
# Test cases
tests = [
    (2, 2),   # 1+1 or 2
    (3, 3),   # 1+1+1, 1+2, 2+1
    (1, 1),
    (4, 5),
    (5, 8),
    (10, 89)
]

solver = Solution()
for n, expected in tests:
    result = solver.climbStairs(n)
    assert result == expected, f"Failed for n={n}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n) - We iterate from 2 to n exactly once.
- **Space**: O(1) - We only use two variables to track previous states, regardless of n.

## Edge Cases & Pitfalls
- **n = 0 or 1**: Handle base cases explicitly. For n=0, conventionally return 1 (one way to stay at ground).
- **Integer overflow**: For very large n, the Fibonacci numbers grow exponentially. In Python, integers have arbitrary precision, but in languages like C++/Java, use modulo arithmetic if specified.
- **Off-by-one errors**: Ensure your loop runs from 2 to n inclusive.
- **Variable naming**: Keep track of which variable represents which state (prev1 vs prev2).

## Follow-up Variants
- **Variable step sizes**: What if you can climb 1, 2, or 3 steps at a time? Generalize to k step sizes.
- **Cost to climb**: Each step has a cost; find the minimum cost to reach the top (LC 746).
- **Matrix exponentiation**: For extremely large n, use matrix exponentiation to compute Fibonacci in O(log n) time.
- **Probabilistic variant**: If each step choice is random, what's the expected number of paths or time to reach the top?

## Takeaways
- Climbing Stairs is the canonical introduction to dynamic programming and demonstrates the Fibonacci recurrence.
- Always look for opportunities to optimize space when only a fixed number of previous states are needed.
- The pattern `dp[i] = dp[i-1] + dp[i-2]` appears in many counting problems.
- Understanding this problem deeply helps with more complex DP problems like House Robber and Decode Ways.

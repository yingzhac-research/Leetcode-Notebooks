## Problem Statement
Given an integer `n`, return the number of structurally unique **BST's** (binary search trees) which has exactly `n` nodes of unique values from `1` to `n`.

**Constraints**:
- `1 <= n <= 19`

## Progressive Hints
- Hint 1: Choose each number 1 to n as root, count unique BSTs for left and right subtrees.
- Hint 2: If root = i, left subtree has i-1 nodes, right has n-i nodes.
- Hint 3: Number of BSTs = product of left subtree count × right subtree count.
- Hint 4: Sum over all possible roots.
- Hint 5: This is the n-th Catalan number!
- Hint 6: Recurrence: `dp[n] = sum(dp[i-1] * dp[n-i] for i in 1 to n)`.
- Hint 7: Base case: `dp[0] = 1` (empty tree), `dp[1] = 1`.
- Hint 8: Catalan formula: `C(n) = (2n)! / ((n+1)! * n!)`.

## Solution Overview
Use **dynamic programming** based on **Catalan number** recurrence.

**Key Insight**: For n nodes, try each node as root:
- Root = i: left subtree has i-1 nodes, right has n-i nodes
- Number of unique BSTs = `dp[i-1] × dp[n-i]`
- Sum over all possible roots

**Recurrence**:
```python
dp[n] = sum(dp[i-1] * dp[n-i] for i in 1 to n)
```

**Base cases**: `dp[0] = 1`, `dp[1] = 1`

**This is the Catalan number**: `C(n) = dp[n]`

**Alternative**: Direct Catalan formula (but DP is clearer)

## Detailed Explanation

### BST Property

**Binary Search Tree**: For each node:
- All left subtree values < node value
- All right subtree values > node value

**Given values 1 to n**: BST structure is determined by value arrangement

---

### Recursive Structure

For n = 3, values {1, 2, 3}:

**Root = 1**:
- Left: empty (0 nodes) → 1 way
- Right: {2, 3} (2 nodes) → dp[2] ways
- Total: 1 × dp[2]

**Root = 2**:
- Left: {1} (1 node) → dp[1] ways
- Right: {3} (1 node) → dp[1] ways
- Total: dp[1] × dp[1]

**Root = 3**:
- Left: {1, 2} (2 nodes) → dp[2] ways
- Right: empty (0 nodes) → 1 way
- Total: dp[2] × 1

**Total for n=3**: `dp[2] + dp[1]×dp[1] + dp[2] = 2 + 1 + 2 = 5`

---

### Why dp[0] = 1?

**Empty tree**: One way to arrange 0 nodes (the empty arrangement)

**Multiplication identity**: `dp[1] = dp[0] × dp[0] = 1 × 1 = 1` ✓

---

### Catalan Number Connection

**Catalan numbers** count:
- Valid parentheses expressions
- Full binary trees with n+1 leaves
- Ways to triangulate a polygon
- **Unique BSTs with n nodes**

**Sequence**: 1, 1, 2, 5, 14, 42, 132, ...

**Formula**: `C(n) = (2n)! / ((n+1)! × n!) = C(2n, n) / (n+1)`

---

### DP vs Formula

**DP approach**:
- Time: O(n²)
- Space: O(n)
- Easy to understand

**Direct formula**:
- Time: O(n) with careful factorial computation
- Space: O(1)
- Risk of integer overflow

**For n <= 19**: DP is simpler and sufficient

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| DP | O(n²) | O(n) | Clear, handles all n |
| Catalan formula | O(n) | O(1) | Risk of overflow |
| Recursion + memo | O(n²) | O(n) | Top-down version |
| Brute force | O(n × C(n)) | O(n) | Generate all, exponential |

````python
class Solution:
    def numTrees(self, n: int) -> int:
        """
        Count unique BSTs using Catalan number DP.
        
        Time: O(n²)
        Space: O(n)
        """
        # dp[i] = number of unique BSTs with i nodes
        dp = [0] * (n + 1)
        
        # Base cases
        dp[0] = 1  # Empty tree
        dp[1] = 1  # Single node
        
        # Fill DP table
        for nodes in range(2, n + 1):
            # Try each value as root
            for root in range(1, nodes + 1):
                left_count = root - 1      # Nodes in left subtree
                right_count = nodes - root  # Nodes in right subtree
                
                # Multiply: left ways × right ways
                dp[nodes] += dp[left_count] * dp[right_count]
        
        return dp[n]
````

````python
# Test cases
tests = [
    (1, 1),
    (2, 2),
    (3, 5),
    (4, 14),
    (5, 42),
    (19, 1767263190),  # 19th Catalan number
]

solver = Solution()
for n, expected in tests:
    result = solver.numTrees(n)
    assert result == expected, f"Failed for n={n}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n²)
  - Outer loop: n iterations
  - Inner loop: n iterations per outer
  - Total: n × n
- **Space**: O(n) for DP array

## Edge Cases & Pitfalls
- **n = 0**: Should return 1 (empty tree)
- **n = 1**: One tree
- **Large n**: Within constraints (n <= 19), no overflow in Python
- **Common mistake**: Setting dp[0] = 0 breaks recurrence
- **Catalan sequence**: Verify with known values (1,1,2,5,14,...)

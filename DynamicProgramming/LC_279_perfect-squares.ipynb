## Problem Statement
Given an integer `n`, return the **least number of perfect square numbers** that sum to `n`.

A **perfect square** is an integer that is the square of an integer; in other words, it is the product of some integer with itself. For example, `1`, `4`, `9`, and `16` are perfect squares while `3` and `11` are not.

**Constraints**:
- 1 <= n <= 10^4

## Progressive Hints
- Hint 1: This is the coin change problem where "coins" are perfect squares: 1, 4, 9, 16, ...
- Hint 2: Use dp[i] to represent the minimum perfect squares needed to sum to i.
- Hint 3: For each number, try all perfect squares less than or equal to it.
- Hint 4: This is complete knapsack - each perfect square can be used unlimited times.
- Hint 5: Traverse from left to right to allow reusing the same square.

## Solution Overview
This is essentially **LC 322 Coin Change** with perfect squares as the "coins".

**Approaches**:
1. **Complete Knapsack DP**: Treat as coin change with coins = [1, 4, 9, 16, ...]
2. **BFS**: Treat as shortest path problem
3. **Mathematical** (Lagrange's Four Square Theorem): Every number can be represented as sum of at most 4 perfect squares

**State**: `dp[i]` = minimum perfect squares to sum to i

**Recurrence**: 
```python
dp[i] = min(dp[i], dp[i - j*j] + 1) for all j where j*j <= i
```

## Detailed Explanation

### Connection to Coin Change (LC 322)

This problem is **identical** to LC 322 Coin Change:
- **Coins**: Perfect squares [1, 4, 9, 16, 25, ...]
- **Amount**: n
- **Goal**: Minimum number of coins to make amount

**Only difference**: We generate the "coins" dynamically (all perfect squares ≤ n)

---

### Two Equivalent Approaches

**Approach 1: Iterate through perfect squares (like coin change)**
```python
squares = [i*i for i in range(1, int(n**0.5) + 1)]
for square in squares:
    for i in range(square, n + 1):
        dp[i] = min(dp[i], dp[i - square] + 1)
```
- Outer loop: perfect squares (items)
- Inner loop: numbers from square to n (capacity)
- Left to right: complete knapsack

**Approach 2: For each number, try all squares**
```python
for i in range(1, n + 1):
    for j in range(1, int(i**0.5) + 1):
        dp[i] = min(dp[i], dp[i - j*j] + 1)
```
- Outer loop: numbers 1 to n
- Inner loop: all perfect squares ≤ current number
- More intuitive for this problem

Both are **O(n × √n)** time complexity.

---

### Complete Knapsack Pattern Review

**Complete Knapsack Template**:
```python
for item in items:                    # Outer: items
    for capacity in range(item_weight, max_cap + 1):  # Inner: capacity (left to right)
        dp[capacity] = optimize(dp[capacity], dp[capacity - item_weight] + value)
```

**Applied to this problem**:
- Items: perfect squares [1, 4, 9, ...]
- Capacity: numbers from 1 to n
- Weight of each item: the square value itself
- Value: 1 (each square contributes 1 to the count)
- Optimize: minimize count

---

### Example Walkthrough

**Input**: n = 12

**Perfect squares ≤ 12**: 1, 4, 9

**Initial**: `dp = [0, inf, inf, ..., inf]` (size 13)

**Process**:
- dp[1] = min(inf, dp[0] + 1) = 1 (use 1)
- dp[2] = min(inf, dp[1] + 1) = 2 (use 1+1)
- dp[3] = min(inf, dp[2] + 1) = 3 (use 1+1+1)
- dp[4] = min(inf, dp[3] + 1, dp[0] + 1) = 1 (use 4)
- dp[5] = min(inf, dp[4] + 1, dp[1] + 1) = 2 (use 4+1)
- ...
- dp[9] = min(..., dp[0] + 1) = 1 (use 9)
- dp[12] = min(dp[11] + 1, dp[8] + 1, dp[3] + 1) = min(3, 3, 4) = 3 (use 4+4+4 or 1+1+1+9)

**Result**: dp[12] = 3

**One optimal decomposition**: 12 = 4 + 4 + 4

---

### Mathematical Insight: Lagrange's Four Square Theorem

**Theorem**: Every positive integer can be expressed as the sum of at most **four** perfect squares.

**Implications**:
- Answer is always between 1 and 4
- Can use this for early termination or validation

**Practical use**: In some implementations, you can check:
1. If n is perfect square: return 1
2. If n = a² + b² for some a, b: return 2
3. If n = a² + b² + c² for some a, b, c: return 3
4. Otherwise: return 4 (by Lagrange's theorem)

However, the DP approach is simpler and clearer.

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Backtracking | O(4^n) | O(√n) | Too slow |
| BFS | O(n × √n) | O(n) | Treats as shortest path |
| Complete Knapsack DP | O(n × √n) | O(n) | Most common, clear |
| Mathematical (Lagrange) | O(√n) | O(1) | Complex to implement correctly |

````python
class Solution:
    def numSquares(self, n: int) -> int:
        """
        Complete Knapsack DP solution.
        
        Time: O(n × √n)
        Space: O(n)
        """
        # dp[i] = minimum perfect squares to sum to i
        dp = [float('inf')] * (n + 1)
        dp[0] = 0  # Base case: 0 squares for sum 0
        
        # For each number from 1 to n
        for i in range(1, n + 1):
            # Try all perfect squares <= i
            j = 1
            while j * j <= i:
                square = j * j
                # If we can make (i - square), we can make i
                dp[i] = min(dp[i], dp[i - square] + 1)
                j += 1
        
        return dp[n]
````

````python
# Test cases
tests = [
    (12, 3),    # 4 + 4 + 4
    (13, 2),    # 4 + 9
    (1, 1),     # 1
    (2, 2),     # 1 + 1
    (3, 3),     # 1 + 1 + 1
    (4, 1),     # 4
    (5, 2),     # 1 + 4
    (7, 4),     # 1 + 1 + 1 + 4 (worst case by Lagrange)
    (100, 1),   # 10 * 10
    (43, 3),    # Can verify: not a square, not sum of 2 squares, so at least 3
]

solver = Solution()
for n, expected in tests:
    result = solver.numSquares(n)
    assert result == expected, f"Failed for n={n}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n × √n)
  - For each number from 1 to n (n iterations)
  - Try all perfect squares up to that number (√n iterations)
  - Total: n × √n
- **Space**: O(n)
  - DP array of size n + 1
  - No additional space needed

## Edge Cases & Pitfalls
- **n = 1**: Return 1 (perfect square itself)
- **n is perfect square**: Return 1 immediately (e.g., n = 4, 9, 16, ...)
- **Small numbers**: For n ≤ 3, answer is n (sum of n ones)
- **Lagrange's bound**: Answer is always ≤ 4, can use for validation
- **Initialization**: dp[0] = 0 is crucial, others should be infinity
- **Loop bounds**: Ensure j² ≤ i, not j ≤ i
- **Integer overflow**: Not an issue for this problem (n ≤ 10^4)
- **Optimization**: Can precompute all squares up to √n to avoid repeated computation

## Follow-up Variants
- **Print decomposition**: Return the actual perfect squares that sum to n, not just count
- **Generalized powers**: Minimum cubes, fourth powers, etc. to sum to n
- **Limited uses**: Each perfect square can be used at most k times
- **Weighted squares**: Different perfect squares have different costs
- **Range query**: Preprocess to answer queries for any n in O(1)
- **Mathematical approach**: Implement using Lagrange's Four Square Theorem directly

## Takeaways
- **Coin Change Pattern**: This is exactly coin change with perfect squares as coins
- **Complete Knapsack**: Each perfect square can be used unlimited times

- **Mathematical Connection**: Lagrange's theorem guarantees answer ≤ 4

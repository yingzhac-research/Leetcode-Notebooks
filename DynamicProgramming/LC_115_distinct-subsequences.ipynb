## Problem Statement 原题描述
Given two strings `s` and `t`, return the number of distinct subsequences of `s` which equals `t`.

A string's **subsequence** is a new string formed from the original string by deleting some (can be none) of the characters without disturbing the remaining characters' relative positions. (i.e., `"ACE"` is a subsequence of `"ABCDE"` while `"AEC"` is not).

The test cases are generated so that the answer fits on a 32-bit signed integer.

## Progressive Hints
- Hint 1: Use dp[i][j] to count distinct subsequences of s[0:i] that match t[0:j].
- Hint 2: If s[i-1] == t[j-1], we can either use this match or skip it: dp[i][j] = dp[i-1][j-1] + dp[i-1][j].
- Hint 3: If s[i-1] != t[j-1], we must skip s[i-1]: dp[i][j] = dp[i-1][j].
- Hint 4: Base case: dp[i][0] = 1 (empty t can be formed in one way), dp[0][j] = 0 for j > 0.
- Hint 5: Can optimize space to O(n) using 1D array with careful updates.

## Solution Overview
Use 2D DP to count the number of ways to form `t` from subsequences of `s`.

**State Definition**: `dp[i][j]` = number of distinct subsequences of `s[0:i]` that equal `t[0:j]`

**Recurrence**:
```python
if s[i-1] == t[j-1]:
    dp[i][j] = dp[i-1][j-1] + dp[i-1][j]
    # Use this character match + skip this character in s
else:
    dp[i][j] = dp[i-1][j]
    # Must skip this character in s
```

**Base cases**:
- `dp[i][0] = 1` (empty t can be formed by deleting all characters)
- `dp[0][j] = 0` for j > 0 (empty s cannot form non-empty t)

## Detailed Explanation
**Approach: 2D Dynamic Programming**

1. **Problem Understanding**:
   - We want to count ALL distinct ways to select characters from s to form t
   - Different indices in s count as different subsequences even if same character
   - Example: s="rabbbit", t="rabbit" has 3 ways (use different 'b's)

2. **State Definition**:
   - `dp[i][j]` = count of distinct subsequences of s[0:i] that equal t[0:j]
   - We're asking: "In how many ways can we form t[0:j] using characters from s[0:i]?"

3. **Transition Logic**:
   
   **Case 1**: Characters match (`s[i-1] == t[j-1]`)
   - **Option A**: Use this match
     - We matched s[i-1] with t[j-1]
     - Count how many ways we can form t[0:j-1] from s[0:i-1]
     - Contribution: `dp[i-1][j-1]`
   - **Option B**: Don't use this match (skip s[i-1])
     - Maybe there are other matches for t[j-1] earlier in s
     - Contribution: `dp[i-1][j]`
   - **Total**: `dp[i][j] = dp[i-1][j-1] + dp[i-1][j]`
   
   **Case 2**: Characters don't match (`s[i-1] != t[j-1]`)
   - Can't use s[i-1], must skip it
   - `dp[i][j] = dp[i-1][j]`

4. **Base Cases**:
   - `dp[0][0] = 1`: Empty s forms empty t in exactly 1 way
   - `dp[i][0] = 1` for all i: Any prefix of s can form empty t in 1 way (delete all)
   - `dp[0][j] = 0` for j > 0: Empty s cannot form non-empty t

**Example Walkthrough** (s="rabbbit", t="rabbit"):
```
DP Table:
        ""  r  a  b  b  i  t
""       1  0  0  0  0  0  0
r        1  1  0  0  0  0  0
a        1  1  1  0  0  0  0
b        1  1  1  1  0  0  0
b        1  1  1  2  1  0  0
b        1  1  1  3  3  0  0
i        1  1  1  3  3  3  0
t        1  1  1  3  3  3  3
```

**Key transitions**:
- dp[4][3]: s[3]='b' matches t[2]='b'
  - Use match: dp[3][2] = 1
  - Skip: dp[3][3] = 1
  - Total: 2
- dp[5][3]: s[4]='b' matches t[2]='b'
  - Use match: dp[4][2] = 1
  - Skip: dp[4][3] = 2
  - Total: 3

Result: 3 distinct subsequences

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| 2D DP | O(m×n) | O(m×n) | Standard approach |
| 1D DP | O(m×n) | O(n) | Space-optimized, careful update order |
| Recursion + memo | O(m×n) | O(m×n) | Top-down |
| Naive recursion | O(2^m) | O(m) | Exponential, too slow |

````python
class Solution:
    def numDistinct(self, s: str, t: str) -> int:
        """
        2D DP solution for counting distinct subsequences.
        
        Time: O(m×n)
        Space: O(m×n)
        """
        m, n = len(s), len(t)
        
        # dp[i][j] = count of ways to form t[0:j] from s[0:i]
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        
        # Base case: empty t can be formed in 1 way
        for i in range(m + 1):
            dp[i][0] = 1
        
        # Fill the table
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if s[i-1] == t[j-1]:
                    # Use this match + skip this character
                    dp[i][j] = dp[i-1][j-1] + dp[i-1][j]
                else:
                    # Must skip this character in s
                    dp[i][j] = dp[i-1][j]
        
        return dp[m][n]
````

````python
# Test cases
tests = [
    ("rabbbit", "rabbit", 3),
    ("babgbag", "bag", 5),
    ("", "", 1),              # Both empty
    ("abc", "", 1),           # Empty t
    ("", "a", 0),             # Empty s, non-empty t
    ("aaa", "aa", 3),         # Multiple same characters
    ("abc", "abc", 1),        # Exact match
    ("abc", "def", 0),        # No match
]

solver = Solution()
for s, t, expected in tests:
    result = solver.numDistinct(s, t)
    assert result == expected, f"Failed for s='{s}', t='{t}': got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(m×n) where m = len(s), n = len(t)
  - Fill m×n table, each cell in O(1)
- **Space**: O(m×n) for 2D table
  - Can optimize to O(n) with 1D array

## Edge Cases & Pitfalls
- **Empty strings**: Both empty → 1, s empty but t non-empty → 0, t empty → 1
- **t longer than s**: Impossible to form, return 0
- **Duplicate characters**: Must count each occurrence separately (key challenge)
- **Exact match**: s == t should return 1
- **No common characters**: Return 0
- **Integer overflow**: Problem states answer fits in 32-bit, but be aware for variants
- **Order of updates in 1D DP**: Must iterate j from right to left to avoid overwriting needed values

## Follow-up Variants
- **With wildcards**: Allow * in s or t to match any character
- **Distinct substrings**: Count contiguous substrings instead
- **Maximum instead of count**: Find longest distinct subsequence
- **Multiple target strings**: Count subsequences matching any of k target strings
- **Circular strings**: s is circular, can start matching from any position
- **Return actual subsequences**: Enumerate all distinct subsequences (exponential output)

## Takeaways
- Distinct Subsequences extends LCS from finding longest to counting all matches.
- The key insight: when characters match, both use and skip contribute to the count.
- Understanding when to ADD vs TAKE MAX is crucial (counting vs optimization).
- Base cases are important: empty t can be formed in 1 way, empty s cannot form non-empty t.
- This problem teaches the pattern of counting ways rather than finding optimal solution.
- The recurrence structure is similar to LCS but the aggregation is different (sum vs max).

## Problem Statement
You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed, the only constraint stopping you from robbing each of them is that adjacent houses have security systems connected and **it will automatically contact the police if two adjacent houses were broken into on the same night**.

Given an integer array `nums` representing the amount of money of each house, return the maximum amount of money you can rob tonight **without alerting the police**.

## Progressive Hints
- Hint 1: At each house i, you have two choices: rob it or skip it.
- Hint 2: If you rob house i, you cannot rob house i-1, so you add nums[i] to the max from i-2.
- Hint 3: If you skip house i, your max is the same as the max from i-1. Choose the better option.
- Hint 4: You only need to track the last two states, so use O(1) space.

## Solution Overview
Use dynamic programming with the recurrence: `dp[i] = max(dp[i-1], dp[i-2] + nums[i])`, where:
- `dp[i-1]`: skip current house, take the best from previous house
- `dp[i-2] + nums[i]`: rob current house, add to the best from two houses ago

Base cases: `dp[0] = nums[0]`, `dp[1] = max(nums[0], nums[1])`. We can optimize space to O(1) by keeping only the last two values.

## Detailed Explanation
**Approach: Space-Optimized DP**

1. **Decision at each house**: For house `i`, you either:
   - Rob it: gain `nums[i]` plus the maximum from houses up to `i-2`
   - Skip it: maintain the maximum from houses up to `i-1`

2. **Recurrence relation**:
   ```
   dp[i] = max(dp[i-1], dp[i-2] + nums[i])
   ```

3. **Base cases**:
   - 0 houses: 0 money
   - 1 house: rob that house
   - 2 houses: rob the one with more money

4. **Space optimization**: Since we only need the previous two values, use two variables `prev2` (i-2) and `prev1` (i-1) instead of an array.

**Example Walkthrough** ([2, 7, 9, 3, 1]):
- House 0: rob 2, total = 2
- House 1: skip 2, rob 7, total = 7
- House 2: rob 9 + 2 = 11 vs skip = 7, choose 11
- House 3: rob 3 + 7 = 10 vs skip = 11, choose 11
- House 4: rob 1 + 11 = 12 vs skip = 11, choose 12
- Result: 12

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Recursion (naive) | O(2^n) | O(n) | Explores all subsets, too slow |
| Top-down DP (memo) | O(n) | O(n) | Recursion + memoization |
| Bottom-up DP (array) | O(n) | O(n) | Iterative with full dp array |
| Space-optimized | O(n) | O(1) | Optimal; two variables only |

````python
from typing import List

class Solution:
    def rob(self, nums: List[int]) -> int:
        """
        Space-optimized DP solution.
        
        Time: O(n)
        Space: O(1)
        """
        if not nums:
            return 0
        if len(nums) == 1:
            return nums[0]
        
        # prev2 = dp[i-2], prev1 = dp[i-1]
        prev2 = nums[0]
        prev1 = max(nums[0], nums[1])
        
        for i in range(2, len(nums)):
            current = max(prev1, prev2 + nums[i])
            prev2 = prev1
            prev1 = current
        
        return prev1
````

````python
# Test cases
tests = [
    ([1,2,3,1], 4),           # Rob house 0 and 2: 1+3=4
    ([2,7,9,3,1], 12),        # Rob house 0, 2, 4: 2+9+1=12
    ([2,1,1,2], 4),           # Rob house 0 and 3: 2+2=4
    ([5], 5),                 # Single house
    ([1,2], 2),               # Two houses, pick larger
    ([5,3,4,11,2], 16),       # Rob 5, 11: 5+11=16
    ([2,1], 2)                # Two houses
]

solver = Solution()
for nums, expected in tests:
    result = solver.rob(nums)
    assert result == expected, f"Failed for {nums}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n) - Single pass through the array.
- **Space**: O(1) - Only two variables used regardless of input size.

## Edge Cases & Pitfalls
- **Empty array**: Return 0.
- **Single house**: Return that house's value.
- **Two houses**: Return the maximum of the two.
- **All same values**: Still need to follow the non-adjacent rule.
- **Negative values**: Problem guarantees non-negative integers, but if allowed, need to handle skipping all houses.
- **Variable initialization**: Ensure prev2 and prev1 are correctly initialized for the first two houses.

## Follow-up Variants

## Takeaways
- House Robber demonstrates the classic "choose or skip" DP pattern with a non-adjacent constraint.
- The recurrence `dp[i] = max(dp[i-1], dp[i-2] + nums[i])` is fundamental and appears in many optimization problems.
- Space optimization from O(n) to O(1) is straightforward when only a fixed window of previous states is needed.
- Understanding this problem helps with more complex variants involving trees and circular arrays.

## Problem Statement 原题描述
Given a string `s`, find the longest palindromic **subsequence**'s length in `s`.

A **subsequence** is a sequence that can be derived from another sequence by deleting some or no elements without changing the order of the remaining elements.

**Constraints**:
- 1 <= s.length <= 1000
- s consists only of lowercase English letters

## Progressive Hints
- Hint 1: Unlike substring (contiguous), subsequence allows gaps. Think interval DP.
- Hint 2: Use dp[i][j] to represent the longest palindromic subsequence in s[i:j+1].
- Hint 3: If s[i] == s[j], include both: dp[i][j] = dp[i+1][j-1] + 2.
- Hint 4: If s[i] != s[j], take max: dp[i][j] = max(dp[i+1][j], dp[i][j-1]).
- Hint 5: Base case: single character has length 1.

## Solution Overview
This is a classic **Interval DP** problem.

**Key Difference from LC 5**:
- **LC 5 (Longest Palindromic Substring)**: Substring is **contiguous**
- **LC 516 (This problem)**: Subsequence is **non-contiguous** (can skip characters)

**State**: `dp[i][j]` = length of longest palindromic subsequence in s[i:j+1]

**Recurrence**:
```python
if s[i] == s[j]:
    dp[i][j] = dp[i+1][j-1] + 2  # Include both ends
else:
    dp[i][j] = max(dp[i+1][j], dp[i][j-1])  # Skip one end
```

**Alternative View**: This is equivalent to **LCS(s, reverse(s))**.

## Detailed Explanation

### Understanding Subsequence vs Substring

**Substring** (LC 5):
- Must be contiguous
- Example: "abc" → substrings: "a", "b", "c", "ab", "bc", "abc"

**Subsequence** (this problem):
- Can skip characters, but maintain order
- Example: "abc" → subsequences: "", "a", "b", "c", "ab", "ac", "bc", "abc"

---

### Interval DP Approach

**State Definition**:
- `dp[i][j]` = length of longest palindromic subsequence in s[i:j+1]

**Base Cases**:
- `dp[i][i] = 1` (single character is palindrome of length 1)
- `dp[i][i-1] = 0` (empty string, used for even-length palindromes)

**Recurrence Logic**:

**Case 1**: s[i] == s[j]
- We can include both characters in the palindrome
- Longest palindrome = 2 + longest palindrome in s[i+1:j]
- `dp[i][j] = dp[i+1][j-1] + 2`

**Case 2**: s[i] != s[j]
- We can't include both characters together
- Try skipping left character: dp[i+1][j]
- Try skipping right character: dp[i][j-1]
- Take maximum: `dp[i][j] = max(dp[i+1][j], dp[i][j-1])`

---

### Iteration Order

**Critical**: Must process smaller intervals before larger ones.

**Method 1 - By Length**:
```python
for length in range(2, n + 1):
    for i in range(n - length + 1):
        j = i + length - 1
        # Compute dp[i][j]
```

**Method 2 - Reverse i, Forward j**:
```python
for i in range(n - 1, -1, -1):  # Reverse
    for j in range(i + 1, n):   # Forward
        # Compute dp[i][j]
```

Both ensure dp[i+1][j-1], dp[i+1][j], dp[i][j-1] are computed before dp[i][j].

---

### Example Walkthrough

**Input**: s = "bbbab"

**DP Table** (dp[i][j]):
```
     0   1   2   3   4
     b   b   b   a   b
0 b  1   2   3   3   4
1 b      1   2   2   3
2 b          1   1   3
3 a              1   1
4 b                  1
```

**Process**:
- Length 1: All 1 (single characters)
- Length 2:
  - dp[0][1]: s[0]='b', s[1]='b' → match → dp[1][0] + 2 = 0 + 2 = 2
  - dp[1][2]: s[1]='b', s[2]='b' → match → 2
  - dp[2][3]: s[2]='b', s[3]='a' → no match → max(dp[3][3], dp[2][2]) = 1
  - dp[3][4]: s[3]='a', s[4]='b' → no match → 1

- Length 3:
  - dp[0][2]: s[0]='b', s[2]='b' → match → dp[1][1] + 2 = 1 + 2 = 3
  - dp[1][3]: s[1]='b', s[3]='a' → no match → max(dp[2][3], dp[1][2]) = max(1, 2) = 2
  - dp[2][4]: s[2]='b', s[4]='b' → match → dp[3][3] + 2 = 1 + 2 = 3

- Length 4:
  - dp[0][3]: s[0]='b', s[3]='a' → no match → max(dp[1][3], dp[0][2]) = max(2, 3) = 3
  - dp[1][4]: s[1]='b', s[4]='b' → match → dp[2][3] + 2 = 1 + 2 = 3

- Length 5:
  - dp[0][4]: s[0]='b', s[4]='b' → match → dp[1][3] + 2 = 2 + 2 = 4

**Result**: dp[0][4] = 4

**Actual subsequence**: "bbbb" (skip the 'a')

---

### Alternative: LCS with Reverse

**Insight**: Longest palindromic subsequence = LCS(s, reverse(s))

**Example**: s = "bbbab"
- reverse(s) = "babbb"
- LCS("bbbab", "babbb") = "bbbb" (length 4) ✓

**Why this works**:
- A palindrome reads same forwards and backwards
- Characters in palindromic subsequence appear in both s and reverse(s)
- LCS finds the longest such common sequence

**Implementation**:
```python
def longestPalindromeSubseq(s):
    return lcs(s, s[::-1])
```

Both approaches are O(n²) time and space.

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Brute force | O(2^n) | O(n) | Try all subsequences |
| Interval DP | O(n²) | O(n²) | 2D DP table |
| Interval DP (space optimized) | O(n²) | O(n) | Use rolling array |
| LCS approach | O(n²) | O(n²) | LCS(s, reverse(s)) |

````python
class Solution:
    def longestPalindromeSubseq(self, s: str) -> int:
        """
        Interval DP solution.
        
        Time: O(n²)
        Space: O(n²)
        """
        n = len(s)
        # dp[i][j] = length of longest palindromic subsequence in s[i:j+1]
        dp = [[0] * n for _ in range(n)]
        
        # Base case: single characters
        for i in range(n):
            dp[i][i] = 1
        
        # Build from smaller to larger intervals
        for length in range(2, n + 1):
            for i in range(n - length + 1):
                j = i + length - 1
                
                if s[i] == s[j]:
                    # Include both ends
                    if length == 2:
                        dp[i][j] = 2
                    else:
                        dp[i][j] = dp[i + 1][j - 1] + 2
                else:
                    # Skip one end, take maximum
                    dp[i][j] = max(dp[i + 1][j], dp[i][j - 1])
        
        return dp[0][n - 1]
````

````python
# Test cases
tests = [
    ("bbbab", 4),      # "bbbb"
    ("cbbd", 2),       # "bb"
    ("a", 1),          # "a"
    ("ac", 1),         # "a" or "c"
    ("abcde", 1),      # Any single char
    ("abcdcba", 7),    # Entire string
    ("abacabad", 5),   # "ababa" or "bacab"
    ("aaa", 3),        # All same
]

solver = Solution()
for s, expected in tests:
    result = solver.longestPalindromeSubseq(s)
    assert result == expected, f"Failed for s='{s}': got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n²) where n = len(s)
  - Fill n² entries in DP table
  - Each entry computed in O(1)
- **Space**: O(n²)
  - 2D DP table
  - Can optimize to O(n) using rolling array

## Edge Cases & Pitfalls
- **Single character**: Return 1
- **All same characters**: Return n (entire string)
- **No palindrome > 1**: Return 1 (any single character)
- **Entire string is palindrome**: Return n
- **Iteration order**: Must process smaller intervals first (length or reverse i)
- **Base case**: dp[i][i] = 1 for all i
- **Index bounds**: When length=2, dp[i+1][j-1] is out of bounds, handle separately
- **Comparison with LC 5**: This allows gaps (subsequence), LC 5 doesn't (substring)

## Takeaways

## Problem Statement 原题描述
You are given an array of binary strings `strs` and two integers `m` and `n`.

Return the size of the largest subset of `strs` such that there are **at most** `m` `0`'s and `n` `1`'s in the subset.

A set `x` is a **subset** of a set `y` if all elements of `x` are also elements of `y`.

**Constraints**:
- 1 <= strs.length <= 600
- 1 <= strs[i].length <= 100
- strs[i] consists only of digits '0' and '1'
- 1 <= m, n <= 100

## Progressive Hints
- Hint 1: This is a knapsack problem with two dimensions (capacity for 0s and capacity for 1s).
- Hint 2: For each string, count its 0s and 1s. This represents the "weight" in both dimensions.
- Hint 3: Use dp[i][j] = maximum size of subset with at most i zeros and j ones.
- Hint 4: For each string, decide whether to include it (if we have enough capacity) or skip it.
- Hint 5: Traverse the DP table from right to left in both dimensions to ensure 0/1 knapsack property.

## Solution Overview
This is a **2D 0/1 Knapsack** problem:
- **Items**: binary strings
- **Two capacities**: m (max zeros), n (max ones)
- **Item weight**: (count of 0s, count of 1s) for each string
- **Item value**: 1 (each string contributes 1 to subset size)
- **Goal**: Maximize the number of strings we can include

**Key Insight**: Extend 1D knapsack to 2D by tracking both resource constraints simultaneously.

**State**: `dp[i][j]` = max subset size using at most i zeros and j ones

**Recurrence**: For each string with zeros0 zeros and ones1 ones:
```python
dp[i][j] = max(dp[i][j], dp[i-zeros0][j-ones1] + 1)
```

## Detailed Explanation

### Understanding 2D Knapsack

**Standard 0/1 Knapsack**:
- One capacity constraint (weight limit)
- DP state: dp[w] = max value with weight limit w

**2D 0/1 Knapsack** (this problem):
- Two capacity constraints (m zeros, n ones)
- DP state: dp[i][j] = max subset size with at most i zeros and j ones
- Each item (string) has two "weights": count of 0s and count of 1s
- Each item has value 1 (contributes 1 to subset size)

---

### Algorithm Steps

**Step 1**: Initialize 2D DP table
- `dp[i][j] = 0` for all i, j
- dp[0][0] = 0 (with 0 capacity, we can select 0 strings)

**Step 2**: For each string in strs:
- Count zeros and ones in the string
- For each capacity (i, j) from (m, n) down to (zeros, ones):
  - If we have enough capacity: `dp[i][j] = max(dp[i][j], dp[i-zeros][j-ones] + 1)`

**Step 3**: Return dp[m][n]

---

### Why Traverse Right to Left in BOTH Dimensions?

**Reason**: To ensure each string is used at most once.

**Wrong (left to right)**:
```python
for i in range(zeros, m + 1):
    for j in range(ones, n + 1):
        dp[i][j] = max(dp[i][j], dp[i-zeros][j-ones] + 1)
```
- When updating dp[i][j], we might use dp[i-zeros][j-ones] that was **already updated** in this iteration
- This allows using the same string multiple times!

**Correct (right to left)**:
```python
for i in range(m, zeros - 1, -1):
    for j in range(n, ones - 1, -1):
        dp[i][j] = max(dp[i][j], dp[i-zeros][j-ones] + 1)
```
- When updating dp[i][j], we use dp[i-zeros][j-ones] from the **previous iteration**
- Ensures each string is used at most once

---

### Example Walkthrough

**Input**: strs = ["10", "0001", "111001", "1", "0"], m = 5, n = 3

**Count zeros and ones**:
- "10": 1 zero, 1 one
- "0001": 3 zeros, 1 one
- "111001": 2 zeros, 4 ones (skip: needs 4 ones > n=3)
- "1": 0 zeros, 1 one
- "0": 1 zero, 0 ones

**DP Process** (simplified):
- Initial: dp[5][3] = 0
- After "10" (1,1): dp[5][3] can include this, dp[5][3] = max(0, dp[4][2] + 1) = 1
- After "0001" (3,1): dp[5][3] = max(1, dp[2][2] + 1) = ...
- Continue for all strings...

**Optimal subset**: ["10", "0001", "1", "0"] = 4 strings
- Total: 5 zeros (1+3+0+1), 3 ones (1+1+1+0) ✓

---

### Comparison: 1D vs 2D Knapsack

**1D Knapsack** (e.g., LC 416):
```python
for item in items:
    for capacity in range(max_cap, item_weight - 1, -1):
        dp[capacity] = max(dp[capacity], dp[capacity - item_weight] + item_value)
```

**2D Knapsack** (this problem):
```python
for item in items:
    for cap1 in range(max_cap1, item_weight1 - 1, -1):
        for cap2 in range(max_cap2, item_weight2 - 1, -1):
            dp[cap1][cap2] = max(dp[cap1][cap2], 
                                 dp[cap1 - item_weight1][cap2 - item_weight2] + 1)
```

**Pattern**: Both traverse capacities right to left to ensure 0/1 property.

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Backtracking | O(2^len(strs)) | O(len(strs)) | Try all subsets |
| 3D DP | O(len(strs) × m × n) | O(len(strs) × m × n) | dp[k][i][j] for first k strings |
| 2D DP | O(len(strs) × m × n) | O(m × n) | Space-optimized, most common |
| DFS + memo | O(len(strs) × m × n) | O(len(strs) × m × n) | Top-down approach |

````python
class Solution:
    def findMaxForm(self, strs: list[str], m: int, n: int) -> int:
        """
        2D 0/1 Knapsack solution.
        
        Time: O(len(strs) × m × n)
        Space: O(m × n)
        """
        # dp[i][j] = max subset size with at most i zeros and j ones
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        
        for s in strs:
            # Count zeros and ones in current string
            zeros = s.count('0')
            ones = s.count('1')
            
            # Traverse right to left in both dimensions (0/1 knapsack)
            for i in range(m, zeros - 1, -1):
                for j in range(n, ones - 1, -1):
                    # Choose max: skip this string OR include it
                    dp[i][j] = max(dp[i][j], dp[i - zeros][j - ones] + 1)
        
        return dp[m][n]
````

````python
# Test cases
tests = [
    (["10", "0001", "111001", "1", "0"], 5, 3, 4),  # ["10","0001","1","0"]
    (["10", "0", "1"], 1, 1, 2),                    # ["0","1"]
    (["0", "0", "1", "1"], 2, 2, 4),                # All strings
    (["10"], 1, 1, 1),                              # Single string
    (["00", "11"], 2, 2, 1),                        # Can only pick one
    (["111"], 0, 3, 1),                             # Exactly fits
    (["111"], 0, 2, 0),                             # Doesn't fit
]

solver = Solution()
for strs, m, n, expected in tests:
    result = solver.findMaxForm(strs, m, n)
    assert result == expected, f"Failed for strs={strs}, m={m}, n={n}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(L × m × n) where L = len(strs)
  - For each of L strings, we iterate through m × n states
  - Counting 0s and 1s in each string is O(string length), but typically small
  - Total: O(L × m × n)
- **Space**: O(m × n)
  - 2D DP table of size (m+1) × (n+1)
  - Can be further optimized using rolling array if needed

## Edge Cases & Pitfalls
- **Empty strings**: Should not occur based on constraints (length >= 1)
- **Strings that don't fit**: If a string has more 0s than m or more 1s than n, skip it automatically
- **m=0 or n=0**: Can only select strings with all 1s (if m=0) or all 0s (if n=0)
- **All strings identical**: DP will correctly handle by selecting as many as capacity allows
- **Traversal direction**: MUST go right to left in BOTH dimensions for 0/1 property
- **Off-by-one**: Ensure loop ranges are `range(m, zeros-1, -1)` not `range(m, zeros, -1)`
- **Large constraints**: With m=n=100 and L=600, DP table is manageable (100×100×600 iterations)

## Follow-up Variants
- **K-dimensional knapsack**: Extend to k different resource types (memory, CPU, disk, etc.)
- **Minimum subset size**: Find minimum subset that uses at least m zeros and n ones
- **Weighted values**: Each string has different value, maximize total value instead of count
- **Exact match**: Find subset with exactly m zeros and n ones
- **Online version**: Strings arrive in stream, maintain optimal subset dynamically
- **Resource scheduling**: Schedule tasks with multiple resource constraints (CPU, memory, GPU)

## Takeaways
- **2D Knapsack Pattern**: Extends 1D knapsack to multiple capacity dimensions
- **Right-to-Left Rule**: Traverse ALL capacity dimensions right to left for 0/1 knapsack
- **State Definition**: dp[i][j] represents max value with capacity constraints (i, j)
- **Resource Allocation**: Models real-world problems with multiple resource constraints
- **Template Scalability**: Can extend to 3D, 4D, or K-dimensional knapsack with same pattern
- **Space Optimization**: Can use 2D array instead of 3D by processing strings sequentially

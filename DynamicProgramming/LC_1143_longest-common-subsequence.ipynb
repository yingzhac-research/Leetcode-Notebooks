## Problem Statement
Given two strings `text1` and `text2`, return the length of their longest common subsequence. If there is no common subsequence, return `0`.

A **subsequence** of a string is a new string generated from the original string with some characters (can be none) deleted without changing the relative order of the remaining characters.

- For example, `"ace"` is a subsequence of `"abcde"`.

A **common subsequence** of two strings is a subsequence that is common to both strings.

## Progressive Hints
- Hint 1: Use 2D DP where dp[i][j] represents LCS length of text1[0:i] and text2[0:j].
- Hint 2: If characters match, dp[i][j] = dp[i-1][j-1] + 1.
- Hint 3: If characters don't match, take max of excluding either character: dp[i][j] = max(dp[i-1][j], dp[i][j-1]).
- Hint 4: Base cases: dp[0][j] = 0 and dp[i][0] = 0 (empty string has no common subsequence).
- Hint 5: Can optimize space to O(min(m, n)) using 1D array.

## Solution Overview
Use 2D dynamic programming similar to Edit Distance.

**State Definition**: `dp[i][j]` = length of LCS between `text1[0:i]` and `text2[0:j]`

**Recurrence**:
```python
if text1[i-1] == text2[j-1]:
    dp[i][j] = dp[i-1][j-1] + 1  # Extend LCS by 1
else:
    dp[i][j] = max(dp[i-1][j],   # Exclude text1[i-1]
                   dp[i][j-1])   # Exclude text2[j-1]
```

**Base cases**: `dp[0][j] = 0` and `dp[i][0] = 0` (empty prefix has LCS of 0)

## Detailed Explanation
**Approach: 2D Dynamic Programming**

1. **Understanding Subsequences**:
   - Unlike substrings, subsequences don't need to be contiguous
   - We can skip characters but must maintain order
   - Example: "ace" is a subsequence of "abcde" (skip b, d)

2. **State Definition**:
   - `dp[i][j]` = LCS length for text1[0:i] and text2[0:j]
   - Array size: (m+1) × (n+1) to handle empty prefixes

3. **Transition Logic**:
   
   **Case 1**: Characters match (`text1[i-1] == text2[j-1]`)
   - This character is part of the LCS
   - Add 1 to the LCS of the previous prefixes
   - `dp[i][j] = dp[i-1][j-1] + 1`
   
   **Case 2**: Characters don't match
   - We must skip one of them
   - Try skipping text1[i-1]: use `dp[i-1][j]`
   - Try skipping text2[j-1]: use `dp[i][j-1]`
   - Take the maximum: `dp[i][j] = max(dp[i-1][j], dp[i][j-1])`

4. **Algorithm**:
   ```python
   m, n = len(text1), len(text2)
   dp = [[0] * (n + 1) for _ in range(m + 1)]
   
   for i in range(1, m + 1):
       for j in range(1, n + 1):
           if text1[i-1] == text2[j-1]:
               dp[i][j] = dp[i-1][j-1] + 1
           else:
               dp[i][j] = max(dp[i-1][j], dp[i][j-1])
   
   return dp[m][n]
   ```

**Example Walkthrough** (text1="abcde", text2="ace"):
```
DP Table:
      ""  a  c  e
""     0  0  0  0
a      0  1  1  1
b      0  1  1  1
c      0  1  2  2
d      0  1  2  2
e      0  1  2  3
```

LCS = "ace" (length 3)

**Key transitions**:
- dp[1][1]: 'a' == 'a' → dp[0][0] + 1 = 1
- dp[3][2]: 'c' == 'c' → dp[2][1] + 1 = 2
- dp[5][3]: 'e' == 'e' → dp[4][2] + 1 = 3

**Reconstructing the LCS** (optional):
```python
# Backtrack from dp[m][n]
i, j = m, n
lcs = []
while i > 0 and j > 0:
    if text1[i-1] == text2[j-1]:
        lcs.append(text1[i-1])
        i -= 1
        j -= 1
    elif dp[i-1][j] > dp[i][j-1]:
        i -= 1
    else:
        j -= 1
return ''.join(reversed(lcs))
```

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| 2D DP | O(m×n) | O(m×n) | Standard approach |
| 1D DP | O(m×n) | O(min(m,n)) | Space-optimized |
| Recursion + memo | O(m×n) | O(m×n) | Top-down, clearer |
| Naive recursion | O(2^(m+n)) | O(m+n) | Too slow, exponential |

````python
class Solution:
    def longestCommonSubsequence(self, text1: str, text2: str) -> int:
        """
        2D DP solution for LCS.
        
        Time: O(m×n)
        Space: O(m×n)
        """
        m, n = len(text1), len(text2)
        
        # Create DP table
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        
        # Fill the table
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if text1[i-1] == text2[j-1]:
                    # Characters match, extend LCS
                    dp[i][j] = dp[i-1][j-1] + 1
                else:
                    # Characters don't match, take max of excluding either
                    dp[i][j] = max(dp[i-1][j], dp[i][j-1])
        
        return dp[m][n]
````

````python
# Test cases
tests = [
    ("abcde", "ace", 3),        # "ace"
    ("abc", "abc", 3),          # Identical strings
    ("abc", "def", 0),          # No common subsequence
    ("", "abc", 0),             # Empty string
    ("abc", "", 0),             # Empty string
    ("ABCDGH", "AEDFHR", 3),   # "ADH"
    ("AGGTAB", "GXTXAYB", 4),  # "GTAB"
]

solver = Solution()
for text1, text2, expected in tests:
    result = solver.longestCommonSubsequence(text1, text2)
    assert result == expected, f"Failed for ('{text1}', '{text2}'): got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(m×n) where m = len(text1), n = len(text2)
  - Fill m×n table, each cell in O(1)
- **Space**: O(m×n) for 2D table
  - Can optimize to O(min(m,n)) using 1D array

## Edge Cases & Pitfalls
- **Empty strings**: One or both strings empty → return 0
- **Identical strings**: LCS length equals string length
- **No common characters**: Return 0
- **Single character**: If they match, return 1; otherwise 0
- **Subsequence vs substring**: Remember subsequence can skip characters
- **Index confusion**: dp[i][j] corresponds to text1[i-1] and text2[j-1]
- **Reconstruction**: If you need the actual LCS string, need to backtrack through the table

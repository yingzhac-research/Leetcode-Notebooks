## Problem Statement 原题描述
You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed. All houses at this place are **arranged in a circle**. That means the first house is the neighbor of the last one. Meanwhile, adjacent houses have a security system connected, and **it will automatically contact the police if two adjacent houses were broken into on the same night**.

Given an integer array `nums` representing the amount of money of each house, return the maximum amount of money you can rob tonight **without alerting the police**.

## Progressive Hints
- Hint 1: The circular constraint means you cannot rob both the first and last house.
- Hint 2: Consider two scenarios: (1) rob houses 0 to n-2 (exclude last), (2) rob houses 1 to n-1 (exclude first).
- Hint 3: Run the House Robber I algorithm on each scenario and take the maximum.
- Hint 4: Handle the edge case where there's only one house.

## Solution Overview
The circular arrangement creates a dependency: if you rob the first house, you cannot rob the last house, and vice versa. We break this into two linear subproblems:

1. **Case 1**: Rob houses from index 0 to n-2 (exclude the last house)
2. **Case 2**: Rob houses from index 1 to n-1 (exclude the first house)

Run the standard House Robber algorithm on each case and return the maximum. Special handling for arrays with 1 or 2 houses.

## Detailed Explanation
**Approach: Two Linear DP Problems**

1. **Circular constraint analysis**:
   - If we rob house 0, we cannot rob house n-1
   - If we rob house n-1, we cannot rob house 0
   - If we rob neither, we're solving a subset of the linear problem

2. **Key insight**: The optimal solution must fall into one of these categories:
   - Includes house 0 but not house n-1: solve for [0, n-2]
   - Includes house n-1 but not house 0: solve for [1, n-1]
   - We don't need to explicitly handle "neither" because it's subsumed by the above cases

3. **Algorithm**:
   - Helper function: `rob_linear(start, end)` - standard House Robber for a range
   - Return `max(rob_linear(0, n-2), rob_linear(1, n-1))`

4. **Edge cases**:
   - n = 1: return `nums[0]`
   - n = 2: return `max(nums[0], nums[1])`

**Example Walkthrough** ([2, 3, 2]):
- Case 1: [2, 3] → max = 3
- Case 2: [3, 2] → max = 3
- Result: max(3, 3) = 3

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Two linear DP passes | O(n) | O(1) | Run House Robber I twice |
| Single DP with tracking | O(n) | O(1) | More complex state management |
| Recursion with memo | O(n) | O(n) | Clearer logic but uses stack space |

```python
from typing import List

class Solution:
    def rob(self, nums: List[int]) -> int:
        """
        Circular House Robber using two linear passes.
        
        Time: O(n)
        Space: O(1)
        """
        n = len(nums)
        if n == 1:
            return nums[0]
        if n == 2:
            return max(nums[0], nums[1])
        
        # Rob houses [0, n-2] or [1, n-1]
        return max(self.rob_linear(nums, 0, n - 2),
                   self.rob_linear(nums, 1, n - 1))
    
    def rob_linear(self, nums: List[int], start: int, end: int) -> int:
        """
        Standard House Robber for a linear range [start, end].
        """
        prev2, prev1 = 0, 0
        
        for i in range(start, end + 1):
            current = max(prev1, prev2 + nums[i])
            prev2 = prev1
            prev1 = current
        
        return prev1
```

```python
# Test cases
tests = [
    ([2,3,2], 3),              # Rob house 1: 3
    ([1,2,3,1], 4),            # Rob house 0 and 2: 1+3=4
    ([1,2,3], 3),              # Rob house 2: 3
    ([1], 1),                  # Single house
    ([1,2], 2),                # Two houses
    ([2,1,1,2], 3),            # Rob house 0 or 3: max=2, but [1,1,2] gives 3
    ([1,3,1,3,100], 103),      # Rob 3 and 100
]

solver = Solution()
for nums, expected in tests:
    result = solver.rob(nums)
    assert result == expected, f"Failed for {nums}: got {result}, expected {expected}"
print('All tests passed.')
```

## Complexity Analysis
- **Time**: O(n) - We run the linear House Robber algorithm twice, each taking O(n).
- **Space**: O(1) - Only constant extra space for variables.

## Edge Cases & Pitfalls
- **Single house**: Must return that house's value (no circular constraint).
- **Two houses**: Can only rob one, return the maximum.
- **Off-by-one errors**: Ensure ranges are [0, n-2] and [1, n-1], not [0, n-1] and [1, n].
- **Empty array**: Problem guarantees at least one house, but handle gracefully if needed.
- **All same values**: The circular constraint still applies; cannot rob all houses.

## Follow-up Variants
- **k-circular constraint**: First and last k houses are mutually exclusive.
- **Multiple circles**: Houses arranged in multiple concentric circles with cross-circle constraints.
- **Binary tree circular**: Generalize to a tree where leaf nodes form a circular dependency.
- **Weighted constraints**: Different penalties for robbing adjacent vs circular-adjacent houses.

## Takeaways
- House Robber II demonstrates how to break circular constraints into linear subproblems.
- The key pattern: when first and last are mutually exclusive, solve twice with different ranges and take the maximum.
- Reusing the linear DP solution shows the power of modular, composable algorithms.
- This technique generalizes to other circular dependency problems.

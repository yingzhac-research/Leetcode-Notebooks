## Problem Statement
Given the `root` of a binary tree, return the **length of the diameter** of the tree.

The **diameter** of a binary tree is the **length** of the longest path between any two nodes in a tree. This path may or may not pass through the `root`.

The **length** of a path between two nodes is represented by the number of edges between them.

**Constraints**:
- The number of nodes in the tree is in the range `[1, 10^4]`.
- `-100 <= Node.val <= 100`

## Progressive Hints
- Hint 1: The diameter at any node is the sum of heights of its left and right subtrees.
- Hint 2: The global diameter might not pass through the root; it could be in any subtree.
- Hint 3: Use DFS to compute height of each subtree while tracking the maximum diameter seen.
- Hint 4: For each node, calculate left_height + right_height and update global maximum.
- Hint 5: Return height to parent (not diameter) to enable parent's calculation.
- Hint 6: Height of a node is 1 + max(left_height, right_height).

## Solution Overview
Use **Tree DP** with DFS to compute heights while maintaining global diameter.

**Key Insight**: At each node:
- **Diameter through this node** = left_height + right_height
- **Height of this node** = 1 + max(left_height, right_height)

**Algorithm**:
```python
def dfs(node):
    if not node: return 0
    
    left_height = dfs(node.left)
    right_height = dfs(node.right)
    
    # Update global diameter
    diameter = max(diameter, left_height + right_height)
    
    # Return height to parent
    return 1 + max(left_height, right_height)
```

**Why it works**: We explore all nodes, and at each node we consider the longest path passing through it.

## Detailed Explanation

### Understanding the Problem

**Diameter** = longest path between any two nodes (measured in number of edges)

Key observations:
1. The path **may or may not** pass through root
2. The path connects two leaf nodes (or internal nodes)
3. For any node, the longest path through it = path down left + path down right

---

### Relationship Between Diameter and Height

For any node `n`:
- **Height** = longest path from `n` to any leaf in its subtree
- **Diameter through n** = longest path from left subtree to right subtree passing through `n`
  - This equals `left_height + right_height`

**Crucial insight**: Diameter through node = left_height + right_height (number of edges)

---

### Algorithm Design

**Goal**: Find maximum diameter across all nodes

**Approach**:
1. Use DFS to traverse tree (post-order)
2. For each node, compute:
   - left_height from left subtree
   - right_height from right subtree
   - diameter_through_node = left_height + right_height
3. Update global maximum diameter
4. Return height to parent: 1 + max(left_height, right_height)

**Why return height, not diameter?**
- Parent needs height to calculate its own diameter
- Diameter is maintained as a global variable

---

### Example Walkthrough

Tree:
```
      1
     / \\
    2   3
   / \\
  4   5
```

**Bottom-up computation**:

Node 4 (leaf):
- left_height = 0, right_height = 0
- diameter_through_4 = 0 + 0 = 0
- return height = 1

Node 5 (leaf):
- left_height = 0, right_height = 0
- diameter_through_5 = 0 + 0 = 0
- return height = 1

Node 2:
- left_height = 1 (from node 4)
- right_height = 1 (from node 5)
- diameter_through_2 = 1 + 1 = **2** ← Update global max
- return height = 1 + max(1, 1) = 2

Node 3 (leaf):
- left_height = 0, right_height = 0
- diameter_through_3 = 0 + 0 = 0
- return height = 1

Node 1 (root):
- left_height = 2 (from node 2)
- right_height = 1 (from node 3)
- diameter_through_1 = 2 + 1 = **3** ← Update global max
- return height = 1 + max(2, 1) = 3

**Answer**: max diameter = **3** (path: 4 → 2 → 1 → 3 or 5 → 2 → 1 → 3)

---

### Why Post-Order DFS?

We need children's heights before computing:
1. Current node's diameter (left_height + right_height)
2. Current node's height (1 + max(left_height, right_height))

This naturally requires post-order traversal: children first, then parent.

---

### Edge vs. Node Count

**Important**: Problem asks for number of **edges**, not nodes.

- Height in edges = number of edges from node to leaf
- For leaf node: height = 0 (no edges below)
- For null node: height = 0
- For internal node: height = 1 + max(left, right)

The +1 represents the edge connecting current node to its child.

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| DFS with global var | O(n) | O(h) | Optimal, clean code |
| DFS returning pair | O(n) | O(h) | Returns (height, diameter) |
| Two-pass DFS | O(n) | O(h) | First find heights, then diameters |
| BFS | O(n) | O(n) | More complex, not natural for this problem |

````python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def diameterOfBinaryTree(self, root: TreeNode) -> int:
        """
        Compute diameter of binary tree using DFS.
        
        Time: O(n) - visit each node once
        Space: O(h) - recursion stack
        """
        self.diameter = 0  # Global variable to track maximum diameter
        
        def dfs(node):
            """
            Returns height of subtree rooted at node.
            Height is measured in number of edges.
            """
            if not node:
                return 0
            
            # Recursively get heights of left and right subtrees
            left_height = dfs(node.left)
            right_height = dfs(node.right)
            
            # Diameter through this node is sum of left and right heights
            current_diameter = left_height + right_height
            
            # Update global maximum diameter
            self.diameter = max(self.diameter, current_diameter)
            
            # Return height of current node to its parent
            # Height = 1 edge to child + max height of children
            return 1 + max(left_height, right_height)
        
        dfs(root)
        return self.diameter
````

````python
# Test cases
def build_tree(values):
    """Build binary tree from level-order list (None for null nodes)"""
    if not values:
        return None
    nodes = [TreeNode(v) if v is not None else None for v in values]
    kids = nodes[::-1]
    root = kids.pop()
    for node in nodes:
        if node:
            if kids: node.left = kids.pop()
            if kids: node.right = kids.pop()
    return root

tests = [
    ([1,2,3,4,5], 3),           # Path: 4->2->1->3 or 5->2->1->3
    ([1,2], 1),                 # Single edge
    ([1], 0),                   # Single node, no edges
    ([1,2,3,4,5,None,None,6,7], 5),  # Longer path through subtree
]

solver = Solution()
for values, expected in tests:
    root = build_tree(values)
    result = solver.diameterOfBinaryTree(root)
    assert result == expected, f"Failed for {values}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n) where n = number of nodes
  - Each node visited exactly once
  - Constant work per node
- **Space**: O(h) where h = tree height
  - Recursion call stack
  - O(log n) for balanced tree
  - O(n) worst case for skewed tree

## Edge Cases & Pitfalls
- **Single node**: Diameter is 0 (no edges)
- **Two nodes**: Diameter is 1 (one edge)
- **Skewed tree**: Diameter equals height of tree
- **Balanced tree**: Diameter approximately 2 * height
- **Empty tree**: Not possible per constraints, but would return 0
- **Common mistake**: Returning diameter instead of height from DFS
- **Common mistake**: Counting nodes instead of edges (off-by-one error)
- **Common mistake**: Forgetting to update global maximum at each node
- **Node values irrelevant**: Values don't matter, only tree structure matters

## Follow-up Variants
- **Return the path**: Return actual nodes in the diameter path, not just length
- **Weighted edges**: Edges have weights, find maximum weighted diameter
- **k-diameter**: Find k-th longest path in tree
- **n-ary tree**: Extend to trees with more than 2 children per node
- **Forest**: Multiple trees, find diameter of each
- **Diameter with node deletion**: Find node whose removal minimizes diameter
- **Center of tree**: Find node(s) that minimize maximum distance to any other node

## Takeaways
- **Diameter through node = left_height + right_height** is the key insight.
- Pattern: compute local property (diameter through node) while returning different value (height) to parent.
- **Global variable** is cleaner than returning tuple for this problem.
- Post-order DFS is natural when child values determine parent value.
- Understanding height vs diameter distinction is crucial.
- This pattern (global max while computing local values) appears in many tree problems.
- The algorithm is elegant because it combines two computations (height and diameter) in one pass.

## Problem Statement 原题描述
Given a `m x n` grid filled with non-negative numbers, find a path from top left to bottom right, which minimizes the sum of all numbers along its path.

**Note:** You can only move either down or right at any point in time.

## Progressive Hints
- Hint 1: Similar to Unique Paths, but instead of counting, we minimize cost.
- Hint 2: To reach (i, j), you come from either (i-1, j) or (i, j-1).
- Hint 3: Choose the path with minimum cost: dp[i][j] = grid[i][j] + min(dp[i-1][j], dp[i][j-1]).
- Hint 4: Initialize first row and column by accumulating costs.
- Hint 5: Can optimize to O(n) space using 1D array or modify grid in-place.

## Solution Overview
Use dynamic programming where `dp[i][j]` represents the minimum path sum to reach cell (i, j).

**Recurrence**:
```
dp[i][j] = grid[i][j] + min(dp[i-1][j], dp[i][j-1])
```

**Base cases**:
- `dp[0][0] = grid[0][0]`
- First row: `dp[0][j] = dp[0][j-1] + grid[0][j]`
- First column: `dp[i][0] = dp[i-1][0] + grid[i][0]`

Can optimize space by modifying grid in-place or using 1D array.

## Detailed Explanation
**Approach: 2D Dynamic Programming**

1. **State Definition**: `dp[i][j]` = minimum path sum from (0,0) to (i,j)

2. **Base Cases**:
   - `dp[0][0] = grid[0][0]` (starting cell)
   - First row: can only come from left
     ```
     dp[0][j] = dp[0][j-1] + grid[0][j]
     ```
   - First column: can only come from above
     ```
     dp[i][0] = dp[i-1][0] + grid[i][0]
     ```

3. **Recurrence** (for i > 0, j > 0):
   ```python
   dp[i][j] = grid[i][j] + min(dp[i-1][j],  # from above
                                dp[i][j-1])  # from left
   ```

4. **Answer**: `dp[m-1][n-1]`

**Example** (grid = [[1,3,1],[1,5,1],[4,2,1]]):
```
Grid:      DP values:
1 3 1      1  4  5
1 5 1      2  7  6
4 2 1      6  8  7

Optimal path: 1→3→1→1→1 = 7
```

**In-place optimization**:
```python
# Modify grid directly to save space
for i in range(1, n):
    grid[0][i] += grid[0][i-1]
for i in range(1, m):
    grid[i][0] += grid[i-1][0]
for i in range(1, m):
    for j in range(1, n):
        grid[i][j] += min(grid[i-1][j], grid[i][j-1])
```

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| 2D DP | O(m×n) | O(m×n) | Separate dp array |
| 1D DP | O(m×n) | O(n) | Space-optimized |
| In-place | O(m×n) | O(1) | Modify input grid |
| Dijkstra | O(mn log(mn)) | O(mn) | Overkill for grid problem |

````python
from typing import List

class Solution:
    def minPathSum(self, grid: List[List[int]]) -> int:
        """
        In-place DP solution.
        
        Time: O(m×n)
        Space: O(1)
        """
        if not grid or not grid[0]:
            return 0
        
        m, n = len(grid), len(grid[0])
        
        # Initialize first row (can only come from left)
        for j in range(1, n):
            grid[0][j] += grid[0][j-1]
        
        # Initialize first column (can only come from above)
        for i in range(1, m):
            grid[i][0] += grid[i-1][0]
        
        # Fill rest of the grid
        for i in range(1, m):
            for j in range(1, n):
                grid[i][j] += min(grid[i-1][j], grid[i][j-1])
        
        return grid[m-1][n-1]
````

````python
# Test cases
tests = [
    ([[1,3,1],[1,5,1],[4,2,1]], 7),
    ([[1,2,3],[4,5,6]], 12),
    ([[1]], 1),
    ([[1,2],[1,1]], 3),
]

solver = Solution()
for grid, expected in tests:
    # Create a copy since we modify in-place
    grid_copy = [row[:] for row in grid]
    result = solver.minPathSum(grid_copy)
    assert result == expected, f"Failed for {grid}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(m×n) - Visit each cell once
- **Space**: O(1) - Modify grid in-place (or O(n) with 1D array)

## Edge Cases & Pitfalls
- **Single cell grid**: Return that cell's value
- **Single row or column**: Path is predetermined, just sum all cells
- **Large values**: Ensure no integer overflow in accumulated sum
- **All zeros**: Valid input, return 0
- **Modifying input**: If using in-place approach, original grid is destroyed
- **Initialization order**: Must initialize first row and column before filling rest

## Follow-up Variants
- **With obstacles**: Some cells cannot be visited (combine with LC 63)
- **4-directional movement**: Can move up/down/left/right (use Dijkstra)
- **Negative values**: Grid contains negative numbers (still solvable with DP)
- **Return actual path**: Track path while computing minimum
- **K paths**: Find k non-overlapping minimum paths
- **Maximum path sum**: Change min to max in recurrence
- **Triangle**: Grid is triangular shape (LC 120)

## Takeaways
- Minimum Path Sum extends Unique Paths from counting to optimization.
- The pattern `dp[i][j] = f(grid[i][j], min/max(dp[i-1][j], dp[i][j-1]))` is fundamental.
- In-place modification is a valuable space optimization when allowed.
- This problem demonstrates the transition from combinatorial counting to cost minimization.
- The recurrence structure is the same, only the aggregation function changes (sum → min).

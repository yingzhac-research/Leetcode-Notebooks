## Topic Alignment

## Problem Statement 原题描述
Given a string `s`, return the **longest palindromic substring** in `s`.

**Constraints**:
- 1 <= s.length <= 1000
- s consist of only digits and English letters

## Progressive Hints
- Hint 1: A palindrome reads the same forwards and backwards: s[i:j+1] is palindrome if s[i] == s[j] and s[i+1:j] is palindrome.
- Hint 2: Use 2D DP: dp[i][j] = True if s[i:j+1] is palindrome.
- Hint 3: Base cases: single characters are palindromes, check two consecutive characters.
- Hint 4: Build from smaller substrings to larger ones (bottom-up by length).
- Hint 5: Alternative: expand around center approach is O(n²) time with O(1) space.

## Solution Overview
Multiple approaches exist:

**1. Interval DP** (O(n²) time, O(n²) space):
- `dp[i][j]` = True if s[i:j+1] is palindrome
- Build from length 1 to length n

**2. Expand Around Center** (O(n²) time, O(1) space):
- For each possible center, expand outward while characters match
- Handle both odd-length (single center) and even-length (two centers)

**3. Manacher's Algorithm** (O(n) time, O(n) space):
- Advanced linear-time algorithm
- Uses symmetry properties to avoid recomputation

We'll focus on **Expand Around Center** (most practical) and **Interval DP** (demonstrates DP pattern).

## Detailed Explanation

### Approach 1: Interval DP

**Interval DP** is a DP pattern for problems on contiguous subarrays/substrings.

**State Definition**:
- `dp[i][j]` = True if substring s[i:j+1] is a palindrome

**Recurrence Relation**:
```python
dp[i][j] = (s[i] == s[j]) and (j - i <= 2 or dp[i+1][j-1])
```

**Explanation**:
- If `s[i] == s[j]`:
  - If length ≤ 3 (j - i <= 2): automatically palindrome (e.g., "a", "aa", "aba")
  - If length > 3: check if inner substring `s[i+1:j]` is palindrome
- If `s[i] != s[j]`: not a palindrome

**Iteration Order**:
- Must build from smaller to larger substrings
- Two ways:
  1. **By length**: for length in 1 to n, for i in 0 to n-length
  2. **Diagonal traversal**: for i from n-1 to 0, for j from i to n-1

---

### Approach 2: Expand Around Center

**Key Insight**: Every palindrome has a center.
- **Odd-length palindrome**: single character center (e.g., "aba")
- **Even-length palindrome**: two character center (e.g., "abba")

**Algorithm**:
1. For each possible center (2n - 1 centers total):
   - n centers for odd-length (each character)
   - n-1 centers for even-length (between each pair)
2. Expand outward while characters match
3. Track the longest palindrome found

**Implementation**:
```python
def expandAroundCenter(s, left, right):
    while left >= 0 and right < len(s) and s[left] == s[right]:
        left -= 1
        right += 1
    return right - left - 1  # Length of palindrome

for i in range(len(s)):
    len1 = expandAroundCenter(s, i, i)      # Odd-length
    len2 = expandAroundCenter(s, i, i + 1)  # Even-length
    max_len = max(len1, len2)
```

---

### Example Walkthrough (Expand Around Center)

**Input**: s = "babad"

**Process**:
- Center at index 0 ('b'):
  - Odd: just 'b' (length 1)
  - Even: 'b' vs 'a', no match (length 0)

- Center at index 1 ('a'):
  - Odd: expand to 'bab' (length 3) ✓
  - Even: 'a' vs 'b', no match

- Center at index 2 ('b'):
  - Odd: expand to 'aba' (length 3) ✓
  - Even: 'b' vs 'a', no match

- Center at index 3 ('a'):
  - Odd: just 'a' (length 1)
  - Even: 'a' vs 'd', no match

- Center at index 4 ('d'):
  - Odd: just 'd' (length 1)

**Result**: Longest is "bab" or "aba" (length 3)

### Example Walkthrough (Interval DP)

**Input**: s = "babad"

**DP Table** (dp[i][j] for s[i:j+1]):

```
     0   1   2   3   4
     b   a   b   a   d
0 b  T   F   T   F   F
1 a      T   F   T   F
2 b          T   F   F
3 a              T   F
4 d                  T
```

**Building process**:
- Length 1: All True (single characters)
- Length 2: Check adjacent pairs
- Length 3: dp[0][2] = (s[0]=='b', s[2]=='b') and dp[1][1] = True ✓
- Length 3: dp[1][3] = (s[1]=='a', s[3]=='a') and dp[2][2] = True ✓
- Continue for larger lengths...

**Longest**: "bab" (dp[0][2]) or "aba" (dp[1][3])

---

### Interval DP Pattern Summary

**Interval DP** is used when:
- Problem involves contiguous subarrays/substrings
- Solution for [i, j] depends on solutions for smaller intervals
- Common pattern: dp[i][j] represents property of s[i:j+1]

**Template**:
```python
# Initialize base cases
for i in range(n):
    dp[i][i] = base_value

# Build from smaller to larger intervals
for length in range(2, n + 1):
    for i in range(n - length + 1):
        j = i + length - 1
        # Compute dp[i][j] based on smaller intervals
        dp[i][j] = f(dp[i+1][j], dp[i][j-1], dp[i+1][j-1], ...)
```

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Brute force | O(n³) | O(1) | Check all substrings |
| Interval DP | O(n²) | O(n²) | 2D DP table |
| Expand around center | O(n²) | O(1) | Most practical |
| Manacher's algorithm | O(n) | O(n) | Complex but optimal |

````python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        """
        Expand around center approach.
        
        Time: O(n²)
        Space: O(1)
        """
        if not s:
            return ""
        
        start = 0
        max_len = 0
        
        def expandAroundCenter(left: int, right: int) -> int:
            """Expand around center and return length of palindrome."""
            while left >= 0 and right < len(s) and s[left] == s[right]:
                left -= 1
                right += 1
            return right - left - 1
        
        for i in range(len(s)):
            # Odd-length palindrome (single center)
            len1 = expandAroundCenter(i, i)
            # Even-length palindrome (two centers)
            len2 = expandAroundCenter(i, i + 1)
            
            # Update longest if found longer palindrome
            curr_max = max(len1, len2)
            if curr_max > max_len:
                max_len = curr_max
                # Calculate start position
                start = i - (curr_max - 1) // 2
        
        return s[start:start + max_len]
````

````python
# Test cases
tests = [
    ("babad", ["bab", "aba"]),  # Either is valid
    ("cbbd", ["bb"]),
    ("a", ["a"]),
    ("ac", ["a", "c"]),          # Either single character
    ("racecar", ["racecar"]),   # Entire string
    ("abcba", ["abcba"]),
    ("abcdcba", ["abcdcba"]),
    ("", [""]),
]

solver = Solution()
for s, expected in tests:
    result = solver.longestPalindrome(s)
    assert result in expected, f"Failed for s='{s}': got '{result}', expected one of {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n²) where n = len(s)
  - Try 2n - 1 centers (n odd + n-1 even)
  - Each expansion can take up to O(n) time
  - Total: O(n) × O(n) = O(n²)
- **Space**: O(1)
  - Only use constant extra space for variables
  - No DP table needed

## Edge Cases & Pitfalls
- **Empty string**: Return "" (by constraints, won't happen: length ≥ 1)
- **Single character**: Return that character
- **No palindrome > 1**: Return any single character
- **Entire string is palindrome**: Return entire string
- **Multiple answers**: Any valid longest palindrome is acceptable
- **Even vs odd length**: Must handle both cases (two expand functions)
- **Index calculation**: When updating start, use formula: i - (max_len - 1) // 2
- **Off-by-one**: Ensure substring extraction is s[start:start + max_len], not s[start:end]

- **Two Cases**: Always handle both odd-length and even-length palindromes

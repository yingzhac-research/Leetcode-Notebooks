## Problem Statement
There is a robot on an `m x n` grid. The robot is initially located at the **top-left corner** (i.e., `grid[0][0]`). The robot tries to move to the **bottom-right corner** (i.e., `grid[m-1][n-1]`). The robot can only move either down or right at any point in time.

Given the two integers `m` and `n`, return the number of possible unique paths that the robot can take to reach the bottom-right corner.

The test cases are generated so that the answer will be less than or equal to `2 * 10^9`.

## Progressive Hints
- Hint 1: To reach cell (i, j), you must come from either (i-1, j) or (i, j-1).
- Hint 2: The number of paths to (i, j) equals the sum of paths to (i-1, j) and (i, j-1).
- Hint 3: Use a 2D DP table where dp[i][j] represents paths to position (i, j).
- Hint 4: Base cases: First row and first column all have exactly 1 path.
- Hint 5: Can optimize space to O(n) by using a 1D array.

## Solution Overview
This is a classic 2D grid DP problem. The recurrence relation is:

```
dp[i][j] = dp[i-1][j] + dp[i][j-1]
```

**Approaches**:
1. **2D DP**: O(m×n) time and space
2. **1D DP**: O(m×n) time, O(n) space - reuse a single row
3. **Combinatorics**: O(m+n) time, O(1) space - mathematical formula

The combinatorics approach recognizes that we need exactly (m-1) down moves and (n-1) right moves, totaling (m+n-2) moves. The answer is C(m+n-2, m-1) = (m+n-2)! / ((m-1)! × (n-1)!)

## Detailed Explanation

### Approach 1: 2D Dynamic Programming

1. **State Definition**: `dp[i][j]` = number of unique paths to reach cell (i, j)

2. **Base Cases**:
   - `dp[0][j] = 1` for all j (first row: only one path - keep going right)
   - `dp[i][0] = 1` for all i (first column: only one path - keep going down)

3. **Recurrence**: For i > 0 and j > 0:
   ```
   dp[i][j] = dp[i-1][j] + dp[i][j-1]
   ```
   Paths from above + paths from left

4. **Answer**: `dp[m-1][n-1]`

**Example** (m=3, n=3):
```
Grid indices:      DP values:
(0,0) (0,1) (0,2)  1  1  1
(1,0) (1,1) (1,2)  1  2  3
(2,0) (2,1) (2,2)  1  3  6
```
Answer: 6 paths

---

### Approach 2: Space-Optimized 1D DP

Observation: To compute row i, we only need row i-1. We can use a single array:

```python
dp = [1] * n  # First row all 1s
for i in range(1, m):
    for j in range(1, n):
        dp[j] = dp[j] + dp[j-1]
        # dp[j] currently holds value from previous row (i-1, j)
        # dp[j-1] holds updated value for current row (i, j-1)
```

---

### Approach 3: Combinatorics

**Key Insight**: Any path from (0,0) to (m-1, n-1) consists of:
- Exactly (m-1) down moves
- Exactly (n-1) right moves
- Total: (m+n-2) moves

Question becomes: In how many ways can we choose (m-1) positions out of (m+n-2) for down moves?

**Answer**: C(m+n-2, m-1) = (m+n-2)! / ((m-1)! × (n-1)!)

To avoid overflow, compute incrementally:
```python
result = 1
for i in range(1, min(m, n)):
    result = result * (m + n - 1 - i) // i
```

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| 2D DP | O(m×n) | O(m×n) | Straightforward, clear |
| 1D DP | O(m×n) | O(n) | Space-optimized DP |
| Combinatorics | O(m+n) | O(1) | Mathematical, fastest |
| Recursion + memo | O(m×n) | O(m×n) | Top-down approach |

````python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        """
        Space-optimized 1D DP solution.
        
        Time: O(m×n)
        Space: O(n)
        """
        dp = [1] * n  # First row all 1s
        
        for i in range(1, m):
            for j in range(1, n):
                # dp[j] = paths from above (previous row)
                # dp[j-1] = paths from left (current row)
                dp[j] = dp[j] + dp[j-1]
        
        return dp[n-1]
    
    def uniquePaths_combinatorics(self, m: int, n: int) -> int:
        """
        Combinatorics solution: C(m+n-2, m-1)
        
        Time: O(min(m, n))
        Space: O(1)
        """
        # Need to choose m-1 down moves from m+n-2 total moves
        # C(m+n-2, m-1) = (m+n-2)! / ((m-1)! * (n-1)!)
        
        result = 1
        # Compute C(m+n-2, min(m-1, n-1)) to minimize iterations
        k = min(m-1, n-1)
        for i in range(k):
            result = result * (m + n - 2 - i) // (i + 1)
        
        return result
````

````python
# Test cases
tests = [
    (3, 7, 28),
    (3, 2, 3),
    (1, 1, 1),
    (2, 2, 2),
    (3, 3, 6),
    (10, 10, 48620),
]

solver = Solution()
for m, n, expected in tests:
    result_dp = solver.uniquePaths(m, n)
    result_comb = solver.uniquePaths_combinatorics(m, n)
    assert result_dp == expected, f"DP failed for ({m}, {n}): got {result_dp}, expected {expected}"
    assert result_comb == expected, f"Combinatorics failed for ({m}, {n}): got {result_comb}, expected {expected}"
print('All tests passed for both approaches.')
````

## Complexity Analysis

**1D DP Approach:**
- **Time**: O(m×n) - Fill m rows, each with n elements
- **Space**: O(n) - Single array of size n

**Combinatorics Approach:**
- **Time**: O(min(m, n)) - Compute combination incrementally
- **Space**: O(1) - Only a few variables

## Edge Cases & Pitfalls
- **1×1 grid**: Should return 1 (already at destination)
- **1×n or m×1 grid**: Only one path (straight line)
- **Large grids**: Combinatorics approach is much faster
- **Integer overflow**: In languages with fixed-size integers, be careful with factorial calculations
- **Order of operations**: In combinatorics, multiply before divide to avoid precision loss
- **Base case initialization**: Ensure first row and column are all 1s

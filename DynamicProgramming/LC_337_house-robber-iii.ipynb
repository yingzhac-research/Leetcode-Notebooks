## Problem Statement
The thief has found himself a new place for his thievery again. There is only one entrance to this area, called `root`.

Besides the `root`, each house has one and only one parent house. After a tour, the smart thief realized that all houses in this place form a **binary tree**. It will automatically contact the police if **two directly-linked houses were broken into on the same night**.

Given the `root` of the binary tree, return the maximum amount of money the thief can rob **without alerting the police**.

**Constraints**:
- The number of nodes in the tree is in the range `[1, 10^4]`.
- `0 <= Node.val <= 10^4`

## Progressive Hints
- Hint 1: For each node, you have two choices: rob it or don't rob it.
- Hint 2: If you rob the current node, you cannot rob its children (but can rob grandchildren).
- Hint 3: If you don't rob the current node, you can choose the optimal strategy for both children.
- Hint 4: Use DFS to traverse the tree and compute two states for each node: max value when robbing, max value when not robbing.
- Hint 5: Return a pair (rob, not_rob) from each recursive call to avoid redundant computation.
- Hint 6: The answer at root is max(rob_root, not_rob_root).

## Solution Overview
Use **Tree DP** with post-order DFS traversal.

**State Definition**: For each node, compute two values:
- `rob`: Maximum money if we rob this node
- `not_rob`: Maximum money if we don't rob this node

**Recurrence**:
```python
# If we rob current node, we cannot rob children
rob = node.val + left_not_rob + right_not_rob

# If we don't rob current, we can choose optimal for each child
not_rob = max(left_rob, left_not_rob) + max(right_rob, right_not_rob)
```

**Base case**: For null nodes, return (0, 0)

**Alternative approaches**:
1. Naive recursion with memoization (slower, easier to understand)
2. Post-order DFS returning pair (optimal, clean code)

## Detailed Explanation

### Understanding Tree DP

Tree DP problems involve computing optimal values on trees where:
- Each node's value depends on its children's values
- We use post-order traversal (process children before parent)
- States are typically stored at each node

---

### Problem Analysis

For any node, we have **two choices**:
1. **Rob this node**: Get node.val + optimal from grandchildren
2. **Don't rob this node**: Get optimal from children (they can be robbed or not)

The constraint is: **cannot rob both parent and child**.

---

### State Definition

For each node, we maintain:
- `rob`: Maximum money if we **must rob** this node
- `not_rob`: Maximum money if we **must not rob** this node

---

### Recurrence Relations

**Case 1: Rob current node**
- We get `node.val`
- We **cannot** rob left child or right child
- But we **can** get the optimal from not robbing them
- `rob = node.val + left_not_rob + right_not_rob`

**Case 2: Don't rob current node**
- We don't get `node.val`
- We **can** choose optimally for left and right children
- Each child can be robbed or not (take max)
- `not_rob = max(left_rob, left_not_rob) + max(right_rob, right_not_rob)`

---

### Example Walkthrough

Tree:
```
     3
    / \\
   2   3
    \\   \\
     3   1
```

**Bottom-up computation**:

Node 3 (left leaf): `rob=3, not_rob=0`

Node 1 (right leaf): `rob=1, not_rob=0`

Node 2 (left child of root):
- Left child: null → (0, 0)
- Right child: (3, 0)
- `rob = 2 + 0 + 0 = 2`
- `not_rob = max(0, 0) + max(3, 0) = 3`
- Result: (2, 3)

Node 3 (right child of root):
- Left child: null → (0, 0)
- Right child: (1, 0)
- `rob = 3 + 0 + 0 = 3`
- `not_rob = max(0, 0) + max(1, 0) = 1`
- Result: (3, 1)

Root node 3:
- Left child: (2, 3)
- Right child: (3, 1)
- `rob = 3 + 3 + 1 = 7`
- `not_rob = max(2, 3) + max(3, 1) = 3 + 3 = 6`
- Result: (7, 6)

**Answer**: max(7, 6) = **7** (rob root, don't rob children, rob grandchildren)

---

### Why Post-Order Traversal?

We need children's results before computing parent's result:
1. Recursively compute left subtree → get (left_rob, left_not_rob)
2. Recursively compute right subtree → get (right_rob, right_not_rob)
3. Use children's results to compute current node's values
4. Return (rob, not_rob) to parent

This is classic **bottom-up DP on trees**.

## Complexity Trade-off Table
| Approach | Time | Space | Notes |
| --- | --- | --- | --- |
| Naive recursion | O(2^n) | O(h) | Exponential, recomputes subtrees |
| Recursion + memo | O(n) | O(n) | HashMap to store results |
| Tree DP (optimal) | O(n) | O(h) | Post-order DFS, returns pair |
| Iterative (post-order) | O(n) | O(n) | Using stack, more complex |

````python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def rob(self, root: TreeNode) -> int:
        """
        Tree DP solution using post-order DFS.
        Returns maximum money without robbing adjacent nodes.
        
        Time: O(n) - visit each node once
        Space: O(h) - recursion stack, h = tree height
        """
        def dfs(node):
            """
            Returns (rob, not_rob) for current node.
            rob: max money if we rob this node
            not_rob: max money if we don't rob this node
            """
            if not node:
                return (0, 0)
            
            # Post-order: process children first
            left_rob, left_not_rob = dfs(node.left)
            right_rob, right_not_rob = dfs(node.right)
            
            # If we rob current node, we cannot rob children
            rob = node.val + left_not_rob + right_not_rob
            
            # If we don't rob current, we can choose optimal for each child
            not_rob = max(left_rob, left_not_rob) + max(right_rob, right_not_rob)
            
            return (rob, not_rob)
        
        rob_root, not_rob_root = dfs(root)
        return max(rob_root, not_rob_root)
````

````python
# Test cases
def build_tree(values):
    """Build binary tree from level-order list (None for null nodes)"""
    if not values:
        return None
    nodes = [TreeNode(v) if v is not None else None for v in values]
    kids = nodes[::-1]
    root = kids.pop()
    for node in nodes:
        if node:
            if kids: node.left = kids.pop()
            if kids: node.right = kids.pop()
    return root

tests = [
    ([3,2,3,None,3,None,1], 7),  # Rob root + grandchildren
    ([3,4,5,1,3,None,1], 9),     # Rob 4 + 5 = 9
    ([1], 1),                     # Single node
    ([2,1,3,None,4], 7),         # Rob 2 + 4 + (implicit) = 7 or rob 3+4=7
]

solver = Solution()
for values, expected in tests:
    root = build_tree(values)
    result = solver.rob(root)
    assert result == expected, f"Failed for {values}: got {result}, expected {expected}"
print('All tests passed.')
````

## Complexity Analysis
- **Time**: O(n) where n = number of nodes
  - Each node visited exactly once in DFS
  - Constant work per node
- **Space**: O(h) where h = tree height
  - Recursion call stack depth equals height
  - O(log n) for balanced tree, O(n) for skewed tree
  - No extra data structures needed

## Edge Cases & Pitfalls
- **Single node**: Return node.val
- **Null root**: Should return 0
- **Skewed tree**: Space complexity becomes O(n) due to deep recursion
- **All same values**: Algorithm still works correctly
- **Large values**: Sum might overflow in other languages (not Python)
- **Common mistake**: Forgetting that not robbing current allows **both** rob/not_rob options for children
- **State confusion**: rob means "we rob this node", not "this is max if parent allows"
- **Return value**: Must take max(rob, not_rob) at root, not just one of them

## Follow-up Variants
- **General tree (n-ary)**: Extend to nodes with arbitrary number of children
- **Weighted edges**: Different penalties for robbing connected nodes
- **k-distance constraint**: Cannot rob nodes within k distance (not just direct parent-child)
- **Return path**: Return which nodes to rob, not just maximum value
- **Multiple thieves**: Coordinate k thieves to maximize total without alerting
- **Directed acyclic graph**: Extend to DAG instead of tree
- **Probabilistic**: Each house has probability of being home

## Takeaways
- Tree DP problems use **post-order DFS** to compute values from children to parent.
- **State design is crucial**: returning a pair (rob, not_rob) elegantly captures both options.
- This pattern is fundamental: computing optimal values on subtrees and combining them.
- The constraint "cannot take adjacent" is handled by state definition, not explicit checking.
- Tree DP is more elegant than memoization for tree problems as it naturally fits the recursion.
- This technique extends to many tree optimization problems: tree coloring, independent sets, vertex cover.
- Understanding when to use post-order (children before parent) vs pre-order (parent before children) is key.

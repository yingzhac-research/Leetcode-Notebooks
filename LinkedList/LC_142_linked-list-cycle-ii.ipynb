## Problem Statement
Given the `head` of a linked list, return the node where the cycle begins. If there is no cycle, return `None`.

There is a cycle in a linked list if there is some node in the list that can be reached again by continuously following the `next` pointer. Internally, `pos` is used to denote the index of the node that tail's `next` pointer is connected to (0-indexed). Note that `pos` is not passed as a parameter.

Input: `head` - the head node of a linked list.
Output: The node where the cycle begins, or `None` if there's no cycle.
Constraints: The number of nodes is in the range `[0, 10^4]`. `-10^5 <= Node.val <= 10^5`. Do not modify the linked list.

## Progressive Hints
- Hint 1: First detect if a cycle exists using Floyd's fast-slow pointer technique (LC 141).
- Hint 2: Once you detect a cycle (when fast and slow meet), there's a mathematical relationship between the meeting point and the cycle entrance.
- Hint 3: Let's say the distance from head to cycle start is `a`, the meeting point is `b` steps into the cycle, and the cycle length is `c`. When they meet, slow traveled `a+b` and fast traveled `a+b+kc` for some k.
- Hint 4: Since fast moves twice as fast, `2(a+b) = a+b+kc`, which simplifies to `a = kc - b`. This means starting from head and from the meeting point, moving one step at a time, they'll meet at the cycle entrance!

## Solution Overview
The optimal solution uses Floyd's cycle detection in two phases. Phase 1: Use fast (2 steps) and slow (1 step) pointers to detect if a cycle exists. Phase 2: Once they meet, reset one pointer to head and move both one step at a time; they'll meet at the cycle entrance. This achieves O(n) time with O(1) space, with a beautiful mathematical proof.

## Detailed Explanation
**Floyd's Cycle Detection with Entry Point Finding:**

**Phase 1 - Detect Cycle:**
1. Initialize `slow = head` and `fast = head`.
2. Move slow one step and fast two steps at a time.
3. If fast reaches None (no cycle), return None.
4. If slow equals fast, a cycle exists; proceed to Phase 2.

**Phase 2 - Find Entry Point:**
1. Reset one pointer to head: `slow = head`.
2. Keep fast at the meeting point.
3. Move both pointers one step at a time.
4. When they meet again, that node is the cycle entrance.

**Mathematical Proof:**
- Let distance from head to cycle start = `a`
- Let distance from cycle start to meeting point = `b`
- Let remaining cycle length = `c` (so total cycle length is `b+c`)
- When they meet: slow traveled `a+b`, fast traveled `a+b+n(b+c)` for some integer n
- Since fast is twice as fast: `2(a+b) = a+b+n(b+c)`
- Simplifying: `a+b = n(b+c)`, so `a = n(b+c) - b = (n-1)(b+c) + c`
- This means: moving `a` steps from meeting point lands you at the cycle start (after going around the cycle n-1 times plus `c` more steps)
- Therefore, moving from head and from meeting point simultaneously, both one step at a time, they meet at cycle entrance!

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def detectCycle(head: Optional[ListNode]) -> Optional[ListNode]:
    """Find the node where the cycle begins using Floyd's algorithm."""
    if not head or not head.next:
        return None
    
    # Phase 1: Detect if cycle exists.
    slow = head
    fast = head
    
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        
        if slow == fast:
            # Cycle detected, proceed to Phase 2.
            break
    else:
        # No cycle found.
        return None
    
    # Phase 2: Find the entry point.
    slow = head
    while slow != fast:
        slow = slow.next
        fast = fast.next
    
    return slow  # The entry point of the cycle.
```

## Validation

```python
def create_linked_list_with_cycle(values, pos):
    """Helper to create a linked list with a cycle at position pos (-1 for no cycle)."""
    if not values:
        return None
    
    head = ListNode(values[0])
    current = head
    nodes = [head]
    
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
        nodes.append(current)
    
    # Create cycle if pos is valid.
    if pos >= 0:
        current.next = nodes[pos]
    
    return head, nodes[pos] if pos >= 0 else None


# Test cases
test_cases = [
    ([3, 2, 0, -4], 1),  # Cycle starts at node with value 2
    ([1, 2], 0),         # Cycle starts at head
    ([1], -1),           # No cycle
    ([1, 2, 3, 4], 2),   # Cycle starts at node with value 3
]

for values, pos in test_cases:
    head, expected_node = create_linked_list_with_cycle(values, pos)
    result = detectCycle(head)
    assert result == expected_node, f"Failed for {values}, pos={pos}"

print('All tests passed for LC 142.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. Phase 1 takes O(n) as pointers will meet within one cycle traversal. Phase 2 takes O(n) to find the entrance.
- Space Complexity: O(1) as we only use two pointers regardless of list size.
- Primary Bottleneck: Must detect the cycle first, which requires traversing until pointers meet. This is optimal for constant space.

## Edge Cases & Pitfalls
- Empty list or single node without cycle should return None.
- Cycle starting at the head (pos = 0) is handled correctly by the algorithm.
- Must check both `fast` and `fast.next` in the while condition to avoid null pointer errors.
- The while-else construct in Python elegantly handles the no-cycle case.

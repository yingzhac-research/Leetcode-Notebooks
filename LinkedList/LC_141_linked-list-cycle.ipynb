## Problem Statement
Given `head`, the head of a linked list, determine if the linked list has a cycle in it.

There is a cycle in a linked list if there is some node in the list that can be reached again by continuously following the `next` pointer. Internally, `pos` is used to denote the index of the node that tail's `next` pointer is connected to. Note that `pos` is not passed as a parameter.

Return `True` if there is a cycle in the linked list. Otherwise, return `False`.

Input: `head` - the head node of a linked list.
Output: Boolean indicating whether a cycle exists.
Constraints: The number of nodes in the list is in the range `[0, 10^4]`.

## Progressive Hints
- Hint 1: If you store all visited nodes in a set, you can detect when you revisit a node, but this uses O(n) space.
- Hint 2: Consider using two pointers moving at different speeds - if there's a cycle, they must eventually meet.
- Hint 3: Let one pointer (slow) advance one step at a time, and another pointer (fast) advance two steps - if fast catches slow, there's a cycle.

## Solution Overview
Use Floyd's Cycle Detection algorithm (tortoise and hare): maintain two pointers, slow and fast. Slow moves one step per iteration, fast moves two steps. If there's a cycle, fast will eventually catch up to slow. If fast reaches None, there's no cycle. This achieves O(n) time with O(1) space.

## Detailed Explanation
**Floyd's Cycle Detection (Two Pointers):**
1. Initialize two pointers: `slow = head` and `fast = head`.
2. While `fast` and `fast.next` are not None:
   - Move `slow` one step: `slow = slow.next`.
   - Move `fast` two steps: `fast = fast.next.next`.
   - If `slow == fast`, a cycle exists, return True.
3. If the loop exits (fast reaches None), no cycle exists, return False.

**Why it works:** In a cycle, the fast pointer will eventually lap the slow pointer. The relative speed is one node per iteration, so within n iterations (where n is the cycle length), they must meet.

**Alternative (Hash Set):** Track visited nodes in a set; if you encounter a node already in the set, there's a cycle. This uses O(n) space but is simpler to understand.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Hash set tracking | O(n) | O(n) | Simple but requires extra memory for all nodes. |
| Floyd's fast-slow pointers | O(n) | O(1) | Optimal space efficiency; classic two-pointer technique. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def hasCycle(head: Optional[ListNode]) -> bool:
    """Detect cycle in linked list using Floyd's algorithm (fast-slow pointers)."""
    if not head or not head.next:
        return False  # Empty list or single node cannot have a cycle.
    
    slow = head
    fast = head
    
    while fast and fast.next:
        slow = slow.next       # Move slow pointer one step.
        fast = fast.next.next  # Move fast pointer two steps.
        
        if slow == fast:       # Pointers meet, cycle detected.
            return True
    
    return False  # Fast reached the end, no cycle.


def hasCycleHashSet(head: Optional[ListNode]) -> bool:
    """Detect cycle using a hash set to track visited nodes."""
    visited = set()
    current = head
    
    while current:
        if current in visited:
            return True  # Node already visited, cycle found.
        visited.add(current)
        current = current.next
    
    return False  # Reached end without revisiting any node.
```

## Validation

```python
def create_cycle_list(values, pos):
    """Create a linked list with a cycle at position pos (-1 means no cycle)."""
    if not values:
        return None
    
    head = ListNode(values[0])
    current = head
    cycle_node = None
    
    if pos == 0:
        cycle_node = head
    
    for i, val in enumerate(values[1:], 1):
        current.next = ListNode(val)
        current = current.next
        if i == pos:
            cycle_node = current
    
    # Create cycle if pos is valid
    if pos != -1 and cycle_node:
        current.next = cycle_node
    
    return head


# Test cases
test_cases = [
    ([3, 2, 0, -4], 1, True),   # Cycle at position 1
    ([1, 2], 0, True),           # Cycle at position 0
    ([1], -1, False),            # Single node, no cycle
    ([1, 2, 3, 4], -1, False),   # Multiple nodes, no cycle
    ([], -1, False),             # Empty list
]

for values, pos, expected in test_cases:
    head = create_cycle_list(values, pos)
    result = hasCycle(head)
    assert result == expected, f"Floyd's failed for {values} with pos {pos}"
    
    # Test hash set approach
    head = create_cycle_list(values, pos)
    result = hasCycleHashSet(head)
    assert result == expected, f"HashSet failed for {values} with pos {pos}"

print('All tests passed for LC 141.')
```

## Complexity Analysis
- Time Complexity: O(n) for both approaches where n is the number of nodes. In the worst case with a cycle, the fast pointer needs to lap the slow pointer once.
- Space Complexity: O(1) for Floyd's algorithm (only two pointers), O(n) for hash set approach (storing all nodes).
- Primary Bottleneck: For Floyd's algorithm, the number of iterations is bounded by the cycle length plus the distance to the cycle.

## Edge Cases & Pitfalls
- Empty list (head is None) should return False.
- Single node without cycle should return False.
- Must check both `fast` and `fast.next` before advancing to avoid null pointer errors.
- Cycle can start at any position including the head node itself.

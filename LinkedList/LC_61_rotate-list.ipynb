## Problem Statement
Given the `head` of a linked list, rotate the list to the right by `k` places.

Input: `head` - the head of a singly linked list, `k` - a non-negative integer representing rotation positions.
Output: The head of the rotated linked list.
Constraints: The number of nodes is in the range `[0, 500]`, `-100 <= Node.val <= 100`, `0 <= k <= 2 * 10^9`.

## Progressive Hints
- Hint 1: Rotating right by k is equivalent to moving the last k nodes to the front. If k >= n (list length), use k % n to avoid redundant full rotations.
- Hint 2: First, find the list length n and connect the tail to head to form a circle.
- Hint 3: Calculate the new tail position: (n - k % n - 1) from the head, or move (n - k % n) steps to find new head.
- Hint 4: Break the circle at the new tail to get the rotated list.

## Solution Overview
The optimal approach creates a circular linked list by connecting tail to head, then breaks it at the appropriate position. First, compute the list length and normalize k using modulo. The new head is at position (n - k % n) from the old head. Form the circle, advance to the new tail (one position before new head), and break the link.

## Detailed Explanation
**Circular Approach:**
1. Handle edge cases: if head is None, head.next is None, or k is 0, return head (no rotation needed).
2. Find list length n and locate the tail:
   - Traverse to count nodes and reach the last node.
3. Normalize k: `k = k % n`. If k is 0 after modulo, no rotation needed, return head.
4. Form a circle: `tail.next = head`.
5. Find the new tail position: move (n - k - 1) steps from head to reach the new tail.
6. Identify new head: `new_head = new_tail.next`.
7. Break the circle: `new_tail.next = None`.
8. Return new_head.

**Example:** List [1,2,3,4,5], k=2
- n = 5, k % 5 = 2
- Form circle: 5->1
- New tail is at position 5-2-1=2 from head, which is node 3
- New head is node 4
- Break at 3: result is 4->5->1->2->3

**Key insight:** Forming a circle and breaking at the right position is cleaner than trying to detach and reattach sublists.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Circular | O(n) | O(1) | Optimal; makes two passes (length + rotation). |
| Two pointers | O(n) | O(1) | Alternative; find kth from end, rearrange pointers. |
| Brute force | O(k * n) | O(1) | Rotate one position k times; too slow for large k. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def rotateRight(head: Optional[ListNode], k: int) -> Optional[ListNode]:
    """Rotate list to the right by k positions using circular approach."""
    # Edge cases
    if not head or not head.next or k == 0:
        return head
    
    # Find length and tail
    length = 1
    tail = head
    while tail.next:
        tail = tail.next
        length += 1
    
    # Normalize k
    k = k % length
    if k == 0:
        return head  # No rotation needed
    
    # Form circular list
    tail.next = head
    
    # Find new tail (n - k - 1 steps from head)
    new_tail = head
    for _ in range(length - k - 1):
        new_tail = new_tail.next
    
    # Break circle at new tail
    new_head = new_tail.next
    new_tail.next = None
    
    return new_head


def rotateRightTwoPointers(head: Optional[ListNode], k: int) -> Optional[ListNode]:
    """Alternative: find kth from end, rearrange pointers."""
    if not head or not head.next or k == 0:
        return head
    
    # Find length
    length = 0
    current = head
    while current:
        length += 1
        current = current.next
    
    # Normalize k
    k = k % length
    if k == 0:
        return head
    
    # Find new tail (n - k - 1 from head)
    new_tail = head
    for _ in range(length - k - 1):
        new_tail = new_tail.next
    
    # Rearrange pointers
    new_head = new_tail.next
    new_tail.next = None
    
    # Connect old tail to old head
    current = new_head
    while current.next:
        current = current.next
    current.next = head
    
    return new_head
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4, 5], 2, [4, 5, 1, 2, 3]),
    ([0, 1, 2], 4, [2, 0, 1]),  # k > n, 4 % 3 = 1
    ([1, 2], 1, [2, 1]),
    ([1, 2], 2, [1, 2]),  # Full rotation
    ([1], 0, [1]),
    ([1], 1, [1]),  # Single node, any k returns same
    ([1, 2, 3], 0, [1, 2, 3]),  # k = 0
    ([], 1, []),  # Empty list
    ([1, 2, 3, 4, 5], 10, [1, 2, 3, 4, 5]),  # k % n = 0
]

for values, k, expected in test_cases:
    # Test circular approach
    head = create_linked_list(values)
    result = rotateRight(head, k)
    assert linked_list_to_list(result) == expected, f"Circular failed for {values} with k={k}"
    
    # Test two-pointer approach
    head = create_linked_list(values)
    result = rotateRightTwoPointers(head, k)
    assert linked_list_to_list(result) == expected, f"Two-pointer failed for {values} with k={k}"

print('All tests passed for LC 61.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the list length. Circular approach makes two passes: one to find length/tail, one to find new tail. Two-pointer makes three passes but still O(n).
- Space Complexity: O(1) for both approaches, using only a few pointers.
- Primary Bottleneck: For very large k (up to 2*10^9), the modulo operation is essential to reduce k to manageable size, preventing timeout.

## Edge Cases & Pitfalls
- Empty list or single node should return unchanged.
- k = 0 requires no rotation, return head immediately.
- k >= n requires normalization with k % n to avoid redundant full rotations.
- After forming circle, must remember to break it at the new tail to avoid infinite loop.
- When k % n = 0 after normalization, list remains unchanged, can return early to save time.

- The new head position in right rotation by k is at index (n - k % n) from the start.

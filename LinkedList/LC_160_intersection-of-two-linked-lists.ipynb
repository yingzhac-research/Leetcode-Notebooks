## Problem Statement
Given the heads of two singly linked-lists `headA` and `headB`, return the node at which the two lists intersect. If the two linked lists have no intersection at all, return `None`.

The test cases are generated such that there are no cycles anywhere in the entire linked structure.

Note that the linked lists must retain their original structure after the function returns.

Input: `headA` and `headB` - heads of two linked lists that may intersect.
Output: The intersection node, or None if no intersection.
Constraints: The number of nodes in each list is in the range `[0, 3 * 10^4]`. Lists may have different lengths.

## Progressive Hints
- Hint 1: If you store all nodes from one list in a set, you can check each node in the second list for intersection, but this uses O(n) space.
- Hint 2: The key insight is that if two lists intersect, their tails are the same from the intersection point onward.
- Hint 3: If you traverse both lists and switch to the other list's head when reaching the end, both pointers will meet at the intersection (or None) after traversing the same total distance.

## Solution Overview
Use two pointers starting at each list's head. Traverse both lists, and when a pointer reaches the end, redirect it to the other list's head. The pointers will meet at the intersection node (or None) after traversing the same total distance. This achieves O(n + m) time with O(1) space.

## Detailed Explanation
**Two Pointer Switching Approach:**
1. Initialize `pA = headA` and `pB = headB`.
2. Traverse both lists:
   - If `pA` is not None, move to `pA.next`; else switch to `headB`.
   - If `pB` is not None, move to `pB.next`; else switch to `headA`.
3. Continue until `pA == pB` (either the intersection node or None).
4. Return `pA` (or `pB`, they're equal).

**Why it works:** 
- Let list A have length a with c common nodes, list B have length b with c common nodes.
- Non-intersecting parts: A has (a-c) unique nodes, B has (b-c) unique nodes.
- Pointer A travels: a nodes + (b-c) nodes = a+b-c nodes to reach intersection.
- Pointer B travels: b nodes + (a-c) nodes = a+b-c nodes to reach intersection.
- Both pointers travel the same distance and meet at the intersection.

**Alternative (Hash Set):** Store all nodes from list A in a set, then traverse list B and return the first node found in the set. Uses O(n) space.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Two pointers with switching | O(n + m) | O(1) | Optimal space; elegant mathematical solution. |
| Hash set tracking | O(n + m) | O(n) or O(m) | Simpler logic but requires extra memory. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def getIntersectionNode(headA: ListNode, headB: ListNode) -> Optional[ListNode]:
    """Find intersection node using two pointers with list switching."""
    if not headA or not headB:
        return None
    
    pA = headA
    pB = headB
    
    # Traverse both lists; switch to other list's head when reaching end.
    while pA != pB:
        pA = pA.next if pA else headB
        pB = pB.next if pB else headA
    
    return pA  # Either intersection node or None.


def getIntersectionNodeHashSet(headA: ListNode, headB: ListNode) -> Optional[ListNode]:
    """Find intersection node using a hash set."""
    if not headA or not headB:
        return None
    
    visited = set()
    
    # Store all nodes from list A.
    current = headA
    while current:
        visited.add(current)
        current = current.next
    
    # Check nodes in list B for intersection.
    current = headB
    while current:
        if current in visited:
            return current
        current = current.next
    
    return None
```

## Validation

```python
def create_intersection_lists(listA_vals, listB_vals, skipA, skipB):
    """Create two lists that intersect at a specific position.
    
    listA_vals: values for list A
    listB_vals: values for list B
    skipA: number of nodes before intersection in list A
    skipB: number of nodes before intersection in list B
    """
    if skipA >= len(listA_vals) or skipB >= len(listB_vals):
        # No intersection
        headA = None
        if listA_vals:
            headA = ListNode(listA_vals[0])
            current = headA
            for val in listA_vals[1:]:
                current.next = ListNode(val)
                current = current.next
        
        headB = None
        if listB_vals:
            headB = ListNode(listB_vals[0])
            current = headB
            for val in listB_vals[1:]:
                current.next = ListNode(val)
                current = current.next
        
        return headA, headB, None
    
    # Create list A up to intersection
    headA = ListNode(listA_vals[0]) if listA_vals else None
    current = headA
    for i in range(1, skipA):
        current.next = ListNode(listA_vals[i])
        current = current.next
    
    # Create list B up to intersection
    headB = ListNode(listB_vals[0]) if listB_vals else None
    currentB = headB
    for i in range(1, skipB):
        currentB.next = ListNode(listB_vals[i])
        currentB = currentB.next
    
    # Create intersection part
    intersection = None
    if skipA < len(listA_vals):
        intersection = ListNode(listA_vals[skipA])
        current.next = intersection if current else intersection
        currentB.next = intersection if currentB else intersection
        
        current = intersection
        for i in range(skipA + 1, len(listA_vals)):
            current.next = ListNode(listA_vals[i])
            current = current.next
    
    return headA, headB, intersection


# Test cases
test_cases = [
    ([4, 1, 8, 4, 5], [5, 6, 1, 8, 4, 5], 2, 3, True),   # Intersect at node with value 8
    ([1, 9, 1, 2, 4], [3, 2, 4], 3, 1, True),             # Intersect at node with value 2
    ([2, 6, 4], [1, 5], 3, 2, False),                     # No intersection
]

for listA_vals, listB_vals, skipA, skipB, should_intersect in test_cases:
    headA, headB, expected = create_intersection_lists(listA_vals, listB_vals, skipA, skipB)
    
    # Test two-pointer approach
    result = getIntersectionNode(headA, headB)
    if should_intersect:
        assert result == expected, f"Two-pointer failed for {listA_vals}, {listB_vals}"
    else:
        assert result is None, f"Two-pointer should return None for {listA_vals}, {listB_vals}"
    
    # Test hash set approach
    result = getIntersectionNodeHashSet(headA, headB)
    if should_intersect:
        assert result == expected, f"HashSet failed for {listA_vals}, {listB_vals}"
    else:
        assert result is None, f"HashSet should return None for {listA_vals}, {listB_vals}"

print('All tests passed for LC 160.')
```

## Complexity Analysis
- Time Complexity: O(n + m) where n and m are the lengths of the two lists. Each pointer traverses at most both lists once.
- Space Complexity: O(1) for two-pointer approach (only pointer variables), O(n) or O(m) for hash set approach.
- Primary Bottleneck: Must traverse enough of both lists to equalize the path lengths.

## Edge Cases & Pitfalls
- One or both lists may be empty; return None immediately.
- Lists may have no intersection; pointers will both reach None simultaneously.
- Lists may have different lengths; the switching mechanism handles this automatically.
- Intersection may occur at the very first node (both heads are the same).

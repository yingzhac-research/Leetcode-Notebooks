## Problem Statement
You are given the head of a singly linked list. The list can be represented as:

L0 → L1 → … → Ln-1 → Ln

Reorder the list to be in the following form:

L0 → Ln → L1 → Ln-1 → L2 → Ln-2 → …

You may not modify the values in the list's nodes. Only nodes themselves may be changed.

Input: `head` - the head node of a singly linked list.
Output: None (modify the list in-place).
Constraints: The number of nodes is in the range `[1, 5 * 10^4]`. `1 <= Node.val <= 1000`.

## Progressive Hints
- Hint 1: This problem can be broken down into three sub-problems you've likely solved before.
- Hint 2: First, find the middle of the list using slow-fast pointers (LC 876).
- Hint 3: Second, reverse the second half of the list (LC 206).
- Hint 4: Third, merge the two halves alternately: take one node from first half, one from second half, repeat.

## Solution Overview
The solution uses a three-phase approach: (1) Find the middle using slow-fast pointers, (2) Reverse the second half, (3) Merge the first and reversed second halves alternately. This achieves O(n) time with O(1) space by combining fundamental linked list techniques.

## Detailed Explanation
**Three-Phase Reordering:**

**Phase 1 - Find Middle:**
1. Use slow-fast pointer technique.
2. When fast reaches the end, slow is at the middle.
3. For odd-length lists: slow points to the true middle.
4. For even-length lists: slow points to the first node of the second half.

**Phase 2 - Reverse Second Half:**
1. Split the list: save `second = slow.next`, then set `slow.next = None`.
2. Reverse the second half using standard three-pointer technique.
3. Now we have two lists: first half and reversed second half.

**Phase 3 - Merge Alternately:**
1. Initialize pointers for both halves: `first = head`, `second = reversed_second_half`.
2. While second exists (second half is shorter or equal):
   - Save next pointers: `tmp1 = first.next`, `tmp2 = second.next`.
   - Link: `first.next = second`, `second.next = tmp1`.
   - Advance: `first = tmp1`, `second = tmp2`.
3. The list is now reordered.

**Key Insight:** Breaking a complex problem into simpler sub-problems (find middle, reverse, merge) makes it manageable and testable.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Three-phase (find, reverse, merge) | O(n) | O(1) | Optimal; in-place modification. |
| Stack/Deque | O(n) | O(n) | Store nodes in stack, then interleave; extra space. |
| Recursion | O(n) | O(n) | Recursive merging; call stack depth. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def reorderList(head: Optional[ListNode]) -> None:
    """Reorder list to L0→Ln→L1→Ln-1→... pattern in-place."""
    if not head or not head.next:
        return
    
    # Phase 1: Find the middle of the list.
    slow, fast = head, head
    while fast.next and fast.next.next:
        slow = slow.next
        fast = fast.next.next
    
    # Phase 2: Reverse the second half.
    second = slow.next
    slow.next = None  # Split the list.
    
    # Reverse second half.
    prev = None
    while second:
        tmp = second.next
        second.next = prev
        prev = second
        second = tmp
    second = prev  # New head of reversed second half.
    
    # Phase 3: Merge two halves alternately.
    first = head
    while second:
        tmp1, tmp2 = first.next, second.next
        first.next = second
        second.next = tmp1
        first, second = tmp1, tmp2
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4], [1, 4, 2, 3]),           # Even length
    ([1, 2, 3, 4, 5], [1, 5, 2, 4, 3]),     # Odd length
    ([1, 2], [1, 2]),                        # Two nodes
    ([1], [1]),                              # Single node
    ([1, 2, 3], [1, 3, 2]),                  # Three nodes
]

for values, expected in test_cases:
    head = create_linked_list(values)
    reorderList(head)
    output = linked_list_to_list(head)
    assert output == expected, f"Failed for {values}: expected {expected}, got {output}"

print('All tests passed for LC 143.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. Finding middle is O(n/2), reversing second half is O(n/2), merging is O(n/2), totaling O(n).
- Space Complexity: O(1) as we only use a constant number of pointers regardless of input size.
- Primary Bottleneck: Must traverse the list multiple times (find middle, reverse, merge), but all are linear, making this optimal.

## Edge Cases & Pitfalls
- Single node or two nodes: handled by early return or minimal merging.
- Must split the list (slow.next = None) before reversing to avoid cycles.
- The merging loop terminates when second becomes None (second half is shorter for odd-length lists).
- Using `fast.next and fast.next.next` ensures slow stops at the right position for both odd and even length lists.

## Follow-up Variants
- Reorder in groups of three: L0→Ln→Ln-1→L1→Ln-2→Ln-3→...
- Reverse only alternate segments before merging.
- Generalize to k-way interleaving from k different segments.

## Takeaways
- Complex linked list problems often decompose into simpler sub-problems: find middle, reverse, merge.
- In-place modification with O(1) space is achievable by carefully managing pointers across multiple phases.
- Saving temporary next pointers before modification is crucial to avoid losing references during merging.

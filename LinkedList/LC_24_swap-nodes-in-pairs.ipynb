## Problem Statement
Given a linked list, swap every two adjacent nodes and return its head. You must solve the problem without modifying the values in the list's nodes (i.e., only nodes themselves may be changed).

Input: `head` - the head node of a singly linked list.
Output: The head of the modified linked list with adjacent pairs swapped.
Constraints: The number of nodes is in the range `[0, 100]`. `0 <= Node.val <= 100`.

## Progressive Hints
- Hint 1: For nodes A → B → C → D, you want B → A → D → C. Focus on swapping the first pair, then recursively handle the rest.
- Hint 2: Use a dummy node to simplify handling the first pair swap.
- Hint 3: For each pair, you need to track: the node before the pair, the first node of the pair, the second node of the pair, and the rest of the list.
- Hint 4: The key operations: prev.next = second, first.next = second.next, second.next = first.

## Solution Overview
Two approaches exist: iterative with pointer manipulation and recursive. The iterative method uses a dummy node and swaps pairs in place by carefully updating three pointers. The recursive method swaps the first pair, then recursively processes the rest. Both achieve O(n) time; iterative uses O(1) space while recursive uses O(n) stack space.

## Detailed Explanation
**Iterative Approach:**
1. Create a dummy node pointing to head for uniform handling.
2. Initialize `prev = dummy` (node before the current pair).
3. While there are at least two nodes to swap:
   - Identify `first = prev.next` and `second = first.next`.
   - Swap them: `prev.next = second`, `first.next = second.next`, `second.next = first`.
   - Move prev to first (now the second node in the swapped pair): `prev = first`.
4. Return `dummy.next`.

**Recursive Approach:**
1. Base case: if head is None or head.next is None, return head (nothing to swap).
2. Identify first = head and second = head.next.
3. Recursively swap the rest: first.next = swapPairs(second.next).
4. Swap current pair: second.next = first.
5. Return second as the new head of this segment.

**Key Insight:** The swap requires three pointer updates in a specific order to avoid losing references.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Iterative | O(n) | O(1) | Optimal space; in-place swapping. |
| Recursive | O(n) | O(n) | Elegant but uses call stack; depth = n/2. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def swapPairs(head: Optional[ListNode]) -> Optional[ListNode]:
    """Swap every two adjacent nodes iteratively."""
    dummy = ListNode(0, head)
    prev = dummy
    
    while prev.next and prev.next.next:
        # Identify the pair to swap.
        first = prev.next
        second = first.next
        
        # Swap the pair.
        prev.next = second
        first.next = second.next
        second.next = first
        
        # Move prev to the end of the swapped pair.
        prev = first
    
    return dummy.next


def swapPairsRecursive(head: Optional[ListNode]) -> Optional[ListNode]:
    """Swap every two adjacent nodes recursively."""
    # Base case: no pair to swap.
    if not head or not head.next:
        return head
    
    first = head
    second = head.next
    
    # Recursively swap the rest.
    first.next = swapPairsRecursive(second.next)
    
    # Swap current pair.
    second.next = first
    
    return second  # New head of this segment.
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4], [2, 1, 4, 3]),      # Even number of nodes
    ([1, 2, 3], [2, 1, 3]),            # Odd number of nodes
    ([1], [1]),                         # Single node
    ([], []),                           # Empty list
    ([1, 2], [2, 1]),                   # Two nodes
]

for values, expected in test_cases:
    # Test iterative
    head = create_linked_list(values)
    result = swapPairs(head)
    assert linked_list_to_list(result) == expected, f"Iterative failed for {values}"
    
    # Test recursive
    head = create_linked_list(values)
    result = swapPairsRecursive(head)
    assert linked_list_to_list(result) == expected, f"Recursive failed for {values}"

print('All tests passed for LC 24.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. We visit each node exactly once.
- Space Complexity: O(1) for iterative, O(n/2) = O(n) for recursive due to call stack depth.
- Primary Bottleneck: Must visit each node to perform swaps, which is unavoidable.

## Edge Cases & Pitfalls
- Empty list should return None.
- Single node should return itself unchanged.
- Odd number of nodes: the last node remains in place.
- Must perform pointer updates in the correct order to avoid losing references or creating cycles.
- The condition `prev.next and prev.next.next` ensures we have a pair to swap.

## Takeaways
- Node swapping requires careful tracking of multiple pointers: the node before the pair, both nodes in the pair, and the node after.
- Dummy nodes simplify edge case handling when the head might be modified.
- The order of pointer updates matters: assign in the wrong order and you'll lose references or create cycles.

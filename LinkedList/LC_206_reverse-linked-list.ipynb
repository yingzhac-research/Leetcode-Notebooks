## Problem Statement
Given the `head` of a singly linked list, reverse the list, and return the reversed list.

Input: `head` - the head node of a singly linked list with values in the range `[-5000, 5000]`.
Output: The head of the reversed linked list.
Constraints: The number of nodes in the list is in the range `[0, 5000]`.

## Progressive Hints
- Hint 1: Consider what happens when you traverse the list once and try to change the direction of each pointer.
- Hint 2: You need to track three nodes at once: previous, current, and next to avoid losing references.
- Hint 3: For the iterative approach, use three pointers; for recursive, think about reversing the rest of the list first, then fixing the current node's pointers.

## Solution Overview
Two main approaches exist: iterative with three pointers and recursive. The iterative method scans the list once while reversing each node's `next` pointer. The recursive method reverses the sublist starting from the second node, then fixes the current node's connections. Both achieve O(n) time with different space trade-offs.

## Detailed Explanation
**Iterative Approach (3 pointers):**
1. Initialize `prev = None` (new tail), `current = head` (traverse pointer), and `next_node = None` (temporary storage).
2. While `current` is not None:
   - Save `next_node = current.next` to preserve the rest of the list.
   - Reverse the link: `current.next = prev`.
   - Move pointers forward: `prev = current`, `current = next_node`.
3. Return `prev` as the new head (when `current` becomes None, `prev` points to the last node).

**Recursive Approach:**
1. Base case: if `head` is None or `head.next` is None, return `head`.
2. Recursively reverse the rest: `new_head = reverseList(head.next)`.
3. Fix pointers: `head.next.next = head` (make next node point back), `head.next = None` (break forward link).
4. Return `new_head` (unchanged through all recursive calls).

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Iterative (3 pointers) | O(n) | O(1) | Best space efficiency; in-place reversal. |
| Recursive | O(n) | O(n) | Call stack depth equals list length; elegant but uses more memory. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def reverseList(head: Optional[ListNode]) -> Optional[ListNode]:
    """Reverse a linked list iteratively using three pointers."""
    prev = None
    current = head
    
    while current:
        next_node = current.next  # Save next node before breaking the link.
        current.next = prev       # Reverse the link.
        prev = current            # Move prev forward.
        current = next_node       # Move current forward.
    
    return prev  # prev is now the new head.


def reverseListRecursive(head: Optional[ListNode]) -> Optional[ListNode]:
    """Reverse a linked list recursively."""
    if not head or not head.next:
        return head  # Base case: empty list or single node.
    
    new_head = reverseListRecursive(head.next)  # Reverse rest of list.
    head.next.next = head  # Make next node point back to current.
    head.next = None       # Break forward link to avoid cycle.
    
    return new_head  # Return the new head (unchanged).
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4, 5], [5, 4, 3, 2, 1]),
    ([1, 2], [2, 1]),
    ([1], [1]),
    ([], []),
]

for values, expected in test_cases:
    # Test iterative
    head = create_linked_list(values)
    result = reverseList(head)
    assert linked_list_to_list(result) == expected, f"Iterative failed for {values}"
    
    # Test recursive
    head = create_linked_list(values)
    result = reverseListRecursive(head)
    assert linked_list_to_list(result) == expected, f"Recursive failed for {values}"

print('All tests passed for LC 206.')
```

## Complexity Analysis
- Time Complexity: O(n) for both approaches where n is the number of nodes. Each node is visited exactly once.
- Space Complexity: O(1) for iterative (only three pointers), O(n) for recursive (call stack depth).
- Primary Bottleneck: For very long lists (n = 5000), recursive approach may hit stack overflow in languages with limited stack size.

## Edge Cases & Pitfalls
- Empty list (head is None) should return None immediately.
- Single node list should return itself unchanged.
- Must carefully manage three pointers in iterative approach to avoid losing references or creating cycles.
- In recursive approach, forgetting to set `head.next = None` creates a cycle between the last two nodes.

## Problem Statement
Given the `head` of a singly linked list, group all the nodes with odd indices together followed by the nodes with even indices, and return the reordered list.

The first node is considered odd, and the second node is even, and so on. Note that the relative order inside both the even and odd groups should remain as it was in the input.

Input: `head` - the head node of a singly linked list.
Output: The head of the reordered linked list.
Constraints: The number of nodes is in the range `[0, 10^4]`. `-10^6 <= Node.val <= 10^6`. Follow-up: Solve it in O(1) space and O(n) time.

## Progressive Hints
- Hint 1: You need to separate the list into two lists: odd-positioned nodes and even-positioned nodes.
- Hint 2: Use two pointers, 'odd' and 'even', to build two separate chains simultaneously.
- Hint 3: Keep track of the head of the even list so you can connect it to the tail of the odd list at the end.
- Hint 4: Each iteration, odd.next skips one node (going to next odd), and even.next also skips one node (going to next even).

## Solution Overview
The solution uses two pointers to simultaneously build odd and even sublists. Start with odd pointing to the first node and even pointing to the second node. In each iteration, link odd to the next odd node (odd.next.next) and even to the next even node (even.next.next). After the loop, connect the tail of the odd list to the head of the even list. This achieves O(n) time with O(1) space.

## Detailed Explanation
**Two-Pointer Odd-Even Separation:**
1. Handle edge cases: if list is empty or has only one node, return head.
2. Initialize `odd = head` and `even = head.next`. Save `even_head = even` to reconnect later.
3. While even and even.next exist (ensuring we can advance odd):
   - Link odd to next odd node: `odd.next = even.next`.
   - Move odd forward: `odd = odd.next`.
   - Link even to next even node: `even.next = odd.next`.
   - Move even forward: `even = even.next`.
4. Connect odd list to even list: `odd.next = even_head`.
5. Return head (which is still the start of the odd list).

**Key Insight:** By advancing odd and even pointers two steps at a time (odd.next.next and even.next.next), we naturally separate odd and even positioned nodes while preserving their relative order.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Two-pointer separation | O(n) | O(1) | Optimal; in-place reorganization. |
| Array conversion | O(n) | O(n) | Convert to array, rearrange, rebuild; extra space. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def oddEvenList(head: Optional[ListNode]) -> Optional[ListNode]:
    """Group odd-indexed nodes followed by even-indexed nodes."""
    if not head or not head.next:
        return head
    
    odd = head
    even = head.next
    even_head = even  # Save the head of even list for later connection.
    
    # Separate odd and even positioned nodes.
    while even and even.next:
        odd.next = even.next     # Link odd to next odd node.
        odd = odd.next           # Move odd pointer.
        even.next = odd.next     # Link even to next even node.
        even = even.next         # Move even pointer.
    
    # Connect odd list to even list.
    odd.next = even_head
    
    return head
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4, 5], [1, 3, 5, 2, 4]),      # Odd length
    ([2, 1, 3, 5, 6, 4, 7], [2, 3, 6, 7, 1, 5, 4]),  # Odd length
    ([1, 2, 3, 4], [1, 3, 2, 4]),            # Even length
    ([1], [1]),                               # Single node
    ([1, 2], [1, 2]),                         # Two nodes
    ([], []),                                 # Empty list
]

for values, expected in test_cases:
    head = create_linked_list(values)
    result = oddEvenList(head)
    output = linked_list_to_list(result)
    assert output == expected, f"Failed for {values}: expected {expected}, got {output}"

print('All tests passed for LC 328.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. We traverse the list once, visiting each node exactly once.
- Space Complexity: O(1) as we only use two pointers regardless of input size.
- Primary Bottleneck: Must visit each node to reorganize, which is optimal.

## Edge Cases & Pitfalls
- Empty list or single node should return unchanged.
- Two-node list: first node (odd) followed by second node (even), already correct.
- Must save even_head before starting the loop to reconnect at the end.
- The loop condition `while even and even.next` ensures we don't access null pointers.

## Follow-up Variants
- Group nodes by k-index modulo (e.g., indices mod 3 = 0, 1, 2).
- Separate nodes based on value parity (odd values vs even values) instead of position.
- Reverse the even group before appending to the odd group.

## Takeaways
- The pattern of building two separate lists simultaneously with alternating pointers is powerful for partitioning problems.
- Saving the head of the second list before reorganization is crucial for reconnection.
- The two-step advancement (odd.next.next) naturally maintains relative order within each group.

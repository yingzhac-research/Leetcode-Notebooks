## Problem Statement
Given the `head` of a linked list, return the list after sorting it in ascending order.

Input: `head` - the head of a singly linked list with integer values.
Output: The head of the sorted linked list.
Constraints: The number of nodes is in the range `[0, 5 * 10^4]`, `-10^5 <= Node.val <= 10^5`.

Follow up: Can you sort the linked list in O(n log n) time and O(1) memory (i.e., constant space complexity)?

## Progressive Hints
- Hint 1: Simple approaches like bubble sort are O(n^2). Think about divide-and-conquer algorithms that achieve O(n log n).
- Hint 2: Merge sort works well with linked lists because merging two sorted lists is O(n) and doesn't require random access.
- Hint 3: Use fast-slow pointers to find the middle of the list for splitting. Recursively sort left and right halves, then merge.
- Hint 4: For O(1) space, use bottom-up merge sort instead of top-down recursion to avoid call stack overhead.

## Solution Overview
Merge sort is the optimal algorithm for sorting linked lists. The top-down recursive approach splits the list in half using fast-slow pointers, recursively sorts each half, then merges them. Time complexity is O(n log n) but space is O(log n) due to recursion depth. The bottom-up iterative approach achieves O(1) space by merging progressively larger sublists (size 1, 2, 4, ...) without recursion.

## Detailed Explanation
**Top-Down Merge Sort (Recursive):**
1. Base case: if list has 0 or 1 nodes, return head (already sorted).
2. Find middle using fast-slow pointers: slow moves one step, fast moves two steps.
3. Split list at middle: set `slow.next = None` to separate left and right halves.
4. Recursively sort left half and right half.
5. Merge the two sorted halves using the standard merge procedure for linked lists.

**Merge procedure:**
- Use a dummy node to simplify edge cases.
- Compare heads of two lists, append smaller node to result, advance that list's pointer.
- After one list exhausted, append remainder of other list.
- Return `dummy.next` as merged result.

**Bottom-Up Merge Sort (Iterative, O(1) space):**
1. Start with merge size = 1 (individual nodes).
2. Merge adjacent pairs of sublists of current size.
3. Double merge size and repeat until size >= list length.
4. Uses split and merge helpers but avoids recursion, maintaining O(1) space.

The key insight is that merge sort doesn't require random access, making it ideal for linked lists, unlike quicksort which performs poorly without efficient indexing.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Top-Down Merge Sort | O(n log n) | O(log n) | Recursive call stack; easier to implement. |
| Bottom-Up Merge Sort | O(n log n) | O(1) | Iterative; achieves optimal space but more complex. |
| Bubble/Insertion Sort | O(n^2) | O(1) | Not recommended for large n; simple but inefficient. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def sortList(head: Optional[ListNode]) -> Optional[ListNode]:
    """Sort linked list using top-down merge sort."""
    if not head or not head.next:
        return head  # Base case: 0 or 1 node
    
    # Find middle using fast-slow pointers
    slow, fast = head, head.next
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    # Split list at middle
    mid = slow.next
    slow.next = None
    
    # Recursively sort both halves
    left = sortList(head)
    right = sortList(mid)
    
    # Merge sorted halves
    return merge(left, right)


def merge(l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
    """Merge two sorted linked lists."""
    dummy = ListNode(0)
    current = dummy
    
    while l1 and l2:
        if l1.val < l2.val:
            current.next = l1
            l1 = l1.next
        else:
            current.next = l2
            l2 = l2.next
        current = current.next
    
    # Append remaining nodes
    current.next = l1 if l1 else l2
    
    return dummy.next


def sortListBottomUp(head: Optional[ListNode]) -> Optional[ListNode]:
    """Sort linked list using bottom-up merge sort (O(1) space)."""
    if not head or not head.next:
        return head
    
    # Get list length
    length = 0
    current = head
    while current:
        length += 1
        current = current.next
    
    dummy = ListNode(0, head)
    size = 1
    
    # Merge sublists of increasing sizes
    while size < length:
        tail = dummy
        current = dummy.next
        
        while current:
            left = current
            right = split(left, size)
            current = split(right, size)
            tail = mergeTo(left, right, tail)
        
        size *= 2
    
    return dummy.next


def split(head: Optional[ListNode], n: int) -> Optional[ListNode]:
    """Split first n nodes from list and return the rest."""
    for i in range(n - 1):
        if not head:
            break
        head = head.next
    
    if not head:
        return None
    
    next_head = head.next
    head.next = None
    return next_head


def mergeTo(l1: Optional[ListNode], l2: Optional[ListNode], 
            tail: ListNode) -> ListNode:
    """Merge l1 and l2, append to tail, return new tail."""
    current = tail
    
    while l1 and l2:
        if l1.val < l2.val:
            current.next = l1
            l1 = l1.next
        else:
            current.next = l2
            l2 = l2.next
        current = current.next
    
    current.next = l1 if l1 else l2
    
    # Move to the end of merged list
    while current.next:
        current = current.next
    
    return current
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([4, 2, 1, 3], [1, 2, 3, 4]),
    ([-1, 5, 3, 4, 0], [-1, 0, 3, 4, 5]),
    ([], []),
    ([1], [1]),
    ([2, 1], [1, 2]),
    ([5, 4, 3, 2, 1], [1, 2, 3, 4, 5]),
    ([1, 1, 1, 1], [1, 1, 1, 1]),
]

for values, expected in test_cases:
    # Test top-down approach
    head = create_linked_list(values)
    result = sortList(head)
    assert linked_list_to_list(result) == expected, f"Top-down failed for {values}"
    
    # Test bottom-up approach
    head = create_linked_list(values)
    result = sortListBottomUp(head)
    assert linked_list_to_list(result) == expected, f"Bottom-up failed for {values}"

print('All tests passed for LC 148.')
```

## Complexity Analysis
- Time Complexity: O(n log n) for both approaches. Each level of merge sort processes all n nodes, and there are log n levels.
- Space Complexity: O(log n) for top-down (recursion stack depth), O(1) for bottom-up (only using pointers).
- Primary Bottleneck: For very large lists (n = 50,000), the recursive approach may hit stack limits in some environments, making bottom-up preferred despite complexity.

## Edge Cases & Pitfalls
- Empty list or single node should return immediately without sorting.
- List with duplicate values must maintain stability (though not required here).
- When finding middle with fast-slow pointers, initialize fast to head.next to ensure proper splitting for even-length lists.
- In bottom-up approach, carefully handle sublists shorter than current merge size.
- Must properly terminate each split sublist by setting the last node's next to None.

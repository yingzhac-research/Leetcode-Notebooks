## Problem Statement
Given the `head` of a singly linked list, return the middle node of the linked list.

If there are two middle nodes, return the second middle node.

Input: `head` - the head node of a singly linked list.
Output: The middle node of the list.
Constraints: The number of nodes in the list is in the range `[1, 100]`.

## Progressive Hints
- Hint 1: One approach is to count the total nodes, then traverse half that distance, but this requires two passes.
- Hint 2: Consider using two pointers that move at different speeds to find the middle in a single pass.
- Hint 3: If one pointer moves twice as fast as the other, when the fast pointer reaches the end, the slow pointer will be at the middle.

## Solution Overview
Use the fast-slow pointer technique: initialize both pointers at the head. Move slow one step and fast two steps per iteration. When fast reaches the end, slow will be at the middle. This achieves O(n) time with O(1) space in a single pass.

## Detailed Explanation
**Fast-Slow Pointer Approach:**
1. Initialize `slow = head` and `fast = head`.
2. While `fast` and `fast.next` are not None:
   - Move `slow` one step: `slow = slow.next`.
   - Move `fast` two steps: `fast = fast.next.next`.
3. When fast reaches the end (None or last node), slow is at the middle.
4. Return `slow`.

**Why it works:** 
- For odd-length lists (e.g., 1→2→3→4→5): when fast reaches node 5, slow is at node 3 (exact middle).
- For even-length lists (e.g., 1→2→3→4): when fast moves off the end, slow is at node 3 (second middle).

**Alternative (Two-Pass):** Count total nodes in first pass, then traverse n/2 nodes in second pass. This uses two passes but is conceptually simpler.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Fast-slow pointers | O(n) | O(1) | Single pass; optimal for both time and space. |
| Count then traverse | O(n) | O(1) | Two passes; simpler logic but less efficient. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def middleNode(head: Optional[ListNode]) -> Optional[ListNode]:
    """Find the middle node using fast-slow pointers."""
    slow = head
    fast = head
    
    while fast and fast.next:
        slow = slow.next       # Move slow one step.
        fast = fast.next.next  # Move fast two steps.
    
    return slow  # Slow is at the middle when fast reaches the end.


def middleNodeTwoPass(head: Optional[ListNode]) -> Optional[ListNode]:
    """Find the middle node using two-pass approach (count then traverse)."""
    # First pass: count nodes.
    count = 0
    current = head
    while current:
        count += 1
        current = current.next
    
    # Second pass: traverse to middle.
    mid = count // 2
    current = head
    for _ in range(mid):
        current = current.next
    
    return current
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases: (input values, expected middle value)
test_cases = [
    ([1, 2, 3, 4, 5], 3),        # Odd length: exact middle
    ([1, 2, 3, 4, 5, 6], 4),     # Even length: second middle
    ([1], 1),                     # Single node
    ([1, 2], 2),                  # Two nodes: second one
    ([1, 2, 3], 2),               # Three nodes: middle
]

for values, expected_val in test_cases:
    # Test fast-slow approach
    head = create_linked_list(values)
    result = middleNode(head)
    assert result.val == expected_val, f"Fast-slow failed for {values}"
    
    # Test two-pass approach
    head = create_linked_list(values)
    result = middleNodeTwoPass(head)
    assert result.val == expected_val, f"Two-pass failed for {values}"

print('All tests passed for LC 876.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. Fast-slow approach makes one pass with fast traversing at most n/2 actual nodes.
- Space Complexity: O(1) for both approaches (only pointer variables).
- Primary Bottleneck: Must traverse at least half the list to find the middle.

## Edge Cases & Pitfalls
- Single node list returns that node as the middle.
- Two node list returns the second node (as specified in problem).
- Must check both `fast` and `fast.next` before advancing to avoid null pointer errors.
- For even-length lists, ensure returning the second of the two middle nodes.

## Follow-up Variants
- Return the first middle node for even-length lists instead of the second.
- Find the kth node from the end using similar two-pointer technique with a gap.
- Split the list into two halves at the middle (used in merge sort for linked lists).

## Takeaways
- Fast-slow pointer technique efficiently finds the middle in a single pass without counting.
- When fast moves two steps per iteration, slow will be at the halfway point when fast finishes.
- This pattern is fundamental for many linked list algorithms including cycle detection and finding kth elements.

## Problem Statement
Given the `head` of a sorted linked list, delete all nodes that have duplicate numbers, leaving only distinct numbers from the original list. Return the linked list sorted as well.

Input: `head` - the head node of a sorted singly linked list.
Output: The head of the modified linked list with all duplicates completely removed.
Constraints: The number of nodes is in the range `[0, 300]`. `-100 <= Node.val <= 100`. The list is guaranteed to be sorted in ascending order.

## Progressive Hints
- Hint 1: Unlike LC 83 where you keep one copy, here you must remove ALL nodes with duplicate values.
- Hint 2: Use a dummy node because the head itself might be a duplicate and need removal.
- Hint 3: Use a 'prev' pointer to track the last node that will remain in the result. Only update prev.next when you've confirmed the next segment has unique values.
- Hint 4: For each unique value region, check if curr and curr.next have the same value. If yes, skip all nodes with that value. If no, it's unique and can be kept.

## Solution Overview
The solution uses a dummy node and a two-pointer approach. Maintain a 'prev' pointer for the last confirmed unique node and a 'curr' pointer to scan ahead. When duplicates are detected (curr.val == curr.next.val), skip all nodes with that value. When a unique value is found, link it to the result and advance prev. This achieves O(n) time with O(1) space.

## Detailed Explanation
**Dummy Node with Duplicate Detection:**
1. Create a dummy node pointing to head to handle head removal cases.
2. Initialize `prev = dummy` (last confirmed unique node) and `curr = head` (scanner).
3. While curr is not None:
   - Check if curr.next exists and curr.val == curr.next.val (duplicate detected):
     - Save the duplicate value: `dup_val = curr.val`.
     - Skip all nodes with this value: `while curr and curr.val == dup_val: curr = curr.next`.
     - Update prev.next to skip all duplicates: `prev.next = curr`.
   - Otherwise (curr is unique):
     - Move prev forward: `prev = prev.next`.
     - Move curr forward: `curr = curr.next`.
4. Return `dummy.next`.

**Key Insight:** Only advance 'prev' when we confirm the current node is unique. When duplicates are found, we skip them all and keep prev pointing to the last good node.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Two-pointer with dummy | O(n) | O(1) | Optimal; single pass with in-place modification. |
| Hash map counting | O(n) | O(n) | Count frequencies first, then rebuild; extra space. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def deleteDuplicates(head: Optional[ListNode]) -> Optional[ListNode]:
    """Remove all nodes that have duplicate values from a sorted list."""
    dummy = ListNode(0, head)
    prev = dummy
    curr = head
    
    while curr:
        # Check if current value is duplicated.
        if curr.next and curr.val == curr.next.val:
            # Skip all nodes with this duplicate value.
            dup_val = curr.val
            while curr and curr.val == dup_val:
                curr = curr.next
            # Connect prev to the node after all duplicates.
            prev.next = curr
        else:
            # Current node is unique, advance prev.
            prev = prev.next
            curr = curr.next
    
    return dummy.next
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 3, 4, 4, 5], [1, 2, 5]),     # Multiple duplicates
    ([1, 1, 1, 2, 3], [2, 3]),              # Duplicates at start
    ([1, 2, 2], [1]),                        # Duplicates at end
    ([1, 1], []),                            # All duplicates
    ([1, 2, 3], [1, 2, 3]),                  # No duplicates
    ([], []),                                # Empty list
]

for values, expected in test_cases:
    head = create_linked_list(values)
    result = deleteDuplicates(head)
    output = linked_list_to_list(result)
    assert output == expected, f"Failed for {values}: expected {expected}, got {output}"

print('All tests passed for LC 82.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. Each node is visited at most once by the curr pointer.
- Space Complexity: O(1) as we only use two pointers and a dummy node regardless of input size.
- Primary Bottleneck: Must scan through all nodes to identify duplicates, which is unavoidable.

## Edge Cases & Pitfalls
- All nodes are duplicates should return an empty list (None).
- Duplicates at the beginning require the dummy node to handle head removal.
- Must skip ALL occurrences of a duplicate value, not just extras.
- The list is sorted, which simplifies duplicate detection (only check consecutive nodes).

## Takeaways
- Complete duplicate removal is trickier than keeping one copy because you can't confirm uniqueness until scanning ahead.

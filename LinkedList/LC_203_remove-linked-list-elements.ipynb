## Problem Statement
Given the `head` of a linked list and an integer `val`, remove all the nodes of the linked list that have `Node.val == val`, and return the new head.

Input: `head` - the head node of a linked list, and `val` - the value to remove.
Output: Head of the modified linked list with all nodes containing `val` removed.
Constraints: The number of nodes in the list is in the range `[0, 10^4]`. Node values are in the range `[1, 50]`. `val` is in the range `[0, 50]`.

## Progressive Hints
- Hint 1: Consider what happens if the head itself needs to be removed - you'll need to update the head pointer.
- Hint 2: A dummy node can simplify handling the case where the head is removed.
- Hint 3: Traverse with a pointer tracking the previous node so you can skip nodes that match the target value.

## Solution Overview
Use a dummy node to handle edge cases where the head is removed. Traverse the list with a pointer, checking each node's value. When a node matches the target value, skip it by updating the previous node's next pointer. Return dummy.next as the new head.

## Detailed Explanation
**Iterative Approach (Dummy Node):**
1. Create a dummy node with next pointing to head to simplify head removal.
2. Initialize `prev = dummy` and `current = head`.
3. While `current` is not None:
   - If `current.val == val`, skip this node: `prev.next = current.next`.
   - Otherwise, move prev forward: `prev = current`.
   - Always advance current: `current = current.next`.
4. Return `dummy.next` (may be None if all nodes were removed).

**Recursive Approach:**
1. Base case: if head is None, return None.
2. Recursively process the rest: `head.next = removeElements(head.next, val)`.
3. If current head's value equals val, skip it by returning `head.next`.
4. Otherwise, return `head`.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Iterative with dummy node | O(n) | O(1) | Best space efficiency; handles all cases cleanly. |
| Recursive | O(n) | O(n) | Call stack depth equals list length; elegant but uses more memory. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def removeElements(head: Optional[ListNode], val: int) -> Optional[ListNode]:
    """Remove all nodes with the specified value from the linked list."""
    dummy = ListNode(0)  # Dummy node simplifies head removal.
    dummy.next = head
    prev = dummy
    current = head
    
    while current:
        if current.val == val:
            # Skip the current node.
            prev.next = current.next
        else:
            # Move prev forward only when not deleting.
            prev = current
        current = current.next
    
    return dummy.next


def removeElementsRecursive(head: Optional[ListNode], val: int) -> Optional[ListNode]:
    """Remove all nodes with the specified value recursively."""
    if not head:
        return None
    
    # Process the rest of the list.
    head.next = removeElementsRecursive(head.next, val)
    
    # If current node matches, skip it.
    return head.next if head.val == val else head
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 6, 3, 4, 5, 6], 6, [1, 2, 3, 4, 5]),
    ([7, 7, 7, 7], 7, []),
    ([1, 2, 3, 4, 5], 6, [1, 2, 3, 4, 5]),
    ([], 1, []),
    ([1], 1, []),
    ([1, 2, 2, 1], 2, [1, 1]),
]

for values, val, expected in test_cases:
    # Test iterative
    head = create_linked_list(values)
    result = removeElements(head, val)
    assert linked_list_to_list(result) == expected, f"Iterative failed for {values}, val={val}"
    
    # Test recursive
    head = create_linked_list(values)
    result = removeElementsRecursive(head, val)
    assert linked_list_to_list(result) == expected, f"Recursive failed for {values}, val={val}"

print('All tests passed for LC 203.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. Each node is visited exactly once.
- Space Complexity: O(1) for iterative (only pointers), O(n) for recursive (call stack depth).
- Primary Bottleneck: Must traverse entire list to find all matching values.

## Edge Cases & Pitfalls
- Head node may need to be removed; dummy node handles this gracefully.
- All nodes may have the target value; result should be None.
- Empty list should return None.
- Multiple consecutive nodes with target value must all be removed.

## Follow-up Variants
- Remove nodes based on a predicate function instead of a fixed value.
- Remove nodes at specific positions rather than by value.

## Takeaways
- Dummy nodes eliminate special case handling for head removal.
- When deleting nodes, only advance the previous pointer when not deleting.
- Recursive solutions process from the end backward, making it natural to handle head removal.

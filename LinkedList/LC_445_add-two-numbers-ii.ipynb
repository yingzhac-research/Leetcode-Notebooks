## Problem Statement
You are given two non-empty linked lists representing two non-negative integers. The most significant digit comes first and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list. You may assume the two numbers do not contain any leading zero, except the number 0 itself.

Input: `l1` and `l2` - heads of two linked lists representing integers in forward order (most significant digit first).
Output: A linked list representing the sum in forward order.
Constraints: The number of nodes in each list is in the range `[1, 100]`, `0 <= Node.val <= 9`.

Follow-up: Could you solve it without reversing the input lists?

## Progressive Hints
- Hint 1: Addition starts from the rightmost (least significant) digit, but the lists start with the leftmost (most significant) digit.
- Hint 2: Use stacks to reverse the order: push all values onto stacks, then pop to process from right to left.
- Hint 3: Build the result list in reverse order (from right to left) by prepending each new digit to the head.
- Hint 4: Alternative approach: reverse both input lists, add using LC 2 logic, then reverse the result, but stack approach is cleaner.

## Solution Overview
The stack-based approach pushes all digits from both lists onto stacks, then pops and adds digits with carry propagation. Results are prepended to build the answer in correct order. This avoids modifying input lists. Alternatively, reverse both lists, use LC 2's algorithm, and reverse the result, but this modifies input unless we make copies.

## Detailed Explanation
**Stack Approach (Preferred):**
1. Create two stacks and push all values from l1 and l2 respectively.
2. Initialize carry to 0 and result head to None.
3. While either stack is non-empty or carry exists:
   - Pop values from stacks (or use 0 if stack is empty).
   - Calculate sum: `total = val1 + val2 + carry`.
   - Update carry: `carry = total // 10`, digit: `total % 10`.
   - Create new node with digit and prepend to result: `new_node.next = result_head`, `result_head = new_node`.
4. Return result_head (naturally in correct forward order).

**Reverse Approach:**
1. Reverse both input lists.
2. Apply LC 2's addition algorithm (add from heads with carry).
3. Reverse the result list.
4. Restore original lists by reversing them back (if modification is not allowed).

**Key insight:** Stacks effectively reverse the order without modifying the original lists, and prepending nodes naturally builds the result in correct order.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Stack | O(max(m, n)) | O(m + n) | Clean; uses stack space for input values. |
| Reverse lists | O(max(m, n)) | O(1) | Modifies input unless making copies; complex restoration. |
| Recursion | O(max(m, n)) | O(max(m, n)) | Call stack space; elegant but harder to manage carry. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def addTwoNumbers(l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
    """Add two numbers in forward order using stacks."""
    # Push all values onto stacks
    stack1, stack2 = [], []
    
    current = l1
    while current:
        stack1.append(current.val)
        current = current.next
    
    current = l2
    while current:
        stack2.append(current.val)
        current = current.next
    
    # Process from right to left, building result by prepending
    carry = 0
    result_head = None
    
    while stack1 or stack2 or carry:
        val1 = stack1.pop() if stack1 else 0
        val2 = stack2.pop() if stack2 else 0
        
        total = val1 + val2 + carry
        carry = total // 10
        digit = total % 10
        
        # Prepend new node to result
        new_node = ListNode(digit)
        new_node.next = result_head
        result_head = new_node
    
    return result_head


def addTwoNumbersReverse(l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
    """Add two numbers by reversing lists (modifies input)."""
    # Reverse both lists
    l1_rev = reverse_list(l1)
    l2_rev = reverse_list(l2)
    
    # Add using LC 2 logic
    dummy = ListNode(0)
    current = dummy
    carry = 0
    
    while l1_rev or l2_rev or carry:
        val1 = l1_rev.val if l1_rev else 0
        val2 = l2_rev.val if l2_rev else 0
        
        total = val1 + val2 + carry
        carry = total // 10
        digit = total % 10
        
        current.next = ListNode(digit)
        current = current.next
        
        l1_rev = l1_rev.next if l1_rev else None
        l2_rev = l2_rev.next if l2_rev else None
    
    # Reverse result to get correct order
    return reverse_list(dummy.next)


def reverse_list(head: Optional[ListNode]) -> Optional[ListNode]:
    """Helper to reverse a linked list."""
    prev = None
    current = head
    
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    
    return prev
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([7, 2, 4, 3], [5, 6, 4], [7, 8, 0, 7]),  # 7243 + 564 = 7807
    ([2, 4, 3], [5, 6, 4], [8, 0, 7]),  # 243 + 564 = 807
    ([0], [0], [0]),  # 0 + 0 = 0
    ([9, 9, 9], [1], [1, 0, 0, 0]),  # 999 + 1 = 1000
    ([5], [5], [1, 0]),  # 5 + 5 = 10
    ([1], [9, 9, 9], [1, 0, 0, 0]),  # 1 + 999 = 1000
]

for l1_vals, l2_vals, expected in test_cases:
    # Test stack approach
    l1 = create_linked_list(l1_vals)
    l2 = create_linked_list(l2_vals)
    result = addTwoNumbers(l1, l2)
    assert linked_list_to_list(result) == expected, f"Stack approach failed for {l1_vals} + {l2_vals}"
    
    # Test reverse approach
    l1 = create_linked_list(l1_vals)
    l2 = create_linked_list(l2_vals)
    result = addTwoNumbersReverse(l1, l2)
    assert linked_list_to_list(result) == expected, f"Reverse approach failed for {l1_vals} + {l2_vals}"

print('All tests passed for LC 445.')
```

## Complexity Analysis
- Time Complexity: O(max(m, n)) for both approaches where m and n are the lengths of the input lists. Stack approach makes two passes (push then pop), reverse approach makes three passes (reverse, add, reverse).
- Space Complexity: O(m + n) for stack approach (storing all input values), O(1) for reverse approach (if modifying input is allowed).
- Primary Bottleneck: For 100-digit numbers, stack approach doubles space temporarily but avoids modifying input, which is often preferred in practice.

## Edge Cases & Pitfalls
- Lists of different lengths handled by treating empty stack as 0.
- Final carry must be added as a new node at the front (most significant position).
- When prepending nodes, ensure new_node.next = result_head, then result_head = new_node.
- Reverse approach modifies input lists unless you reverse them back, which adds complexity.
- Both lists guaranteed to have at least one node (per constraints).

## Problem Statement
You are given the heads of two sorted linked lists `list1` and `list2`.

Merge the two lists into one sorted list. The list should be made by splicing together the nodes of the first two lists.

Return the head of the merged linked list.

Input: `list1` and `list2` - heads of two sorted linked lists.
Output: Head of the merged sorted linked list.
Constraints: The number of nodes in both lists is in the range `[0, 50]`. Both lists are sorted in non-decreasing order.

## Progressive Hints
- Hint 1: Compare the values at the heads of both lists and pick the smaller one to add to the result.
- Hint 2: Use a dummy node to simplify edge cases and avoid special handling for the first node.
- Hint 3: After one list is exhausted, directly attach the remainder of the other list since it's already sorted.

## Solution Overview
Use a dummy node as the starting point and maintain a tail pointer. Compare the current nodes of both lists, attach the smaller one to the tail, and advance that list's pointer. When one list is exhausted, attach the remaining nodes of the other list. Both iterative and recursive solutions achieve O(n + m) time.

## Detailed Explanation
**Iterative Approach (Dummy Node):**
1. Create a dummy node to serve as the predecessor of the result list.
2. Initialize a `tail` pointer to the dummy node.
3. While both `list1` and `list2` are not None:
   - Compare `list1.val` and `list2.val`.
   - Attach the smaller node to `tail.next` and advance that list's pointer.
   - Move `tail` forward.
4. After the loop, one list may have remaining nodes; attach them to `tail.next`.
5. Return `dummy.next` as the head of the merged list.

**Recursive Approach:**
1. Base case: if `list1` is None, return `list2`; if `list2` is None, return `list1`.
2. Compare heads: if `list1.val < list2.val`, set `list1.next = mergeTwoLists(list1.next, list2)` and return `list1`.
3. Otherwise, set `list2.next = mergeTwoLists(list1, list2.next)` and return `list2`.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Iterative with dummy node | O(n + m) | O(1) | Best space efficiency; single pass through both lists. |
| Recursive | O(n + m) | O(n + m) | Call stack depth equals total number of nodes; elegant but uses more memory. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def mergeTwoLists(list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
    """Merge two sorted linked lists iteratively using a dummy node."""
    dummy = ListNode(0)  # Dummy node simplifies edge cases.
    tail = dummy
    
    while list1 and list2:
        if list1.val < list2.val:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next
        tail = tail.next
    
    # Attach remaining nodes from the non-empty list.
    tail.next = list1 if list1 else list2
    
    return dummy.next


def mergeTwoListsRecursive(list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
    """Merge two sorted linked lists recursively."""
    if not list1:
        return list2
    if not list2:
        return list1
    
    if list1.val < list2.val:
        list1.next = mergeTwoListsRecursive(list1.next, list2)
        return list1
    else:
        list2.next = mergeTwoListsRecursive(list1, list2.next)
        return list2
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 4], [1, 3, 4], [1, 1, 2, 3, 4, 4]),
    ([], [], []),
    ([], [0], [0]),
    ([1, 3, 5], [2, 4, 6], [1, 2, 3, 4, 5, 6]),
    ([1], [2], [1, 2]),
]

for list1_vals, list2_vals, expected in test_cases:
    # Test iterative
    list1 = create_linked_list(list1_vals)
    list2 = create_linked_list(list2_vals)
    result = mergeTwoLists(list1, list2)
    assert linked_list_to_list(result) == expected, f"Iterative failed for {list1_vals}, {list2_vals}"
    
    # Test recursive
    list1 = create_linked_list(list1_vals)
    list2 = create_linked_list(list2_vals)
    result = mergeTwoListsRecursive(list1, list2)
    assert linked_list_to_list(result) == expected, f"Recursive failed for {list1_vals}, {list2_vals}"

print('All tests passed for LC 21.')
```

## Complexity Analysis
- Time Complexity: O(n + m) where n and m are the lengths of the two lists. Each node is visited exactly once.
- Space Complexity: O(1) for iterative (only pointers), O(n + m) for recursive (call stack depth).
- Primary Bottleneck: The total number of comparisons is bounded by n + m.

## Edge Cases & Pitfalls
- One or both lists may be empty; handle by returning the non-empty list directly.
- Lists may have different lengths; ensure remaining nodes are properly attached.
- Duplicate values across lists should maintain sorted order in the result.
- Dummy node technique avoids special handling for the first node.

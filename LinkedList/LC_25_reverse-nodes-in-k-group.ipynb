## Problem Statement
Given the `head` of a linked list, reverse the nodes of the list `k` at a time, and return the modified list. `k` is a positive integer and is less than or equal to the length of the linked list. If the number of nodes is not a multiple of `k` then left-out nodes, in the end, should remain as is. You may not alter the values in the list's nodes, only nodes themselves may be changed.

Input: `head` - the head of a singly linked list, `k` - the group size for reversal.
Output: The head of the modified linked list with k-group reversals.
Constraints: The number of nodes is in the range `[1, 5000]`, `0 <= Node.val <= 1000`, `1 <= k <= n` (where n is the number of nodes).

## Progressive Hints
- Hint 1: First, check if there are at least k nodes remaining. If not, return head unchanged (base case).
- Hint 2: Reverse the first k nodes using standard reversal with three pointers. Track the new head of this reversed segment.
- Hint 3: After reversing k nodes, the original first node becomes the tail. Recursively process the rest of the list and connect the tail to the result.
- Hint 4: Use a dummy node before head to simplify edge cases. Track group boundaries carefully: prev (before group), start (group head), end (group tail), next (after group).

## Solution Overview
Two approaches exist: iterative and recursive. Both first verify k nodes exist, then reverse that segment, and continue with remaining nodes. Iterative uses a dummy node and tracks group boundaries (prev, start, end, next) to connect reversed segments. Recursive elegantly reverses the first k nodes, recursively processes the rest, and connects them. Both achieve O(n) time with different space complexity.

## Detailed Explanation
**Recursive Approach:**
1. Count if at least k nodes remain. If not, return head (base case).
2. Reverse the first k nodes using three-pointer technique:
   - Initialize prev=None, current=head.
   - For k iterations: save next, reverse link, advance pointers.
   - After loop, current points to (k+1)th node, prev is new head of reversed segment.
3. The original head is now the tail of reversed segment. Recursively reverse rest: `head.next = reverseKGroup(current, k)`.
4. Return prev (new head of this segment).

**Iterative Approach:**
1. Use a dummy node before head. Initialize group_prev = dummy.
2. While there are at least k nodes:
   - Identify group boundaries: group_start, group_end (k-1 steps from start), group_next (after end).
   - Reverse nodes from group_start to group_end.
   - Connect: `group_prev.next = group_end` (new head), `group_start.next = group_next` (new tail).
   - Update group_prev to group_start for next iteration.
3. Return dummy.next.

**Reversal subroutine for segment [start, end]:**
```python
prev, current = None, start
while current != end.next:  # Stop after end
    next_node = current.next
    current.next = prev
    prev = current
    current = next_node
return prev  # New head of reversed segment
```

**Key insights:**
- Must verify k nodes exist before reversing to handle remainder correctly.
- After reversing k nodes, the original first becomes last (tail of segment).
- Dummy node simplifies connecting the first reversed segment to the result.
- Iterative approach requires careful pointer tracking but avoids recursion stack.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Recursive | O(n) | O(n/k) | Recursion depth is number of groups; elegant but uses stack. |
| Iterative | O(n) | O(1) | Only uses pointers; more complex logic but optimal space. |
| In-place reversal | O(n) | O(1) | Core subroutine for both approaches. |

## Reference Implementation

````python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def reverseKGroup(head: Optional[ListNode], k: int) -> Optional[ListNode]:
    """Reverse nodes in k-group using recursive approach."""
    # Check if at least k nodes remain
    current = head
    count = 0
    while current and count < k:
        current = current.next
        count += 1
    
    if count < k:
        return head  # Less than k nodes, don't reverse
    
    # Reverse first k nodes
    prev = None
    current = head
    for _ in range(k):
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    
    # prev is new head, current is (k+1)th node
    # head is now tail of reversed segment
    # Recursively reverse rest and connect
    head.next = reverseKGroup(current, k)
    
    return prev  # New head of this segment


def reverseKGroupIterative(head: Optional[ListNode], k: int) -> Optional[ListNode]:
    """Reverse nodes in k-group using iterative approach."""
    if not head or k == 1:
        return head
    
    dummy = ListNode(0, head)
    group_prev = dummy
    
    while True:
        # Check if at least k nodes remain
        kth = get_kth_node(group_prev, k)
        if not kth:
            break  # Less than k nodes remaining
        
        # Save group boundaries
        group_next = kth.next
        
        # Reverse k nodes
        prev, current = kth.next, group_prev.next
        while current != group_next:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node
        
        # Connect reversed segment
        tmp = group_prev.next  # Original group start (now tail)
        group_prev.next = kth  # Connect to new head (reversed)
        group_prev = tmp  # Move to next group
    
    return dummy.next


def get_kth_node(current: ListNode, k: int) -> Optional[ListNode]:
    """Get the kth node from current (current is 0th)."""
    while current and k > 0:
        current = current.next
        k -= 1
    return current


def reverseKGroupClean(head: Optional[ListNode], k: int) -> Optional[ListNode]:
    """Alternative iterative with cleaner segment reversal."""
    dummy = ListNode(0, head)
    group_prev = dummy
    
    while True:
        # Find kth node from group_prev
        kth = group_prev
        for _ in range(k):
            kth = kth.next
            if not kth:
                return dummy.next  # Less than k nodes
        
        # Reverse segment [group_prev.next, kth]
        group_start = group_prev.next
        group_next = kth.next
        
        prev, current = group_next, group_start
        while current != group_next:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node
        
        # Connect reversed segment
        group_prev.next = kth
        group_prev = group_start
    
    return dummy.next
````

## Validation

````python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4, 5], 2, [2, 1, 4, 3, 5]),  # Groups of 2, last node unchanged
    ([1, 2, 3, 4, 5], 3, [3, 2, 1, 4, 5]),  # Groups of 3, last 2 unchanged
    ([1, 2, 3, 4, 5], 1, [1, 2, 3, 4, 5]),  # k=1, no change
    ([1], 1, [1]),  # Single node
    ([1, 2], 2, [2, 1]),  # Exactly k nodes
    ([1, 2, 3, 4, 5, 6, 7, 8], 3, [3, 2, 1, 6, 5, 4, 7, 8]),  # Multiple groups
    ([1, 2, 3, 4], 4, [4, 3, 2, 1]),  # Entire list reversed
    ([1, 2, 3, 4, 5], 5, [5, 4, 3, 2, 1]),  # Entire list reversed
]

for values, k, expected in test_cases:
    # Test recursive approach
    head = create_linked_list(values)
    result = reverseKGroup(head, k)
    assert linked_list_to_list(result) == expected, f"Recursive failed for {values} with k={k}"
    
    # Test iterative approach
    head = create_linked_list(values)
    result = reverseKGroupIterative(head, k)
    assert linked_list_to_list(result) == expected, f"Iterative failed for {values} with k={k}"
    
    # Test clean iterative
    head = create_linked_list(values)
    result = reverseKGroupClean(head, k)
    assert linked_list_to_list(result) == expected, f"Clean iterative failed for {values} with k={k}"

print('All tests passed for LC 25.')
````

## Complexity Analysis
- Time Complexity: O(n) for both approaches where n is the number of nodes. Each node is visited a constant number of times (check + reverse).
- Space Complexity: O(n/k) for recursive (recursion depth equals number of groups), O(1) for iterative (only using pointers).
- Primary Bottleneck: For k=2 and n=5000, recursive uses O(2500) stack frames. For k=5000 and n=5000, only one recursion. Iterative always uses O(1) space.

## Edge Cases & Pitfalls
- When k=1, no reversal needed, return head immediately.
- When remaining nodes < k, leave them unchanged (don't reverse partial group).
- Must verify k nodes exist before starting reversal to avoid incomplete groups.
- After reversing k nodes, original first node becomes tail, must connect to next segment.
- Dummy node critical for iterative approach to handle first group consistently.
- In iterative reversal loop, must stop at correct boundary (kth.next or group_next).
- When k equals list length, entire list is reversed (special case of general solution).

## Follow-up Variants
- Reverse nodes in k-group from right to left (reverse last k, then previous k, etc.).
- Reverse alternating k groups (reverse first k, keep next k, reverse next k, etc.).
- Generalize to reverse nodes in groups of varying sizes [k1, k2, k3, ...].
- Implement in-place reversal for doubly linked list updating both next and prev pointers.

## Takeaways
- Always verify sufficient nodes exist before group operations to handle remainders correctly.
- Track four critical pointers for each group: prev (before), start (first), end (kth), next (after).
- After reversing a segment, the original start becomes the tail (crucial for connecting segments).
- Dummy node pattern simplifies boundary cases when the first segment needs special handling.
- Recursive solution is elegant but iterative achieves O(1) space, important for production code.

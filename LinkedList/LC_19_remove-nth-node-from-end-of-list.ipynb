## Problem Statement
Given the `head` of a linked list, remove the `nth` node from the end of the list and return its head.

Input: `head` - the head node of a singly linked list, `n` - a positive integer representing the position from the end.
Output: The head of the modified linked list.
Constraints: The number of nodes in the list is `sz`, where `1 <= sz <= 30`. `0 <= Node.val <= 100`. `1 <= n <= sz`.

## Progressive Hints
- Hint 1: The naive approach requires two passes: one to count nodes, another to remove. Can you do it in one pass?
- Hint 2: Use two pointers with a gap of n+1 nodes between them. When the fast pointer reaches the end, the slow pointer will be at the node before the target.
- Hint 3: Use a dummy node pointing to head to handle the edge case where you need to remove the head itself.
- Hint 4: Move the fast pointer n+1 steps first, then move both pointers together until fast reaches None.

## Solution Overview
The optimal solution uses the two-pointer technique with a maintained gap. Create a dummy node before head, then use two pointers separated by n+1 nodes. Advance the fast pointer n+1 steps, then move both pointers together until fast reaches None. At this point, slow.next is the node to remove. This achieves removal in a single pass with O(1) extra space.

## Detailed Explanation
**Two-Pointer with Gap Approach:**
1. Create a dummy node pointing to head to handle edge cases uniformly (especially when removing the head).
2. Initialize two pointers: `fast = dummy` and `slow = dummy`.
3. Move `fast` forward by `n+1` steps. This creates a gap of n+1 nodes between slow and fast.
4. Move both `fast` and `slow` forward together until `fast` reaches None.
5. At this point, `slow` is positioned right before the node to remove.
6. Remove the target node: `slow.next = slow.next.next`.
7. Return `dummy.next` as the new head (which handles the case where the original head was removed).

**Why n+1 gap?** We want `slow` to point to the node BEFORE the one we're removing, so we can easily update `slow.next`. If we used gap n, slow would point to the node to remove itself, making deletion awkward.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Two-pass (count then remove) | O(n) | O(1) | Simple but requires two traversals. |
| Two-pointer with gap | O(n) | O(1) | Optimal; single pass with dummy node trick. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def removeNthFromEnd(head: Optional[ListNode], n: int) -> Optional[ListNode]:
    """Remove the nth node from the end of the list using two pointers."""
    # Dummy node to handle edge case of removing head.
    dummy = ListNode(0, head)
    fast = dummy
    slow = dummy
    
    # Move fast pointer n+1 steps ahead to create the gap.
    for _ in range(n + 1):
        fast = fast.next
    
    # Move both pointers until fast reaches the end.
    while fast:
        fast = fast.next
        slow = slow.next
    
    # Remove the nth node from end.
    slow.next = slow.next.next
    
    return dummy.next
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4, 5], 2, [1, 2, 3, 5]),       # Remove 4 (2nd from end)
    ([1], 1, []),                              # Remove only node
    ([1, 2], 1, [1]),                          # Remove last node
    ([1, 2], 2, [2]),                          # Remove first node
    ([1, 2, 3], 3, [2, 3]),                    # Remove head
]

for values, n, expected in test_cases:
    head = create_linked_list(values)
    result = removeNthFromEnd(head, n)
    output = linked_list_to_list(result)
    assert output == expected, f"Failed for {values}, n={n}: expected {expected}, got {output}"

print('All tests passed for LC 19.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. We traverse the list once with the fast pointer moving n+1 steps, then both pointers moving together for the remaining nodes.
- Space Complexity: O(1) as we only use two pointers and a dummy node, regardless of list size.
- Primary Bottleneck: Single pass through the list is optimal; cannot be improved further as we must at least visit each node once.

## Edge Cases & Pitfalls
- Removing the head node (n equals list length) requires the dummy node trick.
- Single-node list should return None after removal.
- Must use gap of n+1, not n, to position slow at the node before the target.
- The problem guarantees n is valid (1 <= n <= sz), so we don't need to handle invalid n.

## Follow-up Variants
- Remove nth node from the beginning (trivial with counter).
- Remove all nodes at positions divisible by k from the end.
- Find the nth node from the end without removing it (same technique, just return slow.next instead of removing).

## Takeaways
- The two-pointer technique with a maintained gap is powerful for accessing relative positions in linked lists.
- Dummy nodes simplify edge case handling, especially when the head might be modified.
- The gap size matters: use n+1 when you need to access the node before your target for deletion operations.

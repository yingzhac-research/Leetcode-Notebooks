## Problem Statement
You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list. You may assume the two numbers do not contain any leading zero, except the number 0 itself.

Input: `l1` and `l2` - heads of two linked lists representing integers in reverse order (least significant digit first).
Output: A linked list representing the sum in reverse order.
Constraints: The number of nodes in each list is in the range `[1, 100]`, `0 <= Node.val <= 9`, numbers do not contain leading zeros.

## Progressive Hints
- Hint 1: Since digits are in reverse order, you can add from the heads directly, just like manual addition starting from the rightmost digit.
- Hint 2: Track a carry variable that propagates to the next digit position. Each sum is (l1.val + l2.val + carry), new digit is sum % 10, new carry is sum // 10.
- Hint 3: Handle lists of different lengths by treating missing nodes as 0.
- Hint 4: Don't forget to add a final node if carry is 1 after processing all digits (e.g., 999 + 1 = 1000).

## Solution Overview
Traverse both lists simultaneously, adding corresponding digits plus any carry from the previous position. Create new nodes for each digit of the result. Continue until both lists are exhausted and carry is 0. The reverse order storage simplifies the implementation since we naturally process from least to most significant digit.

## Detailed Explanation
**Iterative Approach:**
1. Initialize a dummy head for the result list and a carry variable to 0.
2. Use a current pointer starting at dummy to build the result list.
3. While l1, l2, or carry is non-zero:
   - Get digit values: `val1 = l1.val if l1 else 0`, `val2 = l2.val if l2 else 0`.
   - Calculate sum: `total = val1 + val2 + carry`.
   - Create new node with digit: `current.next = ListNode(total % 10)`.
   - Update carry: `carry = total // 10`.
   - Advance pointers: `l1 = l1.next if l1 else None`, same for l2, and `current = current.next`.
4. Return `dummy.next` as the result head.

**Key insights:**
- Reverse order storage means we naturally align digits without needing to pad or reverse.
- Treating missing nodes as 0 elegantly handles different-length lists.
- The loop continues while any of (l1, l2, carry) is truthy, ensuring we don't miss the final carry.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Iterative | O(max(m, n)) | O(max(m, n)) | m, n are list lengths; result list length. |
| Recursive | O(max(m, n)) | O(max(m, n)) | Additional call stack space; less preferred. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def addTwoNumbers(l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
    """Add two numbers represented as linked lists in reverse order."""
    dummy = ListNode(0)
    current = dummy
    carry = 0
    
    # Process while any list has nodes or carry exists
    while l1 or l2 or carry:
        # Get values (0 if node is None)
        val1 = l1.val if l1 else 0
        val2 = l2.val if l2 else 0
        
        # Calculate sum and new carry
        total = val1 + val2 + carry
        carry = total // 10
        digit = total % 10
        
        # Create new node for result
        current.next = ListNode(digit)
        current = current.next
        
        # Advance input pointers
        l1 = l1.next if l1 else None
        l2 = l2.next if l2 else None
    
    return dummy.next


def addTwoNumbersRecursive(l1: Optional[ListNode], l2: Optional[ListNode], 
                           carry: int = 0) -> Optional[ListNode]:
    """Add two numbers recursively."""
    # Base case: both lists exhausted and no carry
    if not l1 and not l2 and carry == 0:
        return None
    
    # Get values
    val1 = l1.val if l1 else 0
    val2 = l2.val if l2 else 0
    
    # Calculate sum
    total = val1 + val2 + carry
    new_carry = total // 10
    digit = total % 10
    
    # Create current node
    result = ListNode(digit)
    
    # Recursively process next nodes
    next1 = l1.next if l1 else None
    next2 = l2.next if l2 else None
    result.next = addTwoNumbersRecursive(next1, next2, new_carry)
    
    return result
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([2, 4, 3], [5, 6, 4], [7, 0, 8]),  # 342 + 465 = 807
    ([0], [0], [0]),  # 0 + 0 = 0
    ([9, 9, 9, 9, 9, 9, 9], [9, 9, 9, 9], [8, 9, 9, 9, 0, 0, 0, 1]),  # Carry propagation
    ([9], [1, 9, 9, 9], [0, 0, 0, 0, 1]),  # Different lengths with carry
    ([1], [9, 9], [0, 0, 1]),  # 1 + 99 = 100
]

for l1_vals, l2_vals, expected in test_cases:
    # Test iterative
    l1 = create_linked_list(l1_vals)
    l2 = create_linked_list(l2_vals)
    result = addTwoNumbers(l1, l2)
    assert linked_list_to_list(result) == expected, f"Iterative failed for {l1_vals} + {l2_vals}"
    
    # Test recursive
    l1 = create_linked_list(l1_vals)
    l2 = create_linked_list(l2_vals)
    result = addTwoNumbersRecursive(l1, l2)
    assert linked_list_to_list(result) == expected, f"Recursive failed for {l1_vals} + {l2_vals}"

print('All tests passed for LC 2.')
```

## Complexity Analysis
- Time Complexity: O(max(m, n)) where m and n are the lengths of the two input lists. We process each digit exactly once.
- Space Complexity: O(max(m, n)) for the result list. The iterative approach uses O(1) auxiliary space, while recursive uses O(max(m, n)) call stack.
- Primary Bottleneck: For very long numbers (100 digits), space for the result list dominates, but this is unavoidable as we must return the sum.

## Edge Cases & Pitfalls
- Lists of different lengths must be handled by treating missing nodes as 0.
- Final carry must be added as a new node (e.g., 999 + 1 requires a final carry node).
- Both input lists contain at least one node (guaranteed by constraints), but result might be longer.
- Maximum digit value is 9, so max sum before carry is 9 + 9 + 1 (carry) = 19, which fits in carry logic.
- Don't forget the loop condition includes carry, not just l1 and l2.

## Problem Statement
Given the `head` of a linked list and a value `x`, partition it such that all nodes with values less than `x` come before nodes with values greater than or equal to `x`.

You should preserve the original relative order of the nodes in each of the two partitions.

Input: `head` - the head node of a singly linked list, `x` - an integer partition value.
Output: The head of the partitioned linked list.
Constraints: The number of nodes is in the range `[0, 200]`. `-100 <= Node.val <= 100`. `-200 <= x <= 200`.

## Progressive Hints
- Hint 1: Create two separate linked lists: one for nodes with values less than x, and one for nodes with values >= x.
- Hint 2: Use dummy nodes for both lists to simplify edge case handling.
- Hint 3: Traverse the original list once, appending each node to the appropriate list based on its value.
- Hint 4: After traversal, connect the tail of the 'less than' list to the head of the 'greater or equal' list.

## Solution Overview
The solution uses two dummy nodes to build two separate lists: 'less' for nodes < x and 'greater' for nodes >= x. Traverse the original list, appending each node to the appropriate list. Finally, connect less's tail to greater's head and ensure greater's tail points to None. This achieves O(n) time with O(1) space (excluding the result list).

## Detailed Explanation
**Two-List Partitioning:**
1. Create two dummy nodes: `less_dummy` and `greater_dummy`.
2. Initialize two tail pointers: `less = less_dummy` and `greater = greater_dummy`.
3. Traverse the original list with a current pointer:
   - If `current.val < x`: append to less list: `less.next = current`, `less = less.next`.
   - Otherwise: append to greater list: `greater.next = current`, `greater = greater.next`.
   - Move current forward: `current = current.next`.
4. After traversal:
   - Set `greater.next = None` to terminate the greater list (important to avoid cycles).
   - Connect the two lists: `less.next = greater_dummy.next`.
5. Return `less_dummy.next` as the new head.

**Key Insight:** By maintaining two separate lists and appending nodes in order, we preserve relative ordering within each partition. The final concatenation ensures proper structure.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Two-list with dummy nodes | O(n) | O(1) | Optimal; single pass, in-place reorganization. |
| Array conversion | O(n) | O(n) | Convert to array, partition, rebuild; extra space. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def partition(head: Optional[ListNode], x: int) -> Optional[ListNode]:
    """Partition list so nodes < x come before nodes >= x."""
    # Create two dummy nodes for the two partitions.
    less_dummy = ListNode(0)
    greater_dummy = ListNode(0)
    
    # Tail pointers for building the two lists.
    less = less_dummy
    greater = greater_dummy
    
    # Traverse the original list and partition.
    current = head
    while current:
        if current.val < x:
            less.next = current
            less = less.next
        else:
            greater.next = current
            greater = greater.next
        current = current.next
    
    # Terminate the greater list to avoid cycles.
    greater.next = None
    
    # Connect the two partitions.
    less.next = greater_dummy.next
    
    return less_dummy.next
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 4, 3, 2, 5, 2], 3, [1, 2, 2, 4, 3, 5]),  # Standard case
    ([2, 1], 2, [1, 2]),                           # Partition at boundary
    ([1], 0, [1]),                                 # Single node, >= x
    ([1], 2, [1]),                                 # Single node, < x
    ([], 5, []),                                   # Empty list
    ([3, 1, 2], 3, [1, 2, 3]),                     # All elements partitioned
]

for values, x, expected in test_cases:
    head = create_linked_list(values)
    result = partition(head, x)
    output = linked_list_to_list(result)
    assert output == expected, f"Failed for {values}, x={x}: expected {expected}, got {output}"

print('All tests passed for LC 86.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. We traverse the list once, examining each node exactly once.
- Space Complexity: O(1) as we only use two dummy nodes and pointers regardless of input size.
- Primary Bottleneck: Must examine each node to determine its partition, which is unavoidable.

## Edge Cases & Pitfalls
- Empty list should return None.
- All nodes less than x: greater list will be empty, but connection still works.
- All nodes >= x: less list will be empty (less_dummy.next will be greater_dummy.next).
- Must set `greater.next = None` to avoid cycles, as the last node in greater might originally point to a node now in the less list.

## Follow-up Variants

## Takeaways
- The two-list approach with dummy nodes is a clean pattern for partitioning based on a criterion.
- Always terminate the last list (greater.next = None) to prevent cycles from the original list structure.
- This technique preserves relative order within each partition, which is crucial for stable partitioning.

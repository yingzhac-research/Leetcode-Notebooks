## Problem Statement
Given the `head` of a singly linked list, return `True` if it is a palindrome or `False` otherwise.

Input: `head` - the head node of a singly linked list.
Output: Boolean indicating whether the list is a palindrome.
Constraints: The number of nodes in the list is in the range `[1, 10^5]`. Node values are in the range `[0, 9]`.

Follow-up: Could you do it in O(n) time and O(1) space?

## Progressive Hints
- Hint 1: The simplest approach is to copy all values to an array and check if the array is a palindrome, but this uses O(n) space.
- Hint 2: To achieve O(1) space, you need to work with the list structure itself without copying.
- Hint 3: Find the middle of the list, reverse the second half, then compare the first half with the reversed second half.

## Solution Overview
The optimal O(1) space solution combines three techniques: (1) use fast-slow pointers to find the middle, (2) reverse the second half of the list, (3) compare the first half with the reversed second half. If all corresponding nodes match, the list is a palindrome.

## Detailed Explanation
**O(1) Space Approach (Find Middle + Reverse + Compare):**
1. Use fast-slow pointers to find the middle of the list:
   - Initialize `slow = head`, `fast = head`.
   - While `fast` and `fast.next` exist, move `slow` one step and `fast` two steps.
   - When done, `slow` is at the middle (or start of second half).

2. Reverse the second half starting from `slow`:
   - Use the three-pointer technique from LC 206.
   - Store the reversed head in `second_half`.

3. Compare first half with reversed second half:
   - Initialize two pointers: `p1 = head`, `p2 = second_half`.
   - While `p2` is not None (second half may be shorter for odd-length lists):
     - If `p1.val != p2.val`, return False.
     - Move both pointers forward.
   - If all values match, return True.

4. (Optional) Restore the list by reversing the second half again.

**O(n) Space Approach (Array):**
Copy all values to an array, then use two pointers (start and end) to check if the array is a palindrome.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Find middle + reverse + compare | O(n) | O(1) | Optimal space; modifies list but can restore. |
| Copy to array | O(n) | O(n) | Simpler logic but uses extra memory for all values. |
| Recursive with global pointer | O(n) | O(n) | Call stack depth; elegant but impractical for large lists. |

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def isPalindrome(head: Optional[ListNode]) -> bool:
    """Check if linked list is a palindrome using O(1) space."""
    if not head or not head.next:
        return True  # Single node or empty list is a palindrome.
    
    # Step 1: Find the middle using fast-slow pointers.
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    # Step 2: Reverse the second half starting from slow.
    prev = None
    current = slow
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    second_half = prev
    
    # Step 3: Compare first half with reversed second half.
    p1 = head
    p2 = second_half
    is_palindrome = True
    while p2:  # Second half may be shorter for odd-length lists.
        if p1.val != p2.val:
            is_palindrome = False
            break
        p1 = p1.next
        p2 = p2.next
    
    # (Optional) Step 4: Restore the list by reversing second half again.
    # prev = None
    # current = second_half
    # while current:
    #     next_node = current.next
    #     current.next = prev
    #     prev = current
    #     current = next_node
    
    return is_palindrome


def isPalindromeArray(head: Optional[ListNode]) -> bool:
    """Check if linked list is a palindrome using array (O(n) space)."""
    values = []
    current = head
    while current:
        values.append(current.val)
        current = current.next
    
    # Check if array is a palindrome using two pointers.
    left, right = 0, len(values) - 1
    while left < right:
        if values[left] != values[right]:
            return False
        left += 1
        right -= 1
    
    return True
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


# Test cases
test_cases = [
    ([1, 2, 2, 1], True),
    ([1, 2, 3, 2, 1], True),
    ([1, 2], False),
    ([1], True),
    ([1, 2, 3, 4, 5], False),
    ([1, 0, 1], True),
    ([1, 1, 1, 1], True),
]

for values, expected in test_cases:
    # Test O(1) space approach
    head = create_linked_list(values)
    result = isPalindrome(head)
    assert result == expected, f"O(1) space failed for {values}"
    
    # Test O(n) space approach
    head = create_linked_list(values)
    result = isPalindromeArray(head)
    assert result == expected, f"O(n) space failed for {values}"

print('All tests passed for LC 234.')
```

## Complexity Analysis
- Time Complexity: O(n) for both approaches. Finding middle takes O(n/2), reversing takes O(n/2), comparing takes O(n/2).
- Space Complexity: O(1) for optimal approach (only pointers), O(n) for array approach (storing all values).
- Primary Bottleneck: Must traverse the entire list to verify palindrome property.

## Edge Cases & Pitfalls
- Single node list is a palindrome by definition.
- Two node list is palindrome only if both values are equal.
- For odd-length lists, the middle node is part of the first half and doesn't need to match anything.
- When reversing the second half, the list structure is modified; optionally restore it after comparison.

## Problem Statement
Given the `head` of a singly linked list and two integers `left` and `right` where `left <= right`, reverse the nodes of the list from position `left` to position `right`, and return the reversed list.

Input: `head` - the head node of a singly linked list, `left` and `right` - 1-indexed positions.
Output: The head of the modified linked list.
Constraints: The number of nodes is in the range `[1, 500]`. `1 <= left <= right <= n` where n is the number of nodes. `-500 <= Node.val <= 500`.

## Progressive Hints
- Hint 1: You need to identify four key nodes: the node before the reversal segment, the first node to reverse, the last node to reverse, and the node after the reversal segment.
- Hint 2: Use a dummy node to handle the case where left = 1 (reversing from the head).
- Hint 3: Navigate to the node just before position `left`, then perform standard reversal for (right - left + 1) nodes.
- Hint 4: After reversing, reconnect the segment: the node before `left` should point to the new start of the reversed segment, and the original start should point to the node after `right`.

## Solution Overview
The solution uses a dummy node and performs reversal in three phases: navigate to the start position, reverse the sublist using the standard three-pointer technique, and reconnect the reversed segment to the rest of the list. This achieves O(n) time with O(1) space by modifying pointers in place.

## Detailed Explanation
**Partial Reversal with Reconnection:**
1. Create a dummy node pointing to head to handle edge cases uniformly.
2. Navigate to the node just before position `left` (call it `pre`). This takes left-1 steps from dummy.
3. Identify `start` as the first node to be reversed (pre.next).
4. Perform standard reversal for exactly (right - left + 1) nodes using three pointers (prev, curr, next):
   - Initialize prev = None, curr = start
   - For each of the (right - left + 1) nodes:
     - Save next = curr.next
     - Reverse: curr.next = prev
     - Move forward: prev = curr, curr = next
5. After reversal:
   - `prev` points to the new start of the reversed segment (was the last node)
   - `curr` points to the node after the reversed segment
   - `start` is now the tail of the reversed segment
6. Reconnect: `pre.next = prev` and `start.next = curr`.
7. Return `dummy.next`.

**Key Insight:** Use the same reversal logic as LC 206, but limit it to a specific range and carefully reconnect boundaries.

## Reference Implementation

```python
from typing import Optional


class ListNode:
    """Definition for singly-linked list."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def reverseBetween(head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
    """Reverse nodes from position left to right in a linked list."""
    if not head or left == right:
        return head
    
    # Dummy node to handle edge cases.
    dummy = ListNode(0, head)
    pre = dummy
    
    # Move to the node just before position left.
    for _ in range(left - 1):
        pre = pre.next
    
    # Start of the reversal segment.
    start = pre.next
    
    # Reverse the segment from left to right.
    prev = None
    curr = start
    for _ in range(right - left + 1):
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node
    
    # Reconnect the reversed segment.
    pre.next = prev      # Connect to new start of reversed segment.
    start.next = curr    # Connect old start (now tail) to node after segment.
    
    return dummy.next
```

## Validation

```python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_list(head):
    """Helper function to convert linked list to Python list for comparison."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Test cases
test_cases = [
    ([1, 2, 3, 4, 5], 2, 4, [1, 4, 3, 2, 5]),  # Reverse middle segment
    ([5], 1, 1, [5]),                           # Single node, no change
    ([1, 2, 3], 1, 2, [2, 1, 3]),              # Reverse from head
    ([1, 2, 3], 2, 3, [1, 3, 2]),              # Reverse to tail
    ([1, 2, 3, 4, 5], 1, 5, [5, 4, 3, 2, 1]),  # Reverse entire list
]

for values, left, right, expected in test_cases:
    head = create_linked_list(values)
    result = reverseBetween(head, left, right)
    output = linked_list_to_list(result)
    assert output == expected, f"Failed for {values}, left={left}, right={right}: expected {expected}, got {output}"

print('All tests passed for LC 92.')
```

## Complexity Analysis
- Time Complexity: O(n) where n is the number of nodes. We traverse at most n nodes: (left-1) to reach the segment, (right-left+1) to reverse it.
- Space Complexity: O(1) as we only use a constant number of pointers regardless of input size.
- Primary Bottleneck: Must traverse to the reversal segment, which is unavoidable. The reversal itself is optimal at one pass.

## Edge Cases & Pitfalls
- When left = 1, we're reversing from the head, which is why we need the dummy node.
- When left = right, no reversal needed (though the code handles it correctly).
- Must carefully track four connection points: pre, start, prev (new head of segment), curr (node after segment).
- Off-by-one errors are common when calculating the number of nodes to reverse: it's (right - left + 1), not (right - left).

## Takeaways
- Partial reversal builds on the standard reversal technique by adding boundary tracking and reconnection.
- Dummy nodes are invaluable when the head might be part of the modified segment.
- Always verify reconnection logic: the segment before must point to the new start, and the old start (now tail) must point to the segment after.

## Problem Statement
Given a graph that started as a tree with `n` nodes (labeled from `1` to `n`) and `n` edges, one additional edge was added. The added edge has two different vertices chosen from `1` to `n`, and was not an edge that already existed.

Return an edge that can be removed so that the resulting graph is a tree of `n` nodes. If there are multiple answers, return the answer that occurs last in the input.

## Progressive Hints
- Hint 1: A tree with n nodes has exactly n-1 edges. The nth edge creates a cycle.
- Hint 2: Process edges in order and use Union-Find to track connected components.
- Hint 3: When union(u, v) fails (both already in same set), that edge creates the cycle.
- Hint 4: Return the first edge that fails to union successfully.

## Solution Overview
Use Union-Find to process edges sequentially. The first edge where both vertices share the same root is the redundant connection that creates a cycle.

## Detailed Explanation
1. Initialize Union-Find with n nodes (1 to n).
2. Iterate through edges in the given order.
3. For each edge [u, v], check if find(u) == find(v).
4. If they share the same root, this edge creates a cycle - return it.
5. Otherwise, union(u, v) to merge their components.
6. Path compression and union by rank ensure efficient operations.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(n α(n)) | O(n) | Optimal for cycle detection |
| DFS for each edge | O(n²) | O(n) | Check cycle after each edge addition |
| BFS | O(n²) | O(n) | Similar to DFS |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n + 1))  # 1-indexed
        self.rank = [0] * (n + 1)
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    
    def union(self, x: int, y: int) -> bool:
        root_x, root_y = self.find(x), self.find(y)
        
        if root_x == root_y:
            return False  # Already connected - this edge creates a cycle
        
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1
        
        return True


def findRedundantConnection(edges: List[List[int]]) -> List[int]:
    n = len(edges)
    uf = UnionFind(n)
    
    for u, v in edges:
        if not uf.union(u, v):
            return [u, v]
    
    return []  # Should never reach here given problem constraints
```

## Validation

```python
assert findRedundantConnection([[1,2],[1,3],[2,3]]) == [2,3]
assert findRedundantConnection([[1,2],[2,3],[3,4],[1,4],[1,5]]) == [1,4]
assert findRedundantConnection([[1,2],[1,3],[2,4],[3,4]]) == [3,4]
assert findRedundantConnection([[1,2]]) == []  # Edge case: only one edge
print('All tests passed for LC 684.')
```

## Complexity Analysis
- Time Complexity: O(n α(n)), where n is the number of edges and α(n) is the inverse Ackermann function.
- Space Complexity: O(n) for the parent and rank arrays.
- Bottleneck: Processing each edge with near-constant union-find operations.

## Edge Cases & Pitfalls
- Multiple possible redundant edges: Return the last one in input order.
- Self-loops: Problem guarantees no self-loops.
- 1-indexed nodes: Initialize parent array with size n+1.
- Minimum input: With n=3, at least 3 edges are needed.

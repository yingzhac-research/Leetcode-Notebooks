## Problem Statement
You are given a string `s`, and an array of pairs of indices in the string `pairs` where `pairs[i] = [a, b]` indicates 2 indices (0-indexed) of the string.

You can swap the characters at any pair of indices in the given `pairs` any number of times.

Return the lexicographically smallest string that `s` can be changed to after using the swaps.

## Progressive Hints
- Hint 1: If positions i and j can swap, they can rearrange freely within their connected component.
- Hint 2: Use Union-Find to group positions that can freely interchange characters.
- Hint 3: For each group, collect all characters and sort them.
- Hint 4: Assign the smallest available character to the smallest position in each group.

## Solution Overview
Use Union-Find to group indices that can swap with each other. For each group, sort the characters and assign them back to positions in sorted order.

## Detailed Explanation
1. Initialize Union-Find with n positions (string length).
2. For each pair [a, b], union positions a and b.
3. Group positions by their root representative.
4. For each group:
   - Collect characters at those positions.
   - Sort both positions and characters.
   - Assign sorted characters to sorted positions.
5. Build the result string from the rearranged characters.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(n*log(n) + E*α(n)) | O(n) | n string length, E edges (pairs) |
| DFS graph | O(n*log(n) + E) | O(n + E) | Build graph then sort groups |
| BFS | O(n*log(n) + E) | O(n + E) | Similar to DFS |

## Reference Implementation

```python
from typing import List
from collections import defaultdict


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [0] * n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    
    def union(self, x: int, y: int) -> None:
        root_x, root_y = self.find(x), self.find(y)
        
        if root_x == root_y:
            return
        
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1


def smallestStringWithSwaps(s: str, pairs: List[List[int]]) -> str:
    n = len(s)
    uf = UnionFind(n)
    
    # Union all swappable positions
    for a, b in pairs:
        uf.union(a, b)
    
    # Group positions by their root
    groups = defaultdict(list)
    for i in range(n):
        root = uf.find(i)
        groups[root].append(i)
    
    # Build result array
    result = list(s)
    
    # For each group, sort characters and assign to sorted positions
    for indices in groups.values():
        # Get characters at these positions and sort them
        chars = sorted([s[i] for i in indices])
        # Sort the indices
        indices.sort()
        # Assign sorted characters to sorted positions
        for i, char in zip(indices, chars):
            result[i] = char
    
    return ''.join(result)
```

## Validation

```python
assert smallestStringWithSwaps("dcab", [[0,3],[1,2]]) == "bacd"
assert smallestStringWithSwaps("dcab", [[0,3],[1,2],[0,2]]) == "abcd"
assert smallestStringWithSwaps("cba", [[0,1],[1,2]]) == "abc"
assert smallestStringWithSwaps("abc", []) == "abc"
assert smallestStringWithSwaps("a", []) == "a"
print('All tests passed for LC 1202.')
```

## Complexity Analysis
- Time Complexity: O(n*log(n) + E*α(n)), where n is string length and E is number of pairs.
- Space Complexity: O(n) for Union-Find and grouping structures.
- Bottleneck: Sorting characters within each group.

## Edge Cases & Pitfalls
- No pairs: String remains unchanged.
- All positions connected: Sort entire string.
- Single character: No change possible.
- Disjoint groups: Each group sorted independently.

## Follow-up Variants
- Largest string with swaps: Sort characters in descending order.
- Count swaps needed: Calculate minimum swaps to achieve result.
- Limited swaps: Constrain total number of swap operations.

## Takeaways
- Union-Find identifies positions that can freely exchange characters.
- Within each connected component, optimal arrangement is sorted order.
- Sorting both indices and characters ensures correct assignment.

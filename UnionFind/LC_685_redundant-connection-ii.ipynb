## Problem Statement
In this problem, a rooted tree is a directed graph such that there is exactly one node (the root) for which all other nodes are descendants of this node, plus every node has exactly one parent, except the root node which has no parents.

The given input is a directed graph that started as a rooted tree with `n` nodes (with distinct values from 1 to n), with one additional directed edge added. The added edge has two different vertices chosen from 1 to n, and was not an edge that already existed.

The resulting graph is given as a 2D array of edges. Each element of edges is a pair `[u, v]` that represents a directed edge connecting nodes `u` and `v`, where `u` is a parent of child `v`.

Return an edge that can be removed so that the resulting graph is a rooted tree of `n` nodes. If there are multiple answers, return the answer that occurs last in the given 2D array.

## Progressive Hints
- Hint 1: There are two scenarios: a node has two parents, or there's a cycle without a node having two parents.
- Hint 2: First check if any node has two parents (in-degree > 1).
- Hint 3: If a node has two parents, try removing each edge leading to it and check if the result is a valid tree.
- Hint 4: Use Union-Find to detect cycles while processing edges.

## Solution Overview
Handle two cases: (1) a node with two parents, or (2) a cycle. Track in-degrees to find nodes with multiple parents. Use Union-Find to detect cycles while considering which edge to remove.

## Detailed Explanation
1. **Find node with two parents**: Scan all edges and track parent count for each node.
2. **Two candidates**: If a node has two parents, mark both edges as candidates.
3. **Process edges with Union-Find**:
   - If there are two candidates, skip the second one initially.
   - Use Union-Find to detect if there's a cycle.
   - If a cycle is detected:
     - With two candidates: return the first candidate.
     - Without two candidates: return the edge that creates the cycle.
4. **No cycle found**: Return the second candidate (if it exists).
5. This approach handles both the "two parents" and "cycle" scenarios.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find with parent tracking | O(n α(n)) | O(n) | Optimal for this problem |
| DFS cycle detection | O(n²) | O(n) | Multiple passes needed |
| Try removing each edge | O(n²) | O(n) | Inefficient brute force |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n + 1))
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    
    def union(self, x: int, y: int) -> bool:
        root_x, root_y = self.find(x), self.find(y)
        if root_x == root_y:
            return False  # Cycle detected
        self.parent[root_y] = root_x
        return True


def findRedundantDirectedConnection(edges: List[List[int]]) -> List[int]:
    n = len(edges)
    parent = {}
    candidates = []
    
    # Step 1: Find the node with two parents
    for u, v in edges:
        if v in parent:
            # Found a node with two parents
            candidates = [[parent[v], v], [u, v]]
            break
        parent[v] = u
    
    # Step 2: Use Union-Find to detect cycle
    uf = UnionFind(n)
    
    for u, v in edges:
        # If we have candidates, skip the second candidate
        if candidates and [u, v] == candidates[1]:
            continue
        
        if not uf.union(u, v):
            # Cycle detected
            if candidates:
                # If there are candidates, the first one creates the cycle
                return candidates[0]
            else:
                # No candidates, this edge creates the cycle
                return [u, v]
    
    # No cycle found, return the second candidate
    return candidates[1]
```

## Validation

```python
assert findRedundantDirectedConnection([[1,2],[1,3],[2,3]]) == [2,3]
assert findRedundantDirectedConnection([[1,2],[2,3],[3,4],[4,1],[1,5]]) == [4,1]
assert findRedundantDirectedConnection([[2,1],[3,1],[4,2],[1,4]]) == [2,1]
assert findRedundantDirectedConnection([[1,2],[2,3],[3,1]]) == [3,1]
print('All tests passed for LC 685.')
```

## Complexity Analysis
- Time Complexity: O(n α(n)), where n is the number of nodes and α(n) is the inverse Ackermann function.
- Space Complexity: O(n) for Union-Find parent array and parent tracking dictionary.
- Bottleneck: Union-Find operations dominate, but are nearly constant time.

## Edge Cases & Pitfalls
- Node with two parents vs cycle: Must handle both scenarios correctly.
- Last occurrence: Problem requires returning the last edge in input order.
- Directed vs undirected: Remember this is a directed graph.
- Root node: The root should have no parents in a valid tree.

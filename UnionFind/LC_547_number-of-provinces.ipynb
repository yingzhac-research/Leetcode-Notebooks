## Problem Statement
Given an `n x n` matrix `isConnected` where `isConnected[i][j] = 1` if the `i`th city and `j`th city are directly connected, and `isConnected[i][j] = 0` otherwise. Return the total number of provinces.

A province is a group of directly or indirectly connected cities with no other cities outside of the group.

## Progressive Hints
- Hint 1: Each city starts as its own province. How can you merge provinces when cities are connected?
- Hint 2: Union-Find can track which cities belong to the same province efficiently.
- Hint 3: The answer is the number of distinct root nodes after all unions are performed.
- Hint 4: Use path compression and union by rank for optimal performance.

## Solution Overview
Use Union-Find to group connected cities. Iterate through the adjacency matrix and union cities that are directly connected. The final count of unique roots equals the number of provinces.

## Detailed Explanation
1. Initialize a Union-Find structure with `n` cities, each as its own parent.
2. For each pair `(i, j)` where `isConnected[i][j] == 1`, call `union(i, j)`.
3. Path compression in `find()` flattens the tree for faster lookups.
4. Union by rank ensures balanced merging.
5. Count distinct roots: cities where `parent[i] == i` after all unions.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(n² α(n)) | O(n) | α(n) is inverse Ackermann, nearly constant |
| DFS | O(n²) | O(n) | Simple but no amortization benefits |
| BFS | O(n²) | O(n) | Similar to DFS |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [0] * n
        self.count = n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]
    
    def union(self, x: int, y: int) -> bool:
        root_x, root_y = self.find(x), self.find(y)
        if root_x == root_y:
            return False
        
        # Union by rank
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1
        
        self.count -= 1
        return True
    
    def get_count(self) -> int:
        return self.count


def findCircleNum(isConnected: List[List[int]]) -> int:
    n = len(isConnected)
    uf = UnionFind(n)
    
    for i in range(n):
        for j in range(i + 1, n):
            if isConnected[i][j] == 1:
                uf.union(i, j)
    
    return uf.get_count()
```

## Validation

```python
assert findCircleNum([[1,1,0],[1,1,0],[0,0,1]]) == 2
assert findCircleNum([[1,0,0],[0,1,0],[0,0,1]]) == 3
assert findCircleNum([[1]]) == 1
assert findCircleNum([[1,0,0,1],[0,1,1,0],[0,1,1,1],[1,0,1,1]]) == 1
print('All tests passed for LC 547.')
```

## Complexity Analysis
- Time Complexity: O(n² α(n)), where α(n) is the inverse Ackermann function (effectively constant).
- Space Complexity: O(n) for the parent and rank arrays.
- Bottleneck: Iterating through all pairs in the adjacency matrix.

## Edge Cases & Pitfalls
- Single city: Should return 1.
- Fully connected graph: Should return 1.
- No connections (diagonal only): Should return n.
- Symmetric matrix property: Only need to check upper triangle.

## Follow-up Variants

- Weighted provinces: Find the largest province by population.

## Takeaways
- Union-Find is ideal for dynamic connectivity problems.
- Path compression and union by rank provide near-constant time operations.
- Maintaining a count variable simplifies tracking the number of components.

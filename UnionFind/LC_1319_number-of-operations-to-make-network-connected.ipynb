## Problem Statement
There are `n` computers numbered from `0` to `n-1` connected by ethernet cables `connections` forming a network where `connections[i] = [a, b]` represents a connection between computers `a` and `b`.

Any computer can reach any other computer directly or indirectly through the network.

Given an initial computer network `connections`, you can extract certain cables between two directly connected computers and place them between any pair of disconnected computers to make them directly connected.

Return the minimum number of times you need to do this in order to make all the computers connected. If it's not possible, return -1.

## Progressive Hints
- Hint 1: To connect n computers, you need at least n-1 cables.
- Hint 2: Extra cables (redundant connections) can be repurposed.
- Hint 3: Count the number of connected components and redundant cables using Union-Find.
- Hint 4: If you have k components, you need k-1 cables to connect them all.

## Solution Overview
Use Union-Find to identify connected components and count redundant cables. If we have enough redundant cables to connect all components, return the number of operations needed (components - 1).

## Detailed Explanation
1. Check if total cables < n-1: impossible to connect all computers, return -1.
2. Initialize Union-Find with n computers.
3. Process each connection:
   - If the two computers are already connected (same root), it's a redundant cable.
   - Otherwise, union them.
4. Count the number of connected components after all unions.
5. To connect k components, we need k-1 operations.
6. Verify we have enough redundant cables: if redundant >= k-1, return k-1.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(E α(n)) | O(n) | α(n) is inverse Ackermann, nearly constant |
| DFS | O(n + E) | O(n) | Count components with graph traversal |
| BFS | O(n + E) | O(n) | Similar to DFS |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [0] * n
        self.components = n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]
    
    def union(self, x: int, y: int) -> bool:
        root_x, root_y = self.find(x), self.find(y)
        if root_x == root_y:
            return False  # Already connected, redundant cable
        
        # Union by rank
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1
        
        self.components -= 1
        return True
    
    def get_components(self) -> int:
        return self.components


def makeConnected(n: int, connections: List[List[int]]) -> int:
    # Need at least n-1 cables to connect n computers
    if len(connections) < n - 1:
        return -1
    
    uf = UnionFind(n)
    redundant = 0
    
    for a, b in connections:
        if not uf.union(a, b):
            redundant += 1
    
    # Number of operations needed = components - 1
    return uf.get_components() - 1
```

## Validation

```python
assert makeConnected(4, [[0,1],[0,2],[1,2]]) == 1
assert makeConnected(6, [[0,1],[0,2],[0,3],[1,2],[1,3]]) == 2
assert makeConnected(6, [[0,1],[0,2],[0,3],[1,2]]) == -1
assert makeConnected(5, [[0,1],[0,2],[3,4],[2,3]]) == 0
assert makeConnected(1, []) == 0
print('All tests passed for LC 1319.')
```

## Complexity Analysis
- Time Complexity: O(E α(n)), where E is the number of connections and α(n) is the inverse Ackermann function (effectively constant).
- Space Complexity: O(n) for the Union-Find data structure.
- Bottleneck: Processing all edges through Union-Find operations.

## Edge Cases & Pitfalls
- Not enough cables: If connections.length < n-1, impossible to connect all.
- Already connected: If only 1 component, return 0.
- Single computer: n=1 with no connections returns 0.
- Multiple redundant cables: Correctly count and utilize all redundant cables.

## Follow-up Variants
- Dynamic network: Handle adding/removing cables and query connectivity in real-time.
- Weighted cables: Minimize total cost when reassigning cables.
- Capacity constraints: Each cable has a bandwidth limit.

## Takeaways
- Union-Find elegantly tracks both connectivity and component count.
- Redundant edges can be identified during union operations.
- Mathematical insight: k components need exactly k-1 edges to become fully connected.
- Early termination check (total cables < n-1) saves computation.

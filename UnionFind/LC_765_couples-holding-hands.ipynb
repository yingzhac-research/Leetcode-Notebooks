## Problem Statement
There are `n` couples sitting in `2n` seats arranged in a row and want to hold hands.

The people and seats are represented by an integer array `row` where `row[i]` is the ID of the person sitting in the `i`th seat. The couples are numbered in order, the first couple being `(0, 1)`, the second couple being `(2, 3)`, and so on with the last couple being `(2n - 2, 2n - 1)`.

Return the minimum number of swaps so that every couple is sitting side by side. A swap consists of choosing any two people, then they stand up and switch seats.

## Progressive Hints
- Hint 1: Think of this as a cycle detection problem where couples form connected components.
- Hint 2: Group people by couch positions (pairs of seats), not individual seats.
- Hint 3: If k couples form a cycle, you need exactly k-1 swaps to separate them.
- Hint 4: Use Union-Find to group couches where couples are split across them.

## Solution Overview
Model the problem as a graph where each couch is a node. Connect couches if a couple is split between them. Use Union-Find to count connected components. For each component of size k, we need k-1 swaps.

## Detailed Explanation
1. **Model as graph**: Each couch (pair of seats) is a node. Couples are numbered 0, 1, 2, ..., n-1.
2. **Connect couches**: For each couch, find which couples are sitting there. If couple i and couple j are split across two couches, those couches are connected.
3. **Union-Find**: Union couches that share split couples.
4. **Count swaps**: For a connected component with k couches, we need exactly k-1 swaps to fix all couples.
5. **Formula**: Total swaps = n - (number of connected components).
6. **Why this works**: A cycle of k couches requires k-1 swaps to break into k separate, correctly paired couches.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(n α(n)) | O(n) | Optimal cycle counting approach |
| Greedy simulation | O(n²) | O(n) | Simulate swaps directly |
| Graph DFS | O(n) | O(n) | Count cycles with traversal |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.components = n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]
    
    def union(self, x: int, y: int) -> bool:
        root_x, root_y = self.find(x), self.find(y)
        if root_x == root_y:
            return False
        self.parent[root_y] = root_x
        self.components -= 1
        return True
    
    def get_components(self) -> int:
        return self.components


def minSwapsCouples(row: List[int]) -> int:
    n = len(row) // 2  # Number of couples/couches
    uf = UnionFind(n)
    
    # Process each couch
    for i in range(0, len(row), 2):
        # Get the two people on this couch
        person1, person2 = row[i], row[i + 1]
        # Determine which couples they belong to
        couple1, couple2 = person1 // 2, person2 // 2
        # Union the couples (connect the couches)
        uf.union(couple1, couple2)
    
    # Total swaps = number of couches - number of connected components
    return n - uf.get_components()
```

## Validation

```python
assert minSwapsCouples([0,2,1,3]) == 1
assert minSwapsCouples([3,2,0,1]) == 0
assert minSwapsCouples([0,2,4,6,7,1,3,5]) == 3
assert minSwapsCouples([5,4,2,6,3,1,0,7]) == 2
assert minSwapsCouples([0,1]) == 0
print('All tests passed for LC 765.')
```

## Complexity Analysis
- Time Complexity: O(n α(n)), where n is the number of couples and α(n) is the inverse Ackermann function.
- Space Complexity: O(n) for the Union-Find data structure.
- Bottleneck: Processing each couch with Union-Find operations.

## Edge Cases & Pitfalls
- Already paired correctly: Should return 0.
- Single couple: Always paired, return 0.
- Complete cycle: All couples form one large cycle.
- Couple numbering: Person i belongs to couple i//2.

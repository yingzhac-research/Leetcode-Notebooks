## Problem Statement
Given a list of `accounts` where each element `accounts[i]` is a list of strings, where the first element `accounts[i][0]` is a name, and the rest of the elements are emails representing emails of the account.

Merge accounts that have common emails. Note that even if two accounts have the same name, they may belong to different people as people could have the same name. A person can have any number of accounts initially, but all of their accounts have the same name.

Return the accounts in the following format: the first element of each account is the name, and the rest are emails in sorted order.

## Progressive Hints
- Hint 1: Use Union-Find to group emails that belong to the same person.
- Hint 2: Map each email to an account index using a hash map.
- Hint 3: Union all emails within the same account, and union accounts with shared emails.
- Hint 4: Group emails by their root representative and reconstruct accounts.

## Solution Overview
Use Union-Find to merge accounts with common emails. Map each email to its account index, union emails within and across accounts, then group all emails by their root representative.

## Detailed Explanation
1. Create Union-Find with n accounts.
2. Use a hash map to track which account each email belongs to.
3. For each email in an account:
   - If email seen before, union current account with previous account.
   - Otherwise, map email to current account.
4. Group emails by the root account index.
5. For each group, sort emails and prepend the account name.
6. Return the merged accounts.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(N*K*α(N) + N*K*log(K)) | O(N*K) | N accounts, K emails per account |
| DFS on graph | O(N*K*log(K)) | O(N*K) | Build email graph then traverse |
| BFS | O(N*K*log(K)) | O(N*K) | Similar to DFS |

## Reference Implementation

```python
from typing import List
from collections import defaultdict


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [0] * n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    
    def union(self, x: int, y: int) -> None:
        root_x, root_y = self.find(x), self.find(y)
        
        if root_x == root_y:
            return
        
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1


def accountsMerge(accounts: List[List[str]]) -> List[List[str]]:
    n = len(accounts)
    uf = UnionFind(n)
    email_to_account = {}
    
    # Map emails to accounts and union accounts with shared emails
    for i, account in enumerate(accounts):
        for email in account[1:]:
            if email in email_to_account:
                uf.union(i, email_to_account[email])
            else:
                email_to_account[email] = i
    
    # Group emails by root account
    account_emails = defaultdict(list)
    for email, account_idx in email_to_account.items():
        root = uf.find(account_idx)
        account_emails[root].append(email)
    
    # Build result
    result = []
    for account_idx, emails in account_emails.items():
        name = accounts[account_idx][0]
        result.append([name] + sorted(emails))
    
    return result
```

## Validation

```python
result1 = accountsMerge([
    ["John","johnsmith@mail.com","john_newyork@mail.com"],
    ["John","johnsmith@mail.com","john00@mail.com"],
    ["Mary","mary@mail.com"],
    ["John","johnnybravo@mail.com"]
])
assert len(result1) == 3

result2 = accountsMerge([
    ["Gabe","Gabe0@m.co","Gabe3@m.co","Gabe1@m.co"],
    ["Kevin","Kevin3@m.co","Kevin5@m.co","Kevin0@m.co"],
    ["Ethan","Ethan5@m.co","Ethan4@m.co","Ethan0@m.co"],
    ["Hanzo","Hanzo3@m.co","Hanzo1@m.co","Hanzo0@m.co"],
    ["Fern","Fern5@m.co","Fern1@m.co","Fern0@m.co"]
])
assert len(result2) == 5

print('All tests passed for LC 721.')
```

## Complexity Analysis
- Time Complexity: O(N*K*α(N) + N*K*log(K)), where N is accounts count and K is average emails per account.
- Space Complexity: O(N*K) for hash maps and Union-Find structure.
- Bottleneck: Sorting emails within each merged account.

## Edge Cases & Pitfalls
- Same name, different people: Only merge accounts with shared emails.
- Single account: Should remain unchanged.
- Transitive merging: A-B and B-C should merge A-B-C.
- Email sorting: Must sort emails lexicographically in the output.

## Follow-up Variants
- Real-time account merging: Support incremental account additions.
- Weighted merging: Prefer certain accounts as primary based on metadata.
- Multi-attribute merging: Merge based on multiple fields (phone, address, etc.).

## Takeaways
- Union-Find elegantly handles transitive relationships in entity resolution.
- Hash maps bridge the gap between entities (accounts) and their attributes (emails).
- Grouping by root representative consolidates distributed information efficiently.

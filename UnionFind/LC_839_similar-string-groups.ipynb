## Problem Statement
Two strings `X` and `Y` are similar if we can swap two letters (in different positions) of `X`, so that it equals `Y`. Also two strings are similar if they are equal.

For example, `"tars"` and `"rats"` are similar (swapping at positions 0 and 2), and `"rats"` and `"arts"` are similar, but `"star"` is not similar to `"tars"`, `"rats"`, or `"arts"`.

Together, these form two connected groups: `{"tars", "rats", "arts"}` and `{"star"}`.

Given an array of strings `strs`, return the number of groups.

## Progressive Hints
- Hint 1: Similarity is transitive: if A~B and B~C, then A~C form one group.
- Hint 2: Check if two strings are similar: at most 2 character differences.
- Hint 3: Use Union-Find to merge similar strings into groups.
- Hint 4: Compare all pairs of strings to build the equivalence classes.

## Solution Overview
Check all pairs of strings for similarity. Use Union-Find to merge similar strings into groups. The final number of distinct roots is the answer.

## Detailed Explanation
1. Initialize Union-Find with n strings.
2. Define similarity check: count character differences between two strings.
   - If 0 differences: strings are identical (similar).
   - If 2 differences: check if swapping makes them equal (similar).
   - Otherwise: not similar.
3. For each pair of strings, if similar, union them.
4. Count distinct roots to get the number of groups.
5. Optimization: Early termination if difference count > 2.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(n² * L * α(n)) | O(n) | n strings, L string length |
| DFS | O(n² * L) | O(n) | Build graph then count components |
| BFS | O(n² * L) | O(n) | Similar to DFS |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [0] * n
        self.count = n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    
    def union(self, x: int, y: int) -> bool:
        root_x, root_y = self.find(x), self.find(y)
        
        if root_x == root_y:
            return False
        
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1
        
        self.count -= 1
        return True


def is_similar(s1: str, s2: str) -> bool:
    """Check if two strings are similar (at most one swap away or identical)."""
    if s1 == s2:
        return True
    
    diff = []
    for i, (c1, c2) in enumerate(zip(s1, s2)):
        if c1 != c2:
            diff.append(i)
            if len(diff) > 2:
                return False
    
    # Exactly 2 differences and swapping makes them equal
    return len(diff) == 2 and s1[diff[0]] == s2[diff[1]] and s1[diff[1]] == s2[diff[0]]


def numSimilarGroups(strs: List[str]) -> int:
    n = len(strs)
    uf = UnionFind(n)
    
    for i in range(n):
        for j in range(i + 1, n):
            if is_similar(strs[i], strs[j]):
                uf.union(i, j)
    
    return uf.count
```

## Validation

```python
assert numSimilarGroups(["tars","rats","arts","star"]) == 2
assert numSimilarGroups(["omv","ovm"]) == 1
assert numSimilarGroups(["abc","abc"]) == 1
assert numSimilarGroups(["abc","def","ghi"]) == 3
assert numSimilarGroups(["a"]) == 1
print('All tests passed for LC 839.')
```

## Complexity Analysis
- Time Complexity: O(n² * L * α(n)), where n is number of strings, L is string length.
- Space Complexity: O(n) for Union-Find structure.
- Bottleneck: Comparing all pairs of strings with O(L) similarity check.

## Edge Cases & Pitfalls
- Identical strings: Should be in the same group.
- Single string: Should return 1.
- All different: Each string forms its own group.
- Transitive similarity: Must use Union-Find to capture indirect connections.

## Follow-up Variants
- Allow k swaps instead of 1: Requires different similarity definition.
- Find the largest group: Track group sizes in Union-Find.
- Count pairs of similar strings: Compute after grouping.

## Takeaways
- Union-Find efficiently handles transitive relationships even with custom similarity checks.
- Early termination in similarity checks (when difference > 2) improves performance.
- All-pairs comparison is necessary when similarity isn't based on simple hashing.

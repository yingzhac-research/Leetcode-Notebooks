## Problem Statement
You are given two strings of the same length `s1` and `s2` and a string `baseStr`.

We say `s1[i]` and `s2[i]` are equivalent characters.

- For example, if `s1 = "abc"` and `s2 = "cde"`, then we have `'a' == 'c'`, `'b' == 'd'`, and `'c' == 'e'`.

Equivalent characters follow the usual rules of any equivalence relation:
- Reflexivity: `'a' == 'a'`
- Symmetry: `'a' == 'b'` implies `'b' == 'a'`
- Transitivity: `'a' == 'b'` and `'b' == 'c'` implies `'a' == 'c'`

Return the lexicographically smallest equivalent string of `baseStr` by using the equivalency information from `s1` and `s2`.

## Progressive Hints
- Hint 1: Build equivalence classes using Union-Find from character pairs in s1 and s2.
- Hint 2: For each equivalence class, find the lexicographically smallest character.
- Hint 3: Replace each character in baseStr with its smallest equivalent.
- Hint 4: Always union towards the smaller character to maintain lexicographic order.

## Solution Overview
Use Union-Find with a custom union strategy that always makes the lexicographically smaller character the root. Then replace each character in baseStr with its root.

## Detailed Explanation
1. Initialize Union-Find for 26 lowercase letters.
2. For each pair (s1[i], s2[i]), union them with a modified strategy:
   - Always set the smaller character as the parent (root).
3. After all unions, each character's root is the smallest equivalent character.
4. For each character in baseStr, replace it with find(character).
5. Return the transformed string.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(n * α(26)) ≈ O(n) | O(26) = O(1) | n is total string lengths |
| DFS graph | O(n + 26²) | O(26) = O(1) | Build graph then find min in each component |
| BFS | O(n + 26²) | O(26) = O(1) | Similar to DFS |

## Reference Implementation

```python
class UnionFind:
    def __init__(self):
        # Each character is its own parent initially
        self.parent = {chr(i): chr(i) for i in range(ord('a'), ord('z') + 1)}
    
    def find(self, x: str) -> str:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    
    def union(self, x: str, y: str) -> None:
        root_x, root_y = self.find(x), self.find(y)
        
        if root_x == root_y:
            return
        
        # Always make the lexicographically smaller character the root
        if root_x < root_y:
            self.parent[root_y] = root_x
        else:
            self.parent[root_x] = root_y


def smallestEquivalentString(s1: str, s2: str, baseStr: str) -> str:
    uf = UnionFind()
    
    # Build equivalence classes
    for c1, c2 in zip(s1, s2):
        uf.union(c1, c2)
    
    # Replace each character in baseStr with its smallest equivalent
    result = []
    for char in baseStr:
        result.append(uf.find(char))
    
    return ''.join(result)
```

## Validation

```python
assert smallestEquivalentString("parker", "morris", "parser") == "makkek"
assert smallestEquivalentString("hello", "world", "hold") == "hdld"
assert smallestEquivalentString("leetcode", "programs", "sourcecode") == "aauaaaaada"
assert smallestEquivalentString("a", "b", "c") == "c"
print('All tests passed for LC 1061.')
```

## Complexity Analysis
- Time Complexity: O((|s1| + |baseStr|) * α(26)) ≈ O(n), where n is total string length.
- Space Complexity: O(26) = O(1) for fixed alphabet size.
- Bottleneck: Union operations and character replacements.

## Edge Cases & Pitfalls
- No equivalences: baseStr remains unchanged.
- Self-equivalence: Characters equivalent to themselves don't change.
- Transitive chains: Must use Union-Find for correct propagation.
- Custom union strategy: Always union towards the smaller character.

## Follow-up Variants
- Largest equivalent string: Union towards larger characters.
- Count distinct equivalence classes: Count unique roots.
- Multi-character equivalences: Extend to string equivalences.

## Takeaways
- Union-Find can be customized with application-specific union strategies.
- Unioning towards a specific preference (smallest) enables elegant solutions.
- Fixed alphabet size makes the solution O(1) space complexity.

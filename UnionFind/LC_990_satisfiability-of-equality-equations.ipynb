## Problem Statement
You are given an array of strings `equations` that represent relationships between variables where each string `equations[i]` has length 4 and takes one of two forms: `"xi==xj"` or `"xi!=xj"`.

Here, `xi` and `xj` are lowercase letters (not necessarily different) that represent one-letter variable names.

Return `true` if it is possible to assign integers to variable names so as to satisfy all the given equations, or `false` otherwise.

## Progressive Hints
- Hint 1: Process equality equations first to group variables into equivalence classes.
- Hint 2: Use Union-Find to merge variables that must be equal.
- Hint 3: After processing equalities, check inequality constraints.
- Hint 4: An inequality `a != b` is violated if `find(a) == find(b)`.

## Solution Overview
First, process all equality equations to build equivalence classes using Union-Find. Then, verify that no inequality constraint connects two variables in the same equivalence class.

## Detailed Explanation
1. Initialize Union-Find for 26 lowercase letters.
2. Process all `==` equations: for each `a==b`, call `union(a, b)`.
3. Process all `!=` equations: for each `a!=b`, check if `find(a) == find(b)`.
4. If any inequality connects variables in the same set, return `False`.
5. Otherwise, return `True` - all constraints can be satisfied.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(n α(26)) ≈ O(n) | O(26) = O(1) | Optimal with fixed alphabet size |
| Graph DFS | O(n * 26) | O(26) = O(1) | Build graph then check conflicts |
| Brute force | O(26^26) | O(1) | Try all variable assignments |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [0] * n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    
    def union(self, x: int, y: int) -> None:
        root_x, root_y = self.find(x), self.find(y)
        
        if root_x == root_y:
            return
        
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1


def equationsPossible(equations: List[str]) -> bool:
    uf = UnionFind(26)
    
    # Process all equality equations first
    for eq in equations:
        if eq[1] == '=':
            x, y = ord(eq[0]) - ord('a'), ord(eq[3]) - ord('a')
            uf.union(x, y)
    
    # Check all inequality equations
    for eq in equations:
        if eq[1] == '!':
            x, y = ord(eq[0]) - ord('a'), ord(eq[3]) - ord('a')
            if uf.find(x) == uf.find(y):
                return False
    
    return True
```

## Validation

```python
assert equationsPossible(["a==b","b!=a"]) == False
assert equationsPossible(["b==a","a==b"]) == True
assert equationsPossible(["a==b","b==c","a==c"]) == True
assert equationsPossible(["a==b","b!=c","c==a"]) == False
assert equationsPossible(["c==c","b==d","x!=z"]) == True
print('All tests passed for LC 990.')
```

## Complexity Analysis
- Time Complexity: O(n α(26)) ≈ O(n), where n is the number of equations.
- Space Complexity: O(26) = O(1) for fixed alphabet size.
- Bottleneck: Two passes through the equations array.

## Edge Cases & Pitfalls
- Self-equality: `a==a` is always valid.
- Self-inequality: `a!=a` is always invalid.
- Transitive equality: `a==b`, `b==c` implies `a==c`.
- Order matters: Must process all equalities before checking inequalities.

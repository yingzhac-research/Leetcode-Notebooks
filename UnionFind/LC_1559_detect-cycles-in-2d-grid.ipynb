## Problem Statement
Given a 2D array of characters `grid` of size `m x n`, you need to find if there exists any cycle consisting of the same value in `grid`.

A cycle is a path of length 4 or more in the grid that starts and ends at the same cell. From a given cell, you can move to one of the cells adjacent to it (up, down, left, or right) if it has the same value.

Return `true` if any cycle of the same value exists in `grid`, otherwise, return `false`.

## Progressive Hints
- Hint 1: A cycle means there are two different paths to reach the same cell with the same value.
- Hint 2: Union-Find can detect cycles when trying to union cells that already share a root.
- Hint 3: Only connect adjacent cells with the same character value.
- Hint 4: Process edges carefully to avoid immediately connecting a cell with its parent.

## Solution Overview
Use Union-Find to connect adjacent cells with the same value. If we try to union two cells that already have the same root (and they're not parent-child), a cycle exists.

## Detailed Explanation
1. **Initialize Union-Find** with m*n cells, indexed as `i*n + j`.
2. **Process each cell**: For each cell (i, j), check right and down neighbors only (to avoid double-processing edges).
3. **Same value check**: If neighbor has the same character value:
   - Find roots of both cells.
   - If roots are the same, a cycle exists (return True).
   - If roots differ, union them.
4. **Why this works**: If two adjacent cells already share a root through a different path, we've found a cycle.
5. **Directional optimization**: Only check right and down to process each edge once.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find | O(m*n α(m*n)) | O(m*n) | Efficient for detecting any cycle |
| DFS with visited | O(m*n) | O(m*n) | Track parent to avoid false cycles |
| BFS | O(m*n) | O(m*n) | Similar to DFS |

## Reference Implementation

```python
from typing import List


class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [0] * n
    
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]
    
    def union(self, x: int, y: int) -> bool:
        root_x, root_y = self.find(x), self.find(y)
        if root_x == root_y:
            return False  # Cycle detected
        
        # Union by rank
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1
        return True


def containsCycle(grid: List[List[str]]) -> bool:
    if not grid or not grid[0]:
        return False
    
    m, n = len(grid), len(grid[0])
    uf = UnionFind(m * n)
    
    def get_index(row: int, col: int) -> int:
        return row * n + col
    
    # Check right and down neighbors only (to avoid double-processing)
    directions = [(0, 1), (1, 0)]  # right, down
    
    for i in range(m):
        for j in range(n):
            for di, dj in directions:
                ni, nj = i + di, j + dj
                if 0 <= ni < m and 0 <= nj < n and grid[i][j] == grid[ni][nj]:
                    # If union returns False, they're already connected -> cycle
                    if not uf.union(get_index(i, j), get_index(ni, nj)):
                        return True
    
    return False
```

## Validation

```python
assert containsCycle([["a","a","a","a"],["a","b","b","a"],["a","b","b","a"],["a","a","a","a"]]) == True
assert containsCycle([["c","c","c","a"],["c","d","c","c"],["c","c","e","c"],["f","c","c","c"]]) == True
assert containsCycle([["a","b","b"],["b","z","b"],["b","b","a"]]) == False
assert containsCycle([["a"]]) == False
assert containsCycle([["a","a"],["a","a"]]) == True
print('All tests passed for LC 1559.')
```

## Complexity Analysis
- Time Complexity: O(m*n α(m*n)), where m and n are grid dimensions and α is the inverse Ackermann function.
- Space Complexity: O(m*n) for the Union-Find parent and rank arrays.
- Bottleneck: Processing all cells and their neighbors with Union-Find operations.

## Edge Cases & Pitfalls
- Single cell: No cycle possible, return False.
- Minimum cycle: 2x2 grid with same values forms a cycle.
- No same adjacent values: No unions performed, no cycle.
- Large connected component: Must have actual cycle, not just connectivity.

## Follow-up Variants

- Extend to diagonal movements or 3D grids.

## Takeaways
- Union-Find naturally detects cycles when union fails (same root already).
- Processing only right and down neighbors avoids redundant edge processing.
- Grid problems can often be solved by converting to graph representation.
- Path compression makes repeated find operations very efficient.

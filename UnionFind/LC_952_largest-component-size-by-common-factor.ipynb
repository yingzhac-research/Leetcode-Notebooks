## Problem Statement
You have an integer array `nums`. You would like to split them into several groups where each group has the largest size possible and every pair of numbers in a group shares a common factor greater than 1.

Return the size of the largest group.

**Note**: A common factor is an integer greater than 1 that divides both numbers evenly.

**Example 1**:
```
Input: nums = [4,6,15,35]
Output: 4
Explanation: The numbers can all be grouped together as they share common factors:
- 4 and 6 share factor 2
- 6 and 15 share factor 3
- 15 and 35 share factor 5
```

**Example 2**:
```
Input: nums = [20,50,9,63]
Output: 2
```

## Progressive Hints
- Hint 1: Direct pairwise comparison of all numbers is O(n²) - too slow for large inputs.
- Hint 2: Instead of connecting numbers directly, connect them through their prime factors.
- Hint 3: Each number can be connected to all its prime factors in Union-Find.
- Hint 4: Use efficient factorization up to √num to find all factors, then connect numbers sharing any prime factor.

## Solution Overview
Use Union-Find where we connect each number to its prime factors. Numbers sharing any prime factor will be in the same component. Track component sizes and return the maximum.

## Detailed Explanation
1. **Prime Factorization Strategy**: For each number, find all its prime factors efficiently:
   - Check divisibility by 2, then odd numbers from 3 to √num
   - If remainder > 1 after factorization, it's also a prime factor
2. **Union-Find Connections**: Instead of connecting numbers to each other:
   - Connect each number to all its prime factors
   - This creates indirect connections between numbers sharing factors
3. **Component Size Tracking**: Use a hash map to count numbers in each component:
   - Map each number's root to its count
   - Return the maximum count
4. **Mathematical Insight**: Two numbers share a common factor > 1 if and only if they share at least one prime factor.
5. **Optimization**: We only need to factorize up to √num, significantly reducing time complexity.
6. **Edge Cases**: Single element groups, all coprime numbers (no shared factors), large prime numbers.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Union-Find with prime factors | O(n√m α(n)) | O(n + p) | Optimal; m = max(nums), p = unique primes |
| Direct pairwise GCD | O(n² log m) | O(n) | Too slow for large n |
| Graph DFS | O(n² log m) | O(n²) | Requires building full adjacency |
| Sieve factorization | O(m log log m + n) | O(m) | Fast but high space for large m |

## Reference Implementation

````python
from typing import List
from collections import defaultdict


class UnionFind:
    def __init__(self):
        self.parent = {}
        self.rank = {}
    
    def find(self, x: int) -> int:
        if x not in self.parent:
            self.parent[x] = x
            self.rank[x] = 0
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]
    
    def union(self, x: int, y: int):
        root_x, root_y = self.find(x), self.find(y)
        if root_x != root_y:
            # Union by rank
            if self.rank[root_x] > self.rank[root_y]:
                self.parent[root_y] = root_x
            elif self.rank[root_x] < self.rank[root_y]:
                self.parent[root_x] = root_y
            else:
                self.parent[root_y] = root_x
                self.rank[root_x] += 1


def largestComponentSize(nums: List[int]) -> int:
    def get_prime_factors(n: int) -> List[int]:
        """Get all prime factors of n."""
        factors = []
        # Check for factor 2
        if n % 2 == 0:
            factors.append(2)
            while n % 2 == 0:
                n //= 2
        
        # Check odd factors up to sqrt(n)
        i = 3
        while i * i <= n:
            if n % i == 0:
                factors.append(i)
                while n % i == 0:
                    n //= i
            i += 2
        
        # If n > 1, then it's a prime factor
        if n > 1:
            factors.append(n)
        
        return factors
    
    uf = UnionFind()
    
    # Connect each number to its prime factors
    for num in nums:
        primes = get_prime_factors(num)
        # Union num with all its prime factors
        for prime in primes:
            uf.union(num, prime)
    
    # Count component sizes
    component_count = defaultdict(int)
    for num in nums:
        root = uf.find(num)
        component_count[root] += 1
    
    return max(component_count.values())
````

## Validation

````python
assert largestComponentSize([4,6,15,35]) == 4
assert largestComponentSize([20,50,9,63]) == 2
assert largestComponentSize([2,3,6,7,4,12,21,39]) == 8
assert largestComponentSize([1]) == 1
assert largestComponentSize([83,99,39,11,19,30,31]) == 4
print('All tests passed for LC 952.')
````

## Complexity Analysis
- Time Complexity: O(n√m α(n)), where n is length of nums, m is max value in nums, and α is inverse Ackermann function.
  - Factorization: O(√m) per number
  - Union operations: O(α(n)) per operation
- Space Complexity: O(n + p), where p is the number of unique prime factors encountered.
- Bottleneck: Prime factorization for large numbers.

## Edge Cases & Pitfalls
- Single element: Should return 1 (component of size 1).
- All coprime numbers: Each forms its own component of size 1.
- Large prime numbers: Have no shared factors with other numbers unless duplicates exist.
- Number 1: Has no prime factors; forms isolated component (though not in constraints).
- Duplicate numbers: Share all factors, must be in same component.

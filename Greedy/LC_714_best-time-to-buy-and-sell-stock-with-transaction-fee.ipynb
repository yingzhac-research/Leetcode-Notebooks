## Problem Statement
You are given an array `prices` where `prices[i]` is the price of a given stock on the `i-th` day, and an integer `fee` representing a transaction fee.

Find the maximum profit you can achieve. You may complete as many transactions as you like, but you need to pay the transaction fee for each transaction.

**Note**: You may not engage in multiple transactions simultaneously (i.e., you must sell the stock before you buy again).

**Constraints:**
- `1 <= prices.length <= 5 * 10^4`
- `1 <= prices[i] < 5 * 10^4`
- `0 <= fee < 5 * 10^4`

## Progressive Hints
- Hint 1: Think about the state at each day: either you hold a stock or you don't. How does the transaction fee affect your decision?
- Hint 2: Use dynamic programming or greedy approach to track the maximum profit in two states: holding vs not holding.
- Hint 3: When you sell, subtract the fee. This way, you only pay once per complete transaction.
- Hint 4: The greedy insight: keep track of the minimum effective buy price and sell only when profit exceeds the fee.

## Solution Overview
We can solve this using either DP or a greedy approach:

**DP Approach** (State Machine):
1. Track two states at each day: `hold` (holding stock) and `free` (not holding).
2. Transition equations:
   - `hold[i] = max(hold[i-1], free[i-1] - prices[i])`
   - `free[i] = max(free[i-1], hold[i-1] + prices[i] - fee)`
3. Answer is `free[n-1]`.

**Greedy Approach**:
1. Track the minimum buy price seen so far.
2. For each price, calculate potential profit if we sell now.
3. If profit > fee, accumulate profit and update minimum buy price.
4. This greedy approach works because we can always adjust our buy price to maximize profit.

Both approaches are O(n) time and can be optimized to O(1) space.

## Detailed Explanation
### DP State Machine Approach:
Define two states:
- `hold`: max profit while holding stock
- `free`: max profit while not holding stock

**State transitions**:
- To hold today: either held yesterday or buy today
  - `hold = max(hold, free - prices[i])`
- To be free today: either free yesterday or sell today
  - `free = max(free, hold + prices[i] - fee)`

**Initialization**:
- `hold = -prices[0]` (buy on day 0)
- `free = 0` (don't buy on day 0)

**Algorithm flow**:
```
hold = -prices[0]
free = 0

for i in range(1, n):
    new_hold = max(hold, free - prices[i])
    new_free = max(free, hold + prices[i] - fee)
    hold, free = new_hold, new_free

return free
```

**Example walkthrough** for `prices = [1,3,2,8,4,9], fee = 2`:
- Day 0: hold = -1, free = 0
- Day 1: hold = max(-1, 0-3) = -1, free = max(0, -1+3-2) = 0
- Day 2: hold = max(-1, 0-2) = -1, free = max(0, -1+2-2) = 0
- Day 3: hold = max(-1, 0-8) = -1, free = max(0, -1+8-2) = 5
- Day 4: hold = max(-1, 5-4) = 1, free = max(5, -1+4-2) = 5
- Day 5: hold = max(1, 5-9) = 1, free = max(5, 1+9-2) = 8
- Result: 8

## Complexity Trade-off Table

## Reference Implementation

````python
from typing import List


def max_profit(prices: List[int], fee: int) -> int:
    """
    Calculate maximum profit with unlimited transactions and a fee per transaction.
    
    Args:
        prices: Daily stock prices
        fee: Transaction fee
    
    Returns:
        Maximum profit achievable
    """
    if not prices:
        return 0
    
    # State: hold (holding stock) and free (not holding stock)
    hold = -prices[0]  # Buy on day 0
    free = 0  # Don't buy on day 0
    
    for i in range(1, len(prices)):
        # To hold: either already holding or buy today
        new_hold = max(hold, free - prices[i])
        
        # To be free: either already free or sell today (pay fee)
        new_free = max(free, hold + prices[i] - fee)
        
        hold, free = new_hold, new_free
    
    # At the end, we should not be holding stock
    return free
````

## Validation

````python
cases = [
    ([1, 3, 2, 8, 4, 9], 2, 8),  # Buy at 1, sell at 8 (profit 7-2=5), buy at 4, sell at 9 (profit 5-2=3), total 8
    ([1, 3, 7, 5, 10, 3], 3, 6),  # Buy at 1, sell at 10 (profit 9-3=6)
    ([1, 2, 3, 4, 5], 1, 3),  # Buy at 1, sell at 5 (profit 4-1=3)
    ([5, 4, 3, 2, 1], 1, 0),  # No transactions profitable
    ([1], 0, 0),  # Single day, no transaction
    ([1, 4, 6, 2, 8, 3, 10, 14], 3, 13),  # Multiple transactions
]

for prices, fee, expected in cases:
    result = max_profit(prices, fee)
    assert result == expected, f"Failed for prices={prices}, fee={fee}: got {result}, expected {expected}"

print('All tests passed for LC 714.')
````

## Complexity Analysis
- **Time Complexity**: O(n) where n is the number of days
  - Single pass through the prices array
  - Constant time operations per day
- **Space Complexity**: O(1)
  - Only two variables (hold and free) are maintained
  - No additional data structures needed
- **Bottleneck**: The single iteration through prices is unavoidable.

## Edge Cases & Pitfalls
- **Single day**: Cannot make any transaction, return 0.
- **Decreasing prices**: No profitable transactions, return 0.
- **Fee larger than price differences**: May make all transactions unprofitable.
- **Fee = 0**: Reduces to the simpler stock problem (buy and sell unlimited times).
- **State update order**: Must use temporary variables to avoid overwriting states needed for calculation.
- **Initial state**: `hold = -prices[0]` represents buying on day 0.

## Takeaways
- State machine DP is powerful for problems with distinct states and transitions.
- For stock problems, track states: holding vs not holding.
- Space optimization: reduce DP arrays to variables when only previous state is needed.
- Transaction costs can be incorporated into state transitions.
- The greedy aspect: at each step, choose the best action given current state.

## Problem Statement
You are given an array of `events` where `events[i] = [startDayi, endDayi]`. Every event `i` starts at `startDayi` and ends at `endDayi`.

You can attend an event `i` at any day `d` where `startTimei <= d <= endTimei`. You can only attend one event at any time `d`.

Return the **maximum number** of events you can attend.

**Constraints:**
- `1 <= events.length <= 10^5`
- `events[i].length == 2`
- `1 <= startDayi <= endDayi <= 10^5`

## Progressive Hints
- Hint 1: Process days in chronological order. On each day, consider which events are available.
- Hint 2: Among available events on a given day, which one should you attend? Think about urgency.
- Hint 3: Use a min heap to track events by their end day. Always attend the event that ends soonest.
- Hint 4: Sort events by start day, then simulate day by day, adding events that start on current day to heap.

## Solution Overview
The optimal greedy approach:
1. Sort events by start day.
2. Iterate through each day from the earliest start to the latest end.
3. For each day:
   - Add all events starting on this day to a min heap (keyed by end day).
   - Remove events that have already ended (end day < current day).
   - If heap is not empty, attend the event with the earliest end day (pop from heap).
4. Count the number of events attended.

**Why greedy works**: By always attending the event that ends soonest, we maximize flexibility for future days. This is an application of the "earliest deadline first" scheduling principle.

## Detailed Explanation
1. **Sort events**: Sort by start day to process chronologically.
2. **Initialize**:
   - Min heap to store end days of available events
   - Event index pointer
   - Counter for attended events
3. **Simulate each day**:
   - Determine the range: from min start day to max end day
   - For each day d:
     - Add all events starting on day d to heap
     - Remove expired events (end day < d)
     - If heap not empty: attend event (pop), increment counter
4. **Optimization**: Skip days with no available events.

**Algorithm flow**:
```
events.sort()
heap = []
i = 0
attended = 0

for day in range(min_start, max_end + 1):
    # Add events starting today
    while i < n and events[i][0] == day:
        heappush(heap, events[i][1])
        i += 1
    
    # Remove expired events
    while heap and heap[0] < day:
        heappop(heap)
    
    # Attend the event ending soonest
    if heap:
        heappop(heap)
        attended += 1

return attended
```

**Example walkthrough** for `events = [[1,2],[2,3],[3,4]]`:
- Day 1: Add event [1,2], attend it, count = 1
- Day 2: Add event [2,3], attend it, count = 2
- Day 3: Add event [3,4], attend it, count = 3
- Result: 3

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy with heap | O(n log n + d log n) | O(n) | d is the number of days, n is events. Optimal. |
| Greedy with sorting only | O(n^2) | O(1) | Repeatedly find best event each day. |
| Brute force | O(2^n) | O(n) | Try all subsets of events. Too slow. |

## Reference Implementation

````python
import heapq
from typing import List


def max_events(events: List[List[int]]) -> int:
    """
    Find maximum number of events that can be attended.
    
    Args:
        events: List of [start_day, end_day] for each event
    
    Returns:
        Maximum number of events attendable
    """
    # Sort events by start day
    events.sort()
    
    # Min heap to track events by end day
    heap = []
    event_idx = 0
    attended = 0
    n = len(events)
    
    # Simulate each day
    day = events[0][0]  # Start from first event's start day
    
    while event_idx < n or heap:
        # Add all events starting on current day
        while event_idx < n and events[event_idx][0] == day:
            heapq.heappush(heap, events[event_idx][1])
            event_idx += 1
        
        # Remove events that have ended
        while heap and heap[0] < day:
            heapq.heappop(heap)
        
        # Attend the event ending soonest
        if heap:
            heapq.heappop(heap)
            attended += 1
        
        # Move to next day
        day += 1
    
    return attended
````

## Validation

````python
cases = [
    ([[1, 2], [2, 3], [3, 4]], 3),
    ([[1, 2], [2, 3], [3, 4], [1, 2]], 4),
    ([[1, 4], [4, 4], [2, 2], [3, 4], [1, 1]], 4),
    ([[1, 5], [1, 5], [1, 5], [2, 3], [2, 3]], 5),
    ([[1, 1], [1, 2], [1, 3], [1, 4], [1, 5]], 5),
    ([[1, 100000]], 1),
]

for events, expected in cases:
    result = max_events([e[:] for e in events])  # Deep copy
    assert result == expected, f"Failed for events={events}: got {result}, expected {expected}"

print('All tests passed for LC 1353.')
````

## Complexity Analysis
- **Time Complexity**: O(n log n + d log n) where n is the number of events and d is the number of unique days
  - Sorting events: O(n log n)
  - Each day: O(log n) for heap operations
  - Total days processed: O(d), where d ≤ max_end - min_start
  - Overall: O(n log n + d log n), in worst case O(n log n) when d is small
- **Space Complexity**: O(n) for the heap
- **Bottleneck**: Heap operations and number of days to simulate.

## Edge Cases & Pitfalls
- **Single event**: Simply attend it, return 1.
- **All events on same day**: Can only attend 1.
- **Non-overlapping events**: Can attend all.
- **Events with same end day**: Order doesn't matter among them.
- **Large day range**: If max_end is very large, iterating all days might be slow; optimize by skipping empty days.
- **Heap maintenance**: Regularly remove expired events to keep heap small.
- **Event already ended**: Skip events where end_day < current_day.

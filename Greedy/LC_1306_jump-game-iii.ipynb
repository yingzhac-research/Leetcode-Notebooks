## Problem Statement
Given an array of non-negative integers `arr`, you are initially positioned at `start` index of the array. When you are at index `i`, you can jump to `i + arr[i]` or `i - arr[i]`, check if you can reach **any** index with value 0.

Notice that you can not jump outside of the array at any time.

**Constraints:**
- `1 <= arr.length <= 5 * 10^4`
- `0 <= arr[i] < arr.length`
- `0 <= start < arr.length`

## Progressive Hints
- Hint 1: Think of this as a graph problem where each index is a node and jumps are edges.
- Hint 2: Use BFS or DFS to explore all reachable indices from the start position.
- Hint 3: Keep track of visited indices to avoid infinite loops.
- Hint 4: The greedy insight: we don't need to find the shortest path, just check if any zero is reachable.

## Solution Overview
We can solve this using either BFS or DFS:

**BFS Approach**:
1. Use a queue initialized with the start index.
2. For each index, explore both possible jumps: `i + arr[i]` and `i - arr[i]`.
3. Mark visited indices to avoid cycles.
4. If we reach an index with value 0, return True.
5. If queue is exhausted without finding 0, return False.

**DFS Approach**:
1. Recursively explore from the start index.
2. Mark current index as visited.
3. If current value is 0, return True.
4. Try both jumps recursively.
5. Return True if either direction succeeds.

**Greedy insight**: We explore all reachable states and stop as soon as we find a zero. No optimization needed beyond avoiding revisits.

## Detailed Explanation
### BFS Approach:
1. **Initialize**: Create a queue with start index, and a visited set.
2. **BFS loop**:
   - Pop an index from queue
   - If already visited, skip
   - Mark as visited
   - If value is 0, return True
   - Calculate next indices: `i + arr[i]` and `i - arr[i]`
   - Add valid (in bounds, not visited) indices to queue
3. **Return False**: If queue is empty and no zero found.

**Algorithm flow**:
```
queue = [start]
visited = set()

while queue:
    i = queue.pop(0)
    if i in visited or i < 0 or i >= len(arr):
        continue
    visited.add(i)
    
    if arr[i] == 0:
        return True
    
    queue.append(i + arr[i])
    queue.append(i - arr[i])

return False
```

**Example walkthrough** for `arr = [4,2,3,0,3,1,2], start = 5`:
- Start at index 5 (value 1)
- Can jump to 5+1=6 or 5-1=4
- From index 6 (value 2): can jump to 8 (invalid) or 4
- From index 4 (value 3): can jump to 7 (invalid) or 1
- From index 1 (value 2): can jump to 3 or -1 (invalid)
- Index 3 has value 0 → return True

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| BFS | O(n) | O(n) | Visit each index at most once. Optimal. |
| DFS | O(n) | O(n) | Same complexity, recursion stack. |
| DFS with in-place marking | O(n) | O(1) | Mark visited by negating values (if allowed). |
| Brute force | O(2^n) | O(n) | Try all paths. Too slow. |

## Reference Implementation

````python
from collections import deque
from typing import List


def can_reach(arr: List[int], start: int) -> bool:
    """
    Determine if we can reach any index with value 0 from start.
    
    Args:
        arr: Array of non-negative integers
        start: Starting index
    
    Returns:
        True if we can reach a zero, False otherwise
    """
    n = len(arr)
    visited = set()
    queue = deque([start])
    
    while queue:
        i = queue.popleft()
        
        # Check bounds and visited
        if i < 0 or i >= n or i in visited:
            continue
        
        # Mark as visited
        visited.add(i)
        
        # Check if we found a zero
        if arr[i] == 0:
            return True
        
        # Add next positions to queue
        queue.append(i + arr[i])
        queue.append(i - arr[i])
    
    return False


def can_reach_dfs(arr: List[int], start: int) -> bool:
    """
    DFS solution for comparison.
    """
    n = len(arr)
    visited = set()
    
    def dfs(i: int) -> bool:
        # Check bounds and visited
        if i < 0 or i >= n or i in visited:
            return False
        
        visited.add(i)
        
        # Found zero
        if arr[i] == 0:
            return True
        
        # Try both directions
        return dfs(i + arr[i]) or dfs(i - arr[i])
    
    return dfs(start)
````

## Validation

````python
cases = [
    ([4, 2, 3, 0, 3, 1, 2], 5, True),
    ([4, 2, 3, 0, 3, 1, 2], 0, True),
    ([3, 0, 2, 1, 2], 2, False),
    ([0], 0, True),
    ([1, 1, 1, 1, 1, 1, 1, 0], 3, True),
    ([58, 48, 33, 84, 97, 18, 42, 0], 5, True),
]

for arr, start, expected in cases:
    result_bfs = can_reach(arr.copy(), start)
    result_dfs = can_reach_dfs(arr.copy(), start)
    assert result_bfs == expected, f"BFS failed for arr={arr}, start={start}: got {result_bfs}, expected {expected}"
    assert result_dfs == expected, f"DFS failed for arr={arr}, start={start}: got {result_dfs}, expected {expected}"

print('All tests passed for LC 1306.')
````

## Complexity Analysis
- **Time Complexity**: O(n) where n is the length of the array
  - Each index is visited at most once
  - Each index generates at most 2 new indices to explore
  - Overall: O(n)
- **Space Complexity**: O(n)
  - Visited set: O(n)
  - Queue/recursion stack: O(n) in worst case
- **Bottleneck**: The traversal is necessary to determine reachability.

## Edge Cases & Pitfalls
- **Start is already zero**: Return True immediately.
- **Single element**: If `arr[0] == 0`, return True; otherwise False.
- **Cycles**: Without visited tracking, we might loop infinitely.
- **Out of bounds**: Always check `0 <= i < n` before accessing.
- **All zeros**: Any starting position returns True.
- **No zeros**: Return False after exhaustive search.
- **BFS vs DFS**: BFS finds shortest path but we only need existence; both work.

## Takeaways
- Graph traversal (BFS/DFS) is essential for reachability problems.
- Always track visited nodes to avoid infinite loops in cyclic graphs.
- For existence problems ("can we reach"), we don't need path reconstruction.
- BFS and DFS have similar complexity but different space characteristics.
- In-place marking (e.g., negating visited values) can save space if mutation is allowed.

## Problem Statement
You are given an array of `n` pairs `pairs` where `pairs[i] = [left_i, right_i]` and `left_i < right_i`.

A pair `p2 = [c, d]` follows a pair `p1 = [a, b]` if `b < c`. A chain of pairs can be formed in this fashion.

Return the length longest chain which can be formed.

You do not need to use up all the given intervals. You can select pairs in any order.

**Constraints:**
- `n == pairs.length`
- `1 <= n <= 1000`
- `-1000 <= left_i < right_i <= 1000`

## Progressive Hints
- Hint 1: Think about how sorting the pairs might help. What ordering would be most useful?
- Hint 2: If you sort by the second element of each pair, you can use a greedy approach.
- Hint 3: Always pick the pair with the smallest ending point that doesn't conflict with the previous selection.
- Hint 4: This problem is similar to the Activity Selection problem or LC 435 Non-overlapping Intervals.

## Solution Overview
The optimal approach uses a greedy interval selection strategy:
1. Sort pairs by their second element (right value).
2. Select the first pair and track its end point.
3. For each subsequent pair, if its start is greater than the current end, add it to the chain.
4. Update the current end to the newly selected pair's end.

**Why greedy works**: By always selecting the pair with the earliest end point, we maximize the remaining space for future pairs, leading to the longest possible chain.

## Detailed Explanation
1. **Sort pairs by right value**: `pairs.sort(key=lambda x: x[1])`
2. **Initialize chain length**: Start with 1 (first pair is always selected).
3. **Track current end**: Set to the right value of the first pair.
4. **Iterate through remaining pairs**:
   - If current pair's left value > current end, it can be chained.
   - Increment chain length and update current end.
5. **Return chain length**.

**Key insight**: This is essentially the Activity Selection problem. Choosing the activity/pair that finishes earliest leaves the most room for subsequent selections.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy with sorting | O(n log n) | O(1) | Optimal and most efficient. |
| Dynamic Programming | O(n^2) | O(n) | DP[i] = max chain ending at i. |
| DP with binary search | O(n log n) | O(n) | Similar to LIS with binary search. |
| Brute force | O(2^n) | O(n) | Try all possible chains. |

## Reference Implementation

```python
from typing import List


def find_longest_chain(pairs: List[List[int]]) -> int:
    """
    Find the maximum length of chain that can be formed from pairs.
    
    Args:
        pairs: List of pairs [left, right] where left < right
    
    Returns:
        Maximum length of chain
    """
    if not pairs:
        return 0
    
    # Sort pairs by their second element (end point)
    pairs.sort(key=lambda x: x[1])
    
    chain_length = 1  # At least one pair can be selected
    current_end = pairs[0][1]  # End of the first selected pair
    
    for i in range(1, len(pairs)):
        # If current pair's start is after the previous end, it can be chained
        if pairs[i][0] > current_end:
            chain_length += 1
            current_end = pairs[i][1]  # Update to new end point
    
    return chain_length
```

## Validation

```python
cases = [
    ([[1, 2], [2, 3], [3, 4]], 2),
    ([[1, 2], [7, 8], [4, 5]], 3),
    ([[-6, 9], [1, 6], [8, 10], [-1, 4], [-6, -2], [9, 10], [-5, 8]], 3),
    ([[1, 2]], 1),
    ([[1, 10], [2, 3], [4, 5], [6, 7]], 3),
]
for pairs, expected in cases:
    result = find_longest_chain([p.copy() for p in pairs])
    assert result == expected, f"Failed for pairs={pairs}: got {result}, expected {expected}"
print('All tests passed for LC 646.')
```

## Complexity Analysis
- **Time Complexity**: O(n log n) where n = len(pairs)
  - Sorting takes O(n log n)
  - Single pass through sorted array takes O(n)
  - Overall dominated by sorting
- **Space Complexity**: O(1) if sorting in-place
  - Only constant extra space for variables
- **Bottleneck**: The sorting step is the limiting factor.

## Edge Cases & Pitfalls
- **Empty array**: Should return 0.
- **Single pair**: Always return 1.
- **All non-chainable pairs**: Overlapping or touching pairs that can't be chained.
- **All chainable pairs**: When all pairs can form one long chain.
- **Strict inequality**: Note that b < c is required, not b <= c. Touching endpoints don't allow chaining.
- **Order independence**: Pairs can be selected in any order, not necessarily the given order.

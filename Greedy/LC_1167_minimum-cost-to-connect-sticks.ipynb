## Problem Statement
You have some number of sticks with positive integer lengths. These lengths are given as an array `sticks`, where `sticks[i]` is the length of the ith stick.

You can connect any two sticks of lengths `x` and `y` into one stick by paying a cost of `x + y`. You must connect all the sticks until there is only one stick remaining.

Return the minimum cost of connecting all the given sticks into one stick in this way.

**Example 1:**
```
Input: sticks = [2,4,3]
Output: 14
Explanation: First connect 2 and 3, cost = 5. Then connect 5 and 4, cost = 9. Total = 14.
```

**Example 2:**
```
Input: sticks = [1,8,3,5]
Output: 30
Explanation: First connect 1 and 3 (cost=4), then 4 and 5 (cost=9), then 9 and 8 (cost=17). Total = 4+9+17=30.
```

**Example 3:**
```
Input: sticks = [5]
Output: 0
Explanation: Only one stick, no connection needed.
```

**Constraints:**
- `1 <= sticks.length <= 10^4`
- `1 <= sticks[i] <= 10^4`

## Progressive Hints
- Hint 1: To minimize cost, always combine the two smallest sticks first.
- Hint 2: Use a min-heap to efficiently find and remove the two smallest sticks.
- Hint 3: After combining two sticks, insert the combined stick back into the heap. Repeat until one stick remains.

## Solution Overview
The optimal greedy approach uses a min-heap:
1. Insert all stick lengths into a min-heap.
2. While heap has more than one element:
   - Extract two smallest sticks
   - Combine them (cost = sum of their lengths)
   - Add cost to total
   - Insert combined stick back to heap
3. Return total cost.

**Why greedy works**: This is equivalent to building an optimal Huffman tree. By always merging the smallest elements first, we ensure they appear in more combinations (deeper in the tree), but their small values minimize the overall cost.

## Detailed Explanation
**Algorithm (Huffman-like merging)**:

1. **Initialize min-heap**: `heapify(sticks)`

2. **Greedy merging**:
   ```python
   total_cost = 0
   while len(heap) > 1:
       stick1 = heappop(heap)  # Smallest
       stick2 = heappop(heap)  # Second smallest
       combined = stick1 + stick2
       total_cost += combined
       heappush(heap, combined)
   return total_cost
   ```

3. **Why this is optimal**:
   - Each stick contributes to the cost based on how many times it's involved in merges
   - Smaller sticks should be merged first so they're "buried" in larger combinations
   - This minimizes their contribution to the total cost

**Example trace**: sticks = [2,4,3]
- Heap: [2, 3, 4]
- Pop 2, 3; cost += 5; heap: [4, 5]
- Pop 4, 5; cost += 9; heap: [9]
- Total cost: 14

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Min-heap (optimal) | O(n log n) | O(n) | Each of n-1 merges involves 2 pops and 1 push. |
| Sort and greedy | O(n^2 log n) | O(n) | Sort after each merge. |
| Brute force | O(n!) | O(n) | Try all merge orders. Factorial time. |

## Reference Implementation

````python
from typing import List
import heapq


def connect_sticks(sticks: List[int]) -> int:
    """
    Find minimum cost to connect all sticks into one.
    
    Args:
        sticks: List of stick lengths
    
    Returns:
        Minimum total cost to connect all sticks
    """
    # Edge case: single stick requires no connection
    if len(sticks) <= 1:
        return 0
    
    # Convert to min-heap
    heapq.heapify(sticks)
    
    total_cost = 0
    
    # Keep combining two smallest sticks until one remains
    while len(sticks) > 1:
        # Extract two smallest sticks
        first = heapq.heappop(sticks)
        second = heapq.heappop(sticks)
        
        # Combine them
        combined = first + second
        total_cost += combined
        
        # Put combined stick back
        heapq.heappush(sticks, combined)
    
    return total_cost
````

## Validation

````python
cases = [
    ([2, 4, 3], 14),           # Example 1
    ([1, 8, 3, 5], 30),        # Example 2
    ([5], 0),                  # Example 3: Single stick
    ([1, 1, 1, 1], 8),         # All equal: 1+1=2, 2+1=3, 3+1=4, total=2+3+4=9? No: (1+1)+(2+1)+(3+1)=2+3+4=9? Let me recalc
                                # Actually: 1+1=2(cost 2), 2+1=3(cost 3), 3+1=4(cost 4), but that's sequential
                                # Heap: [1,1,1,1] -> pop 1,1 cost=2, heap=[1,1,2] -> pop 1,1 cost=2, heap=[2,2] -> pop 2,2 cost=4, total=8
    ([1, 2], 3),               # Two sticks
    ([3, 4, 5, 6], 36),        # 3+4=7(cost 7), 5+6=11(cost 11), 7+11=18(cost 18), total=36
    ([1000], 0),               # Single large stick
    ([1, 2, 3, 4, 5], 33),     # Multiple small sticks
]
for sticks, expected in cases:
    result = connect_sticks(sticks.copy())  # Copy to avoid mutation
    assert result == expected, f"Failed for sticks={sticks}: got {result}, expected {expected}"
print('All tests passed for LC 1167.')
````

## Complexity Analysis
- **Time Complexity**: O(n log n) where n is the number of sticks
  - Initial heapify: O(n)
  - We perform n-1 merge operations
  - Each merge involves 2 heappops (O(log n) each) and 1 heappush (O(log n))
  - Total: O(n) + O((n-1) * 3 * log n) = O(n log n)
- **Space Complexity**: O(n) for the heap
- **Bottleneck**: Heap operations dominate the time complexity.

## Edge Cases & Pitfalls
- **Single stick**: Return 0 immediately (no merging needed).
- **Two sticks**: Single merge, cost is their sum.
- **All equal lengths**: Still need to use heap approach; answer is not trivial.
- **Large values**: Be aware of potential integer overflow in languages with fixed integer size.
- **Empty array**: Not possible according to constraints, but good to handle.
- **Heap mutation**: The input array is modified by heapify; make a copy if needed.

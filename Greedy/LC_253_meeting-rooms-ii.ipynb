## Problem Statement
Given an array of meeting time intervals `intervals` where `intervals[i] = [start_i, end_i]`, return the minimum number of conference rooms required.

**Note**: This is a LeetCode Premium problem. Examples are provided for clarity.

**Example 1:**
```
Input: intervals = [[0, 30], [5, 10], [15, 20]]
Output: 2
Explanation: 
- Meeting 1: [0, 30]
- Meeting 2: [5, 10] overlaps with meeting 1, needs a second room
- Meeting 3: [15, 20] overlaps with meeting 1, can use the room freed by meeting 2
- Maximum 2 rooms needed simultaneously
```

**Example 2:**
```
Input: intervals = [[7, 10], [2, 4]]
Output: 1
Explanation: The meetings don't overlap, so only 1 room is needed.
```

**Example 3:**
```
Input: intervals = [[1, 5], [2, 6], [3, 7], [4, 8]]
Output: 4
Explanation: All meetings overlap at time 4, requiring 4 rooms.
```

**Constraints:**
- `1 <= intervals.length <= 10^4`
- `0 <= start_i < end_i <= 10^6`

## Progressive Hints
- Hint 1: Think about what happens when a meeting starts vs. when a meeting ends.
- Hint 2: Use a min-heap to track the end times of ongoing meetings.
- Hint 3: Sort meetings by start time, and for each meeting, check if any room has become available.
- Hint 4: Alternatively, use two pointers: one for starts and one for ends, both sorted separately.

## Solution Overview
Two main approaches:

**Approach 1: Min-Heap**
1. Sort meetings by start time.
2. Use a min-heap to track end times of ongoing meetings.
3. For each meeting, remove ended meetings from heap.
4. Add current meeting's end time to heap.
5. Track maximum heap size.

**Approach 2: Two Pointers**
1. Separate and sort start times and end times.
2. Use two pointers to traverse both arrays.
3. When a meeting starts, increment room count.
4. When a meeting ends, decrement room count.
5. Track maximum room count.

## Detailed Explanation
**Min-Heap Approach:**
1. Sort intervals by start time: `intervals.sort(key=lambda x: x[0])`
2. Initialize min-heap with first meeting's end time.
3. For each subsequent meeting:
   - If heap top (earliest end time) <= current start time, pop from heap (room freed).
   - Push current meeting's end time to heap (room occupied).
   - Track max heap size as the answer.

**Two Pointers Approach:**
1. Create sorted arrays of start times and end times.
2. Use two pointers: start_ptr and end_ptr.
3. When start_times[start_ptr] < end_times[end_ptr], a meeting starts (rooms++).
4. Otherwise, a meeting ends first (rooms--).
5. Track maximum rooms used.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Min-Heap | O(n log n) | O(n) | Intuitive, uses heap for room management. |
| Two Pointers | O(n log n) | O(n) | More space efficient, same time complexity. |
| Brute force | O(n^2) | O(1) | Check overlap for each pair of meetings. |
| Sweep line | O(n log n) | O(n) | Event-based approach, similar to two pointers. |

## Reference Implementation

````python
from typing import List
import heapq


def min_meeting_rooms_heap(intervals: List[List[int]]) -> int:
    """
    Find minimum number of meeting rooms required using min-heap.
    
    Args:
        intervals: List of meeting intervals [start, end]
    
    Returns:
        Minimum number of conference rooms needed
    """
    if not intervals:
        return 0
    
    # Sort by start time
    intervals.sort(key=lambda x: x[0])
    
    # Min-heap to track end times of ongoing meetings
    heap = []
    heapq.heappush(heap, intervals[0][1])  # Add first meeting's end time
    
    for i in range(1, len(intervals)):
        # If earliest ending meeting finishes before current starts, free the room
        if heap[0] <= intervals[i][0]:
            heapq.heappop(heap)
        
        # Add current meeting's end time
        heapq.heappush(heap, intervals[i][1])
    
    # Heap size represents the number of rooms needed
    return len(heap)


def min_meeting_rooms_two_pointers(intervals: List[List[int]]) -> int:
    """
    Find minimum number of meeting rooms required using two pointers.
    
    Args:
        intervals: List of meeting intervals [start, end]
    
    Returns:
        Minimum number of conference rooms needed
    """
    if not intervals:
        return 0
    
    # Separate and sort start and end times
    start_times = sorted([interval[0] for interval in intervals])
    end_times = sorted([interval[1] for interval in intervals])
    
    start_ptr = 0
    end_ptr = 0
    rooms_used = 0
    max_rooms = 0
    
    while start_ptr < len(intervals):
        # A meeting is starting
        if start_times[start_ptr] < end_times[end_ptr]:
            rooms_used += 1
            max_rooms = max(max_rooms, rooms_used)
            start_ptr += 1
        else:
            # A meeting has ended, free up a room
            rooms_used -= 1
            end_ptr += 1
    
    return max_rooms
````

## Validation

````python
cases = [
    ([[0, 30], [5, 10], [15, 20]], 2),
    ([[7, 10], [2, 4]], 1),
    ([[1, 5], [2, 6], [3, 7], [4, 8]], 4),
    ([[1, 10], [2, 7], [3, 19], [8, 12], [10, 20], [11, 30]], 4),
    ([[1, 2]], 1),
    ([[1, 5], [8, 9], [10, 15]], 1),
]
for intervals, expected in cases:
    result1 = min_meeting_rooms_heap([i.copy() for i in intervals])
    result2 = min_meeting_rooms_two_pointers([i.copy() for i in intervals])
    assert result1 == expected, f"Heap approach failed for {intervals}: got {result1}, expected {expected}"
    assert result2 == expected, f"Two pointers approach failed for {intervals}: got {result2}, expected {expected}"
print('All tests passed for LC 253.')
````

## Complexity Analysis
**Min-Heap Approach:**
- **Time Complexity**: O(n log n)
  - Sorting: O(n log n)
  - Heap operations: O(n log n) for n push/pop operations
- **Space Complexity**: O(n) for the heap

**Two Pointers Approach:**
- **Time Complexity**: O(n log n)
  - Sorting start and end times: O(n log n)
  - Two pointer traversal: O(n)
- **Space Complexity**: O(n) for the separate start and end arrays

**Bottleneck**: Sorting dominates both approaches.

## Edge Cases & Pitfalls
- **Empty intervals**: Return 0.
- **Single meeting**: Always requires 1 room.
- **All meetings overlap**: Need n rooms for n meetings.
- **No overlaps**: Only need 1 room.
- **Back-to-back meetings**: If one ends exactly when another starts, same room can be used.
- **Heap underflow**: Ensure heap is not empty before accessing heap[0].
- **Integer boundaries**: Be careful with time values near boundaries.

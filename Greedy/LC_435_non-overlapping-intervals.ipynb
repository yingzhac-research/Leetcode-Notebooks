## Problem Statement
Given an array of intervals `intervals` where `intervals[i] = [starti, endi]`, return the minimum number of intervals you need to remove to make the rest of the intervals non-overlapping.

**Note**: You may assume the interval's end point is always bigger than its start point.

**Constraints:**
- `1 <= intervals.length <= 10^5`
- `intervals[i].length == 2`
- `-5 * 10^4 <= starti < endi <= 5 * 10^4`

## Progressive Hints
- Hint 1: Sort intervals by their end time. Why would this be helpful?
- Hint 2: Greedily select intervals that end earliest to maximize the number of non-overlapping intervals you can keep.
- Hint 3: Track the end time of the last selected interval. Skip intervals that overlap with it.

## Solution Overview
The optimal greedy approach:
1. Sort intervals by end time.
2. Iterate through sorted intervals, keeping track of the last non-overlapping interval's end.
3. If current interval doesn't overlap with the last kept interval, keep it.
4. Otherwise, skip it (count as removed).
5. Return the count of removed intervals.

**Why greedy works**: By always choosing the interval that ends earliest, we leave maximum room for future intervals, maximizing the number we can keep.

## Detailed Explanation
1. **Sort by end time**: `intervals.sort(key=lambda x: x[1])`
   - Sorting by end time ensures we can greedily select intervals that finish early

2. **Initialize tracking variables**:
   - `removals = 0`: Count of intervals to remove
   - `prev_end = float('-inf')`: End time of last kept interval

3. **Iterate through sorted intervals**:
   - If `start >= prev_end`: No overlap, keep this interval
     - Update `prev_end = end`
   - Else: Overlap detected, remove this interval
     - Increment `removals`

4. **Return removals**

**Example**: [[1,2], [2,3], [3,4], [1,3]]
- Sort by end: [[1,2], [2,3], [1,3], [3,4]]
- Keep [1,2], end=2
- Keep [2,3], end=3 (2 >= 2, no overlap)
- Remove [1,3] (1 < 3, overlap)
- Keep [3,4], end=4 (3 >= 3, no overlap)
- Removals: 1

## Reference Implementation

```python
from typing import List


def erase_overlap_intervals(intervals: List[List[int]]) -> int:
    """
    Find minimum number of intervals to remove to make rest non-overlapping.
    
    Args:
        intervals: List of intervals [start, end]
    
    Returns:
        Minimum number of intervals to remove
    """
    if not intervals:
        return 0
    
    # Sort by end time (greedy: keep intervals that end earliest)
    intervals.sort(key=lambda x: x[1])
    
    removals = 0
    prev_end = float('-inf')
    
    for start, end in intervals:
        # No overlap: start >= previous end
        if start >= prev_end:
            # Keep this interval
            prev_end = end
        else:
            # Overlap: remove this interval
            removals += 1
    
    return removals
```

## Validation

```python
cases = [
    ([[1,2],[2,3],[3,4],[1,3]], 1),  # Remove [1,3]
    ([[1,2],[1,2],[1,2]], 2),        # Remove 2 overlapping intervals
    ([[1,2],[2,3]], 0),              # No overlap (touching is OK)
    ([[1,100],[11,22],[1,11],[2,12]], 2),  # Remove 2
    ([[1,5],[2,3],[3,4],[4,5]], 2),  # Multiple overlaps
    ([[0,2],[1,3],[2,4],[3,5],[4,6]], 2),  # Chain of overlaps
    ([[-52,31],[-73,-26],[82,97],[-65,-11],[-62,-49],[95,99],[58,95],[-31,49],[66,98],[-63,2],[30,47],[-40,-26]], 7),
]
for intervals, expected in cases:
    result = erase_overlap_intervals([x[:] for x in intervals])  # Copy to avoid mutation
    assert result == expected, f"Failed for intervals={intervals}: got {result}, expected {expected}"
print('All tests passed for LC 435.')
```

## Complexity Analysis
- **Time Complexity**: O(n log n) where n is the number of intervals
  - Sorting takes O(n log n)
  - Single pass through intervals: O(n)
  - Overall dominated by sorting
- **Space Complexity**: O(1) if sorting in-place, O(n) if sorting creates a copy
  - Only constant extra space for variables
- **Bottleneck**: Sorting step is necessary and dominates.

## Edge Cases & Pitfalls
- **Empty array**: Return 0.
- **Single interval**: Return 0 (no removal needed).
- **All overlapping**: Keep one, remove n-1.
- **No overlaps**: Return 0.
- **Touching intervals**: [1,2] and [2,3] do NOT overlap (start >= prev_end is the condition).
- **Sorting stability**: Not required since we only care about count, not which specific intervals to remove.
- **Negative coordinates**: Handled correctly by comparisons.

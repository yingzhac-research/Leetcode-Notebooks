## Problem Statement
At a lemonade stand, each lemonade costs `$5`. Customers are standing in a queue to buy from you and order one at a time (in the order specified by bills). Each customer will only buy one lemonade and pay with either a `$5`, `$10`, or `$20` bill. You must provide the correct change to each customer so that the net transaction is that the customer pays `$5`.

Note that you do not have any change in hand at first.

Given an integer array `bills` where `bills[i]` is the bill the ith customer pays, return `true` if you can provide every customer with the correct change, or `false` otherwise.

**Constraints:**
- `1 <= bills.length <= 10^5`
- `bills[i]` is either `5`, `10`, or `20`.

## Progressive Hints
- Hint 1: Keep track of the number of $5 and $10 bills you have.
- Hint 2: When receiving a $10 bill, you need to give back one $5. When receiving a $20, you prefer to give back one $10 and one $5 rather than three $5s.
- Hint 3: Use a greedy strategy: always try to use larger bills first when making change.
- Hint 4: You never need to track $20 bills since they can't be used as change.

## Solution Overview
The optimal approach uses a greedy simulation:
1. Track the count of $5 and $10 bills.
2. For each customer:
   - If paying with $5, just keep it.
   - If paying with $10, give back one $5 (if available).
   - If paying with $20, prefer giving one $10 and one $5, otherwise three $5s.
3. Return false if at any point you can't make change.

**Why greedy works**: Preferring to give larger bills as change preserves smaller bills for future flexibility.

## Detailed Explanation
1. **Initialize counters**: `five_count = 0`, `ten_count = 0`
2. **For each bill in bills**:
   - If bill is $5: Increment five_count
   - If bill is $10:
     - Check if five_count > 0
     - Decrement five_count, increment ten_count
   - If bill is $20:
     - Prefer: one $10 + one $5 (if both available)
     - Otherwise: three $5s (if available)
     - Update counts accordingly
   - If can't make change, return False
3. **Return True** if all transactions succeed.

**Key insight**: $10 bills are more valuable than $5 bills for making change for $20, so we should save them when possible but use them when making change for $20.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy simulation | O(n) | O(1) | Optimal solution. |
| Backtracking | O(2^n) | O(n) | Try all change combinations. |
| DP | O(n * max_change) | O(max_change) | Overkill for this problem. |
| Brute force | O(n!) | O(n) | Try all orderings. |

## Reference Implementation

```python
from typing import List


def lemonade_change(bills: List[int]) -> bool:
    """
    Determine if we can provide correct change for all customers.
    
    Args:
        bills: List of bills paid by customers in order
    
    Returns:
        True if can make change for all customers, False otherwise
    """
    five_count = 0
    ten_count = 0
    
    for bill in bills:
        if bill == 5:
            # Just keep the $5 bill
            five_count += 1
        
        elif bill == 10:
            # Need to give back $5
            if five_count == 0:
                return False
            five_count -= 1
            ten_count += 1
        
        else:  # bill == 20
            # Need to give back $15
            # Prefer one $10 + one $5 over three $5s
            if ten_count > 0 and five_count > 0:
                ten_count -= 1
                five_count -= 1
            elif five_count >= 3:
                five_count -= 3
            else:
                return False
    
    return True
```

## Validation

```python
cases = [
    ([5, 5, 5, 10, 20], True),
    ([5, 5, 10, 10, 20], False),
    ([5, 5, 10], True),
    ([10, 10], False),
    ([5, 5, 5, 10, 5, 20, 5, 10, 5, 20], True),
    ([5], True),
    ([5, 5, 5, 5, 20, 20, 5, 5, 20, 5], False),
]
for bills, expected in cases:
    result = lemonade_change(bills)
    assert result == expected, f"Failed for bills={bills}: got {result}, expected {expected}"
print('All tests passed for LC 860.')
```

## Complexity Analysis
- **Time Complexity**: O(n) where n = len(bills)
  - Single pass through the bills array
  - Each transaction processed in constant time
- **Space Complexity**: O(1)
  - Only two counters needed (five_count, ten_count)
  - No additional data structures
- **Bottleneck**: Single pass through the array is optimal.

## Edge Cases & Pitfalls
- **All $5 bills**: Always return True.
- **Starting with large bill**: Cannot make change, return False.
- **Running out of $5s**: Critical since $5 is needed for all change.
- **Choosing change strategy**: Prefer $10+$5 over three $5s for $20.
- **Order matters**: Same bills in different order can have different outcomes.
- **No need to track $20s**: They're never used as change.

## Problem Statement
You are given an integer array `prices` where `prices[i]` is the price of a given stock on the ith day.

On each day, you may decide to buy and/or sell the stock. You can only hold at most one share of the stock at any time. However, you can buy it then immediately sell it on the same day.

Find and return the maximum profit you can achieve.

**Constraints:**
- `1 <= prices.length <= 3 * 10^4`
- `0 <= prices[i] <= 10^4`

## Progressive Hints
- Hint 1: If you can make unlimited transactions, what is the optimal strategy?
- Hint 2: Think about capturing every upward price movement. When prices increase from day i to day i+1, you can profit by buying at i and selling at i+1.
- Hint 3: Sum all positive differences between consecutive days. This greedy approach captures all profit opportunities.

## Solution Overview
The optimal greedy strategy is remarkably simple:
1. Iterate through consecutive days.
2. Whenever price increases from day i to day i+1, add the difference to profit.
3. This captures every upward price movement.

**Why greedy works**: Since we can make unlimited transactions with no transaction cost, we should capture every price increase. Even if prices rise continuously over multiple days, summing individual daily gains equals the total gain from buying at the valley and selling at the peak.

**Example**: If prices go [1, 3, 5], we gain:
- Greedy: (3-1) + (5-3) = 4
- Buy at 1, sell at 5: 5-1 = 4 (same result)

## Detailed Explanation
1. **Initialize profit = 0**

2. **Iterate from day 1 to n-1**:
   - Calculate `diff = prices[i] - prices[i-1]`
   - If `diff > 0`, add it to profit (capture upward movement)
   - If `diff <= 0`, ignore (don't trade during price drops)

3. **Return total profit**

**Mathematical insight**: 
```
max_profit = sum of all (prices[i] - prices[i-1]) where prices[i] > prices[i-1]
```

This is equivalent to buying at every local minimum and selling at every local maximum, but the greedy formulation is simpler to implement.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy (sum positive diffs) | O(n) | O(1) | Optimal and simplest solution. |
| Peak-Valley approach | O(n) | O(1) | Find all peaks and valleys explicitly. |
| Dynamic Programming | O(n) | O(n) or O(1) | Track states (holding/not holding). Overkill for this problem. |

## Reference Implementation

````python
from typing import List


def max_profit(prices: List[int]) -> int:
    """
    Calculate maximum profit from unlimited stock transactions.
    
    Args:
        prices: Daily stock prices
    
    Returns:
        Maximum profit achievable
    """
    profit = 0
    
    # Sum all positive price differences between consecutive days
    for i in range(1, len(prices)):
        # Capture every upward price movement
        if prices[i] > prices[i - 1]:
            profit += prices[i] - prices[i - 1]
    
    return profit


# Alternative one-liner using max
def max_profit_oneliner(prices: List[int]) -> int:
    """Concise version using max to handle negative differences."""
    return sum(max(prices[i] - prices[i - 1], 0) for i in range(1, len(prices)))
````

## Validation

````python
cases = [
    ([7, 1, 5, 3, 6, 4], 7),  # Buy at 1, sell at 5 (+4); buy at 3, sell at 6 (+3)
    ([1, 2, 3, 4, 5], 4),     # Buy at 1, sell at 5, or capture all daily gains
    ([7, 6, 4, 3, 1], 0),     # Only declining prices, no profit
    ([1], 0),                 # Single day, no transaction possible
    ([2, 1, 2, 0, 1], 2),     # Multiple small transactions
    ([3, 3, 5, 0, 0, 3, 1, 4], 8),  # Complex pattern
]
for prices, expected in cases:
    result = max_profit(prices)
    assert result == expected, f"Failed for prices={prices}: got {result}, expected {expected}"
    # Also test the one-liner version
    result2 = max_profit_oneliner(prices)
    assert result2 == expected, f"One-liner failed for prices={prices}: got {result2}, expected {expected}"
print('All tests passed for LC 122.')
````

## Complexity Analysis
- **Time Complexity**: O(n) where n is the number of days
  - Single pass through the array
  - Constant time operation at each step
- **Space Complexity**: O(1)
  - Only storing profit variable
- **Bottleneck**: Must examine all prices to find all profit opportunities.

## Edge Cases & Pitfalls
- **Single element array**: Return 0 (no transaction possible).
- **Monotonically decreasing prices**: Return 0 (no profit opportunity).
- **Monotonically increasing prices**: Sum of all differences equals (last - first).
- **Consecutive equal prices**: Differences are 0, no contribution to profit.
- **Integer overflow**: In other languages, check if prices are very large and use appropriate data types.

- When you can make unlimited transactions, capture every upward price movement.

- Sum of consecutive differences is a powerful pattern for capturing all incremental gains.
- This problem demonstrates when greedy is optimal: when local choices don't constrain future choices.

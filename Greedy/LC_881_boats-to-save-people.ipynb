## Problem Statement
You are given an array `people` where `people[i]` is the weight of the ith person, and an infinite number of boats where each boat can carry a maximum weight of `limit`. Each boat carries at most two people at the same time, provided the sum of the weight of those people is at most `limit`.

Return the minimum number of boats to carry every given person.

**Constraints:**
- `1 <= people.length <= 5 * 10^4`
- `1 <= people[i] <= limit <= 3 * 10^4`

## Progressive Hints
- Hint 1: Try sorting the people by weight. How does this help?
- Hint 2: Use two pointers: one at the lightest person and one at the heaviest.
- Hint 3: Greedily try to pair the heaviest person with the lightest person. If they fit together, move both pointers inward.
- Hint 4: If the heaviest and lightest can't fit together, the heaviest must go alone.

## Solution Overview
The optimal approach uses a greedy two-pointer strategy:
1. Sort people by weight.
2. Use two pointers: left (lightest) and right (heaviest).
3. Try to pair the heaviest with the lightest:
   - If they fit together, move both pointers inward.
   - If not, the heaviest goes alone, move only right pointer.
4. Count each boat used.

**Why greedy works**: Pairing the heaviest with the lightest maximizes the chance of fitting two people per boat, minimizing total boats needed.

## Detailed Explanation
1. **Sort people array**: `people.sort()`
2. **Initialize two pointers**: `left = 0`, `right = len(people) - 1`
3. **Initialize boat count**: `boats = 0`
4. **While left <= right**:
   - If `people[left] + people[right] <= limit`:
     - Both can share a boat, increment left
   - The heaviest person always gets on a boat (alone or paired)
   - Decrement right, increment boats
5. **Return boats count**.

**Key insight**: The heaviest person will always take a boat. The only question is whether they can share it with the lightest remaining person.

## Complexity Trade-off Table

## Reference Implementation

```python
from typing import List


def num_rescue_boats(people: List[int], limit: int) -> int:
    """
    Find minimum number of boats needed to rescue all people.
    
    Args:
        people: List of people's weights
        limit: Maximum weight capacity per boat
    
    Returns:
        Minimum number of boats required
    """
    # Sort people by weight
    people.sort()
    
    left = 0  # Lightest person
    right = len(people) - 1  # Heaviest person
    boats = 0
    
    while left <= right:
        # Try to pair the lightest with the heaviest
        if people[left] + people[right] <= limit:
            left += 1  # Lightest person gets on the boat
        
        # Heaviest person always gets on a boat (alone or with lightest)
        right -= 1
        boats += 1
    
    return boats
```

## Validation

```python
cases = [
    ([1, 2], 3, 1),
    ([3, 2, 2, 1], 3, 3),
    ([3, 5, 3, 4], 5, 4),
    ([5, 1, 4, 2], 6, 2),
    ([1, 1, 1, 1], 2, 2),
    ([3, 3, 3, 3], 3, 4),
    ([2, 2], 6, 1),
]
for people, limit, expected in cases:
    result = num_rescue_boats(people.copy(), limit)
    assert result == expected, f"Failed for people={people}, limit={limit}: got {result}, expected {expected}"
print('All tests passed for LC 881.')
```

## Complexity Analysis
- **Time Complexity**: O(n log n) where n = len(people)
  - Sorting takes O(n log n)
  - Two-pointer traversal takes O(n)
  - Overall dominated by sorting
- **Space Complexity**: O(1) if sorting in-place
  - Only constant extra space for pointers and counter
- **Bottleneck**: Sorting is the limiting factor.

## Edge Cases & Pitfalls
- **Everyone fits in pairs**: Minimum boats = n/2.
- **Everyone goes alone**: When all weights > limit/2, minimum boats = n.
- **Equal weights**: People with equal weights may or may not pair depending on limit.
- **Single person**: Always requires 1 boat.
- **Exact limit**: When two people exactly match the limit, they can share.
- **Sorting side effects**: The input array is modified; copy if needed.

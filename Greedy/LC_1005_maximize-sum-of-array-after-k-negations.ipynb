## Problem Statement
Given an integer array `nums` and an integer `k`, modify the array in the following way:

- Choose an index `i` and replace `nums[i]` with `-nums[i]`.

You should apply this process exactly `k` times. You may choose the same index `i` multiple times.

Return the largest possible sum of the array after modifying it in this way.

**Constraints:**
- `1 <= nums.length <= 10^4`
- `1 <= k <= 10^4`
- `-100 <= nums[i] <= 100`

## Progressive Hints
- Hint 1: To maximize the sum, you want to make negative numbers positive. Which negative numbers should you flip first?
- Hint 2: Always flip the smallest (most negative) number first to maximize the increase in sum.
- Hint 3: After flipping all negative numbers, if you still have operations left, what happens?
- Hint 4: If k is odd after flipping all negatives, you must flip one number. Which one minimizes the loss?

## Solution Overview
The optimal approach uses a greedy strategy:
1. Sort the array to easily find the smallest (most negative) values.
2. Flip negative numbers from smallest to largest until they're all positive or k is exhausted.
3. If k is still positive and odd, flip the smallest absolute value in the array (to minimize loss).
4. Return the sum of the modified array.

**Why greedy works**: Flipping the most negative number gives the maximum increase in sum at each step. After all negatives are positive, flipping any number decreases the sum, so we minimize damage by flipping the smallest absolute value.

## Detailed Explanation
1. **Sort the array**: This allows us to process the most negative numbers first.
2. **Flip negatives greedily**:
   - Iterate through sorted array
   - For each negative number, if k > 0, flip it and decrement k
   - Stop when we encounter a non-negative or k becomes 0
3. **Handle remaining k**:
   - If k is even, we can flip the same number back and forth, resulting in no net change
   - If k is odd, we must flip one number once, so flip the smallest absolute value to minimize loss
4. **Calculate sum**: Sum all elements in the modified array.

**Algorithm flow**:
```
nums.sort()

# Flip negatives
for i in range(n):
    if k > 0 and nums[i] < 0:
        nums[i] = -nums[i]
        k -= 1

# If k is odd, flip the smallest absolute value
if k % 2 == 1:
    nums.sort()  # Re-sort to find smallest
    nums[0] = -nums[0]

return sum(nums)
```

**Example walkthrough** for `nums = [4,-2,-3,1], k = 4`:
- Sorted: [-3, -2, 1, 4]
- Flip -3 → [3, -2, 1, 4], k = 3
- Flip -2 → [3, 2, 1, 4], k = 2
- All positive now, k = 2 (even)
- k is even, so no more changes needed
- Sum = 3 + 2 + 1 + 4 = 10

## Reference Implementation

````python
from typing import List


def largest_sum_after_k_negations(nums: List[int], k: int) -> int:
    """
    Maximize array sum after exactly k negations.
    
    Args:
        nums: Array of integers
        k: Number of negations to perform
    
    Returns:
        Maximum possible sum after k negations
    """
    # Sort to process most negative first
    nums.sort()
    
    # Flip negative numbers greedily
    for i in range(len(nums)):
        if k > 0 and nums[i] < 0:
            nums[i] = -nums[i]
            k -= 1
        else:
            break
    
    # If k is odd, we must flip the smallest absolute value
    if k % 2 == 1:
        # Find the smallest absolute value
        nums.sort()  # Re-sort after negations
        nums[0] = -nums[0]
    
    return sum(nums)
````

## Validation

````python
cases = [
    ([4, 2, 3], 1, 5),  # Flip 2 → [4, -2, 3], sum = 5
    ([3, -1, 0, 2], 3, 6),  # Flip -1 → [3, 1, 0, 2], then flip 0 twice, sum = 6
    ([2, -3, -1, 5, -4], 2, 13),  # Flip -4 and -3 → [2, 3, -1, 5, 4], sum = 13
    ([-2, 5, 0, 2, -2], 3, 11),  # Flip both -2s and 0, sum = 11
    ([5, 6, 9, -3, 3], 2, 20),  # Flip -3 then 3, sum = 20
]

for nums, k, expected in cases:
    result = largest_sum_after_k_negations(nums.copy(), k)
    assert result == expected, f"Failed for nums={nums}, k={k}: got {result}, expected {expected}"

print('All tests passed for LC 1005.')
````

## Complexity Analysis
- **Time Complexity**: O(n log n) where n is the length of nums
  - Initial sort: O(n log n)
  - First pass to flip negatives: O(n)
  - Second sort (if needed): O(n log n)
  - Sum calculation: O(n)
  - Overall: O(n log n)
- **Space Complexity**: O(1) for in-place sorting (or O(n) depending on sort implementation)
- **Bottleneck**: The sorting operations dominate the time complexity.

## Edge Cases & Pitfalls
- **All positive numbers**: If k is odd, flip the smallest; if even, no net change.
- **All negative numbers**: Flip from smallest to largest until k is exhausted.
- **k larger than negatives**: After flipping all negatives, handle remaining k based on parity.
- **Zero in array**: Zero is optimal to flip if k is odd and all other numbers are positive.
- **Forgetting to re-sort**: After flipping negatives, the array is no longer sorted; need to re-sort to find the smallest absolute value.
- **Even vs odd k**: Critical distinction for handling leftover operations.

## Takeaways
- Greedy algorithms often involve sorting to establish a natural processing order.
- Pay attention to parity (even/odd) in problems with repeated operations.
- When forced to make a detrimental choice, minimize the damage (flip smallest absolute value).
- After modifying sorted data, remember to re-sort if needed for subsequent steps.
- The greedy choice property: making locally optimal decisions (flip most negative) leads to global optimum.

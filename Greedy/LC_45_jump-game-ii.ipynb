## Problem Statement
You are given a 0-indexed array of integers `nums` of length `n`. You are initially positioned at `nums[0]`.

Each element `nums[i]` represents the maximum length of a forward jump from index `i`. In other words, if you are at `nums[i]`, you can jump to any `nums[i + j]` where:
- `0 <= j <= nums[i]`
- `i + j < n`

Return the minimum number of jumps to reach `nums[n - 1]`. The test cases are generated such that you can reach `nums[n - 1]`.

**Constraints:**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 1000`
- It's guaranteed that you can reach `nums[n - 1]`.

## Progressive Hints
- Hint 1: Think of this as a BFS problem where each level represents the positions reachable in the same number of jumps.
- Hint 2: Instead of exploring all positions at each level, track the farthest position you can reach within the current jump count.
- Hint 3: Greedy strategy: At each position, track the farthest you can reach in the next jump, and increment jump count when you reach the boundary of the current jump range.

## Solution Overview
The optimal approach uses a greedy strategy that simulates BFS without explicit queue:
1. Track the current jump's boundary (`current_end`) and the farthest position reachable in the next jump (`farthest`).
2. Iterate through the array, updating `farthest` at each position.
3. When reaching `current_end`, increment jump count and update `current_end` to `farthest`.
4. This ensures we make the minimum number of jumps.

**Alternative**: Dynamic Programming with O(n^2) time, but greedy is more efficient with O(n).

## Detailed Explanation
1. **Initialize variables**:
   - `jumps = 0`: Count of jumps taken
   - `current_end = 0`: Boundary of current jump range
   - `farthest = 0`: Farthest position reachable in next jump

2. **Iterate through array** (stop before last index since we start there):
   - Update `farthest = max(farthest, i + nums[i])`
   - If `i == current_end` (reached boundary of current jump):
     - Increment `jumps`
     - Set `current_end = farthest` (prepare for next jump)

3. **Return jumps**: The minimum number needed.

**Greedy insight**: At each boundary, we've explored all positions reachable in current jumps, and we know the farthest we can go in the next jump. This guarantees minimum jumps.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy (implicit BFS) | O(n) | O(1) | Optimal solution. |
| Dynamic Programming | O(n^2) | O(n) | dp[i] = min jumps to reach i. |
| Explicit BFS | O(n) | O(n) | Uses queue, same time but more space. |

## Reference Implementation

```python
from typing import List


def jump(nums: List[int]) -> int:
    """
    Find minimum number of jumps to reach the last index.
    
    Args:
        nums: Array where nums[i] is max jump length from index i
    
    Returns:
        Minimum number of jumps to reach last index
    """
    n = len(nums)
    if n <= 1:
        return 0
    
    jumps = 0
    current_end = 0  # Boundary of current jump range
    farthest = 0     # Farthest position reachable in next jump
    
    # We don't need to jump from the last position
    for i in range(n - 1):
        # Update the farthest position we can reach
        farthest = max(farthest, i + nums[i])
        
        # If we've reached the end of current jump range
        if i == current_end:
            jumps += 1
            current_end = farthest  # Start next jump range
            
            # Early exit if we can already reach the end
            if current_end >= n - 1:
                break
    
    return jumps
```

## Validation

```python
cases = [
    ([2, 3, 1, 1, 4], 2),  # Jump 1 step from index 0 to 1, then 3 steps to the last index
    ([2, 3, 0, 1, 4], 2),  # Same as above
    ([1, 2, 3], 2),        # 0 -> 1 -> 3
    ([1, 1, 1, 1], 3),     # Three jumps
    ([1], 0),              # Already at last index
    ([0], 0),              # Single element
    ([5, 9, 3, 2, 1, 0, 2, 3, 3, 1, 0, 0], 3),  # Can jump directly or in steps
]
for nums, expected in cases:
    result = jump(nums)
    assert result == expected, f"Failed for nums={nums}: got {result}, expected {expected}"
print('All tests passed for LC 45.')
```

## Complexity Analysis
- **Time Complexity**: O(n) where n is the length of the array
  - Single pass through the array
  - Each element processed exactly once
- **Space Complexity**: O(1)
  - Only constant extra space for variables
- **Bottleneck**: Linear scan is unavoidable as we must consider all positions to find optimal path.

## Edge Cases & Pitfalls
- **Single element array**: Return 0 immediately (already at destination).
- **Already reachable**: If first element can reach the end, return 1.
- **Off-by-one errors**: Be careful with loop boundary (iterate to n-1, not n).
- **Guaranteed reachability**: Problem states solution always exists, but in real scenarios, check if `farthest` stops advancing.
- **Zero values**: Positions with 0 are okay as long as we can jump over them.

## Takeaways
- Greedy algorithms can simulate BFS implicitly by tracking range boundaries.
- The key insight is recognizing that all positions within a jump range are equivalent for counting purposes.
- Tracking boundaries (current_end, farthest) allows us to avoid explicit queue/level tracking.
- This pattern applies to other range-based optimization problems.

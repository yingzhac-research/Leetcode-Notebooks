## Problem Statement
You are given an array of integers `stones` where `stones[i]` is the weight of the `i-th` stone.

We are playing a game with the stones. On each turn, we choose the **heaviest two stones** and smash them together. Suppose the heaviest two stones have weights `x` and `y` with `x <= y`. The result of this smash is:

- If `x == y`, both stones are destroyed, and
- If `x != y`, the stone of weight `x` is destroyed, and the stone of weight `y` has new weight `y - x`.

At the end of the game, there is **at most one stone left**.

Return the weight of the last remaining stone. If there are no stones left, return `0`.

**Constraints:**
- `1 <= stones.length <= 30`
- `1 <= stones[i] <= 1000`

## Progressive Hints
- Hint 1: You need to repeatedly find and remove the two largest elements. What data structure allows efficient retrieval of the maximum?
- Hint 2: A max heap (priority queue) can give you the largest element in O(log n) time.
- Hint 3: Python's heapq is a min heap, but you can simulate a max heap by negating all values.
- Hint 4: Continue the process until you have 0 or 1 stones left.

## Solution Overview
The optimal approach uses a max heap:
1. Convert the stones array into a max heap.
2. While there are at least 2 stones:
   - Pop the two heaviest stones
   - If they differ, push the difference back to the heap
3. Return the last stone's weight (or 0 if heap is empty).

**Why greedy works**: We always smash the two heaviest stones, which is the optimal local choice at each step. The order doesn't affect the final result because stone collisions are associative in terms of the final outcome.

## Detailed Explanation
1. **Initialize max heap**: Convert the stones array into a max heap. In Python, use negative values with heapq to simulate max heap.
2. **Simulation loop**:
   - While heap has at least 2 elements:
     - Pop the first largest stone (y)
     - Pop the second largest stone (x)
     - If x != y, push (y - x) back to the heap
3. **Return result**: If heap is empty, return 0; otherwise, return the remaining stone's weight.

**Algorithm flow**:
```
heap = [-stone for stone in stones]
heapify(heap)

while len(heap) > 1:
    y = -heappop(heap)  # heaviest
    x = -heappop(heap)  # second heaviest
    if x != y:
        heappush(heap, -(y - x))

return -heap[0] if heap else 0
```

**Example walkthrough** for `stones = [2,7,4,1,8,1]`:
- Heap (as max): [8,7,4,2,1,1]
- Smash 8 and 7 → difference 1, heap: [4,2,1,1,1]
- Smash 4 and 2 → difference 2, heap: [2,1,1,1]
- Smash 2 and 1 → difference 1, heap: [1,1,1]
- Smash 1 and 1 → both destroyed, heap: [1]
- Result: 1

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Max heap | O(n log n) | O(n) | Optimal solution. |
| Sorting repeatedly | O(n^2 log n) | O(1) | Sort after each operation. Too slow. |
| Brute force | O(n^2) | O(1) | Find max twice in each iteration. Slower than heap. |

## Reference Implementation

````python
import heapq
from typing import List


def last_stone_weight(stones: List[int]) -> int:
    """
    Find the weight of the last remaining stone after smashing.
    
    Args:
        stones: List of stone weights
    
    Returns:
        Weight of the last stone, or 0 if none remain
    """
    # Create max heap using negative values
    heap = [-stone for stone in stones]
    heapq.heapify(heap)
    
    # Keep smashing until 0 or 1 stones remain
    while len(heap) > 1:
        # Pop two heaviest stones
        y = -heapq.heappop(heap)  # heaviest
        x = -heapq.heappop(heap)  # second heaviest
        
        # If different weights, push the difference back
        if x != y:
            heapq.heappush(heap, -(y - x))
    
    # Return last stone weight or 0 if all destroyed
    return -heap[0] if heap else 0
````

## Validation

````python
cases = [
    ([2, 7, 4, 1, 8, 1], 1),
    ([1], 1),
    ([2, 2], 0),
    ([1, 3], 2),
    ([7, 6, 7, 6, 9], 3),
    ([3, 7, 2], 2),
]

for stones, expected in cases:
    result = last_stone_weight(stones.copy())
    assert result == expected, f"Failed for stones={stones}: got {result}, expected {expected}"

print('All tests passed for LC 1046.')
````

## Complexity Analysis
- **Time Complexity**: O(n log n) where n is the number of stones
  - Heapify: O(n)
  - Each stone is pushed/popped at most once: O(n log n)
  - Overall: O(n log n)
- **Space Complexity**: O(n) for the heap
- **Bottleneck**: The heap operations (push/pop) dominate the time complexity.

## Edge Cases & Pitfalls
- **Single stone**: Return that stone's weight.
- **All stones same weight**: They all cancel out, return 0.
- **Two stones**: Return their difference.
- **Heap sign**: Remember to negate when using Python's min heap as max heap.
- **Empty heap at end**: When all stones cancel out, heap becomes empty; return 0.

## Takeaways
- Max heaps are ideal for problems requiring repeated extraction of maximum elements.
- In Python, simulate max heap by negating values with heapq (min heap).
- Greedy simulation problems often have a natural ordering that suggests using a priority queue.
- Always consider whether the order of operations affects the final result (in this case, it doesn't).
- Heap operations (push, pop) are O(log n), making them efficient for dynamic data.

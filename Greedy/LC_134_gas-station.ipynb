## Problem Statement
There are `n` gas stations along a circular route, where the amount of gas at the ith station is `gas[i]`.

You have a car with an unlimited gas tank and it costs `cost[i]` of gas to travel from the ith station to its next (i + 1)th station. You begin the journey with an empty tank at one of the gas stations.

Given two integer arrays `gas` and `cost`, return the starting gas station's index if you can travel around the circuit once in the clockwise direction, otherwise return -1. If there exists a solution, it is guaranteed to be unique.

**Constraints:**
- `n == gas.length == cost.length`
- `1 <= n <= 10^5`
- `0 <= gas[i], cost[i] <= 10^4`

## Progressive Hints
- Hint 1: If total gas < total cost, it's impossible to complete the circuit.
- Hint 2: If you can't reach station j from station i, then you can't reach j from any station between i and j-1.
- Hint 3: Track cumulative gas balance. When it becomes negative, reset the starting point to the next station.

## Solution Overview
The optimal greedy approach uses a single pass:
1. Check if total gas >= total cost (necessary condition).
2. Track current gas balance while iterating.
3. When balance becomes negative, it means we can't start from any previous station to reach here.
4. Reset start to next station and reset current balance.
5. If total gas >= total cost and we found a candidate start, it's the answer.

**Key insight**: If starting from station i, we run out of gas before station j, then any station between i and j-1 also can't reach j. So we can skip them all and try starting from j.

## Detailed Explanation
1. **Check feasibility**:
   ```python
   if sum(gas) < sum(cost):
       return -1
   ```

2. **Greedy search for starting point**:
   - Initialize: `start = 0`, `total_tank = 0`, `current_tank = 0`
   - For each station i:
     - `total_tank += gas[i] - cost[i]`
     - `current_tank += gas[i] - cost[i]`
     - If `current_tank < 0`:
       - Can't start from any station up to i
       - Reset: `start = i + 1`, `current_tank = 0`

3. **Return result**:
   - If `total_tank >= 0`, return `start`
   - Otherwise, return -1

**Why this works**:
- If total gas >= total cost, a solution exists.
- The greedy reset ensures we find the rightmost valid starting point.
- Since solution is unique, this is the only valid start.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| One-pass greedy | O(n) | O(1) | Optimal solution. |
| Try each starting point | O(n^2) | O(1) | Simulate full circuit from each station. |
| Prefix sum with two passes | O(n) | O(n) | Track cumulative balance, find minimum. |

## Reference Implementation

````python
from typing import List


def can_complete_circuit(gas: List[int], cost: List[int]) -> int:
    """
    Find starting gas station index to complete circuit, or -1 if impossible.
    
    Args:
        gas: Gas available at each station
        cost: Cost to travel to next station
    
    Returns:
        Starting station index, or -1 if impossible
    """
    n = len(gas)
    total_tank = 0    # Total gas surplus/deficit for entire circuit
    current_tank = 0  # Current gas in tank
    start = 0         # Candidate starting station
    
    for i in range(n):
        # Net gas change at station i
        delta = gas[i] - cost[i]
        total_tank += delta
        current_tank += delta
        
        # If current tank goes negative, can't start from any previous station
        if current_tank < 0:
            # Try starting from next station
            start = i + 1
            current_tank = 0
    
    # If total gas >= total cost, solution exists at 'start'
    return start if total_tank >= 0 else -1
````

## Validation

````python
cases = [
    ([1,2,3,4,5], [3,4,5,1,2], 3),  # Start at index 3
    ([2,3,4], [3,4,3], -1),         # Impossible
    ([5,1,2,3,4], [4,4,1,5,1], 4),  # Start at index 4
    ([1], [1], 0),                  # Single station, break even
    ([1], [2], -1),                 # Single station, not enough
    ([2], [1], 0),                  # Single station, surplus
    ([3,3,4], [3,4,4], 0),          # Multiple valid, return first found
]
for gas, cost, expected in cases:
    result = can_complete_circuit(gas, cost)
    assert result == expected, f"Failed for gas={gas}, cost={cost}: got {result}, expected {expected}"
print('All tests passed for LC 134.')
````

## Complexity Analysis
- **Time Complexity**: O(n) where n is the number of stations
  - Single pass through all stations
  - Constant time operations at each station
- **Space Complexity**: O(1)
  - Only constant extra space for variables
- **Bottleneck**: Must examine all stations to compute total balance and find starting point.

## Edge Cases & Pitfalls
- **Single station**: Return 0 if gas[0] >= cost[0], else -1.
- **Impossible circuit**: Total gas < total cost.
- **All stations break even**: Any starting point works; algorithm returns 0.
- **Starting point at end**: Algorithm correctly handles this (start = n-1 or start = 0).
- **Multiple valid starting points**: Problem guarantees unique solution.
- **Off-by-one errors**: Make sure to reset start to i+1, not i.

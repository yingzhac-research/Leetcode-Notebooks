## Problem Statement
Given string `num` representing a non-negative integer, and an integer `k`, return the smallest possible integer after removing `k` digits from the number.

**Constraints:**
- `1 <= k <= num.length <= 10^5`
- `num` consists of only digits.
- `num` does not have any leading zeros except for the zero itself.

## Progressive Hints
- Hint 1: To make the number smallest, you want higher-order digits to be as small as possible.
- Hint 2: Use a stack to build the result. When you encounter a digit smaller than the stack top, you should remove the larger digit (if you still have removals left).
- Hint 3: After processing all digits, if you haven't removed k digits yet, remove from the end. Also handle leading zeros.

## Solution Overview
The optimal approach uses a monotonic stack (increasing) with greedy removal:
1. Iterate through digits left to right.
2. For each digit, while it's smaller than the stack top and we still have removals left, pop the stack.
3. Push the current digit.
4. After iteration, if removals remain, remove from the end.
5. Remove leading zeros and return the result (or "0" if empty).

**Why greedy works**: Removing a larger digit from a higher position always produces a smaller result than removing a smaller digit or a digit from a lower position.

## Detailed Explanation
1. **Initialize**:
   - `stack = []` to build the result
   - `removals_left = k`

2. **Process each digit**:
   - While stack is not empty, stack top > current digit, and removals_left > 0:
     - Pop from stack (remove larger digit)
     - Decrement removals_left
   - Push current digit to stack

3. **Handle remaining removals**:
   - If removals_left > 0, remove that many digits from the end of stack
   - This handles cases like "12345" where digits are already increasing

4. **Build result**:
   - Join stack to form number string
   - Remove leading zeros using lstrip('0')
   - Return "0" if result is empty

**Example**: num = "1432219", k = 3
- Process '1': stack = ['1']
- Process '4': stack = ['1', '4']
- Process '3': 3 < 4, pop '4', stack = ['1', '3']
- Process '2': 2 < 3, pop '3', stack = ['1', '2']
- Process '2': stack = ['1', '2', '2']
- Process '1': 1 < 2, pop '2', stack = ['1', '2', '1'] (k removals done)
- Process '9': stack = ['1', '2', '1', '9']
- Result: "1219"

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Monotonic stack | O(n) | O(n) | Optimal solution. Each digit pushed/popped at most once. |
| Greedy without stack | O(n * k) | O(n) | Repeatedly find and remove largest digit. |
| Backtracking | O(C(n, k)) | O(k) | Try all combinations. Exponential time. |

## Reference Implementation

```python
def remove_k_digits(num: str, k: int) -> str:
    """
    Remove k digits from num to get the smallest possible number.
    
    Args:
        num: String representing a non-negative integer
        k: Number of digits to remove
    
    Returns:
        Smallest possible number after removing k digits
    """
    stack = []
    removals_left = k
    
    for digit in num:
        # Remove larger digits from stack while we can
        while stack and stack[-1] > digit and removals_left > 0:
            stack.pop()
            removals_left -= 1
        stack.append(digit)
    
    # If we still need to remove digits, remove from the end
    # (handles cases where digits are in increasing order)
    if removals_left > 0:
        stack = stack[:-removals_left]
    
    # Build result and remove leading zeros
    result = ''.join(stack).lstrip('0')
    
    # Return "0" if result is empty
    return result if result else '0'
```

## Validation

```python
cases = [
    ("1432219", 3, "1219"),
    ("10200", 1, "200"),
    ("10", 2, "0"),
    ("9", 1, "0"),
    ("112", 1, "11"),
    ("1234567890", 9, "0"),
    ("10001", 4, "0"),
    ("54321", 2, "321"),  # All decreasing, remove from front
    ("12345", 2, "123"),  # All increasing, remove from end
    ("100", 1, "0"),
]
for num, k, expected in cases:
    result = remove_k_digits(num, k)
    assert result == expected, f"Failed for num={num}, k={k}: got {result}, expected {expected}"
print('All tests passed for LC 402.')
```

## Complexity Analysis
- **Time Complexity**: O(n) where n is the length of num
  - Each digit is pushed to stack exactly once: O(n)
  - Each digit is popped from stack at most once: O(n)
  - Building result string: O(n)
  - Overall: O(n)
- **Space Complexity**: O(n)
  - Stack can hold up to n digits
- **Bottleneck**: The string operations (join, lstrip) are linear but necessary.

## Edge Cases & Pitfalls
- **All digits removed**: Return "0" instead of empty string.
- **Leading zeros**: Must be removed (e.g., "10200" removing 1 digit gives "200" not "0200").
- **k equals length**: Remove all digits, return "0".
- **Monotonically increasing number**: Remove from end (e.g., "12345", k=2 -> "123").
- **Monotonically decreasing number**: Remove from start (e.g., "54321", k=2 -> "321").
- **Single digit**: If k >= 1, result is "0".

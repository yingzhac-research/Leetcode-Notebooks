## Problem Statement
You are given an integer array `nums`. You are initially positioned at the array's first index, and each element in the array represents your maximum jump length at that position.

Return `true` if you can reach the last index, or `false` otherwise.

**Constraints:**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 10^5`

## Progressive Hints
- Hint 1: Think about tracking the farthest position you can reach at any point.
- Hint 2: Use a greedy approach: iterate through the array and update the maximum reachable position.
- Hint 3: If at any point your current index exceeds the maximum reachable position, you cannot proceed further.
- Hint 4: You only need to check if the last index is within the maximum reachable position.

## Solution Overview
The optimal approach uses a greedy strategy:
1. Track the maximum reachable index as you iterate through the array.
2. For each position, update the maximum reachable index based on the jump length.
3. If the current position is beyond the maximum reachable index, return false.
4. If you can reach or exceed the last index, return true.

**Why greedy works**: At each position, we only care about the farthest point we can reach, not the specific path to get there. This greedy choice is sufficient to determine reachability.

## Detailed Explanation
1. **Initialize max_reach**: Set to 0, representing the farthest index reachable so far.
2. **Iterate through the array**:
   - For each index `i`, check if `i > max_reach`. If so, return false (unreachable).
   - Update `max_reach = max(max_reach, i + nums[i])`.
   - If `max_reach >= len(nums) - 1`, return true early.
3. **Return true**: If loop completes, the last index is reachable.

**Key insight**: We don't need to explore all possible paths. We only need to know if the last index falls within the maximum reachable range.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy (forward) | O(n) | O(1) | Optimal solution. |
| Greedy (backward) | O(n) | O(1) | Alternative greedy approach. |
| Dynamic Programming | O(n^2) | O(n) | Overkill for this problem. |
| BFS | O(n^2) | O(n) | Explores all possible jumps. |

## Reference Implementation

```python
from typing import List


def can_jump(nums: List[int]) -> bool:
    """
    Determine if you can reach the last index using greedy approach.
    
    Args:
        nums: List of maximum jump lengths at each position
    
    Returns:
        True if last index is reachable, False otherwise
    """
    max_reach = 0  # Farthest index we can reach
    
    for i in range(len(nums)):
        # If current position is beyond max reach, we're stuck
        if i > max_reach:
            return False
        
        # Update the farthest position we can reach
        max_reach = max(max_reach, i + nums[i])
        
        # Early termination: if we can already reach the end
        if max_reach >= len(nums) - 1:
            return True
    
    return True
```

## Validation

```python
cases = [
    ([2, 3, 1, 1, 4], True),
    ([3, 2, 1, 0, 4], False),
    ([0], True),
    ([2, 0, 0], True),
    ([1, 1, 1, 0], True),
    ([1, 0, 1, 0], False),
    ([2, 5, 0, 0], True),
]
for nums, expected in cases:
    result = can_jump(nums)
    assert result == expected, f"Failed for nums={nums}: got {result}, expected {expected}"
print('All tests passed for LC 55.')
```

## Complexity Analysis
- **Time Complexity**: O(n) where n = len(nums)
  - Single pass through the array
  - Each element is visited at most once
- **Space Complexity**: O(1)
  - Only using a constant amount of extra space (max_reach variable)
- **Bottleneck**: The single array traversal is necessary to determine reachability.

## Edge Cases & Pitfalls
- **Single element array**: Always return true (already at last index).
- **Array with zeros**: A zero in the middle can block progress if you can't jump over it.
- **Large jump values**: Even if nums[i] is very large, you still need to check if position i is reachable.
- **Early termination**: Once max_reach >= len(nums) - 1, you can return true immediately.
- **Off-by-one errors**: Ensure comparison is with len(nums) - 1, not len(nums).

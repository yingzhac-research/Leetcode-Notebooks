## Problem Statement
You are given an array `prices` where `prices[i]` is the price of a given stock on the `i-th` day.

You want to maximize your profit by choosing a **single day** to buy one stock and choosing a **different day in the future** to sell that stock.

Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return `0`.

**Constraints:**
- `1 <= prices.length <= 10^5`
- `0 <= prices[i] <= 10^4`

## Progressive Hints
- Hint 1: For each day, you want to know what the minimum price was before that day.
- Hint 2: The profit on day i is `prices[i] - min_price_so_far`. Track the maximum of these profits.
- Hint 3: You only need one pass through the array, maintaining the minimum price seen and maximum profit found.
- Hint 4: This is a greedy algorithm: always buy at the lowest price seen so far.

## Solution Overview
The optimal greedy approach:
1. Track the minimum price seen so far (potential buy point).
2. For each day, calculate the profit if we sell today: `price - min_price`.
3. Update the maximum profit if current profit is higher.
4. Update the minimum price if today's price is lower.
5. Return the maximum profit.

**Why greedy works**: To maximize profit on selling day i, we should have bought at the minimum price before day i. By tracking the running minimum, we ensure we're always considering the best buy point for each potential sell point.

## Detailed Explanation
1. **Initialize**:
   - `min_price = infinity` (or first price)
   - `max_profit = 0`
2. **Iterate through prices**:
   - For each price:
     - Calculate potential profit: `price - min_price`
     - Update `max_profit` if this is better
     - Update `min_price` if current price is lower
3. **Return** `max_profit`

**Algorithm flow**:
```
min_price = float('inf')
max_profit = 0

for price in prices:
    min_price = min(min_price, price)
    profit = price - min_price
    max_profit = max(max_profit, profit)

return max_profit
```

**Example walkthrough** for `prices = [7,1,5,3,6,4]`:
- Day 0 (7): min_price = 7, profit = 0, max_profit = 0
- Day 1 (1): min_price = 1, profit = 0, max_profit = 0
- Day 2 (5): min_price = 1, profit = 4, max_profit = 4
- Day 3 (3): min_price = 1, profit = 2, max_profit = 4
- Day 4 (6): min_price = 1, profit = 5, max_profit = 5
- Day 5 (4): min_price = 1, profit = 3, max_profit = 5
- Result: 5 (buy at 1, sell at 6)

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| One-pass greedy | O(n) | O(1) | Optimal solution. |
| Brute force | O(n^2) | O(1) | Try all buy-sell pairs. |
| DP with arrays | O(n) | O(n) | Track min price for each day. Unnecessary space. |

## Reference Implementation

````python
from typing import List


def max_profit(prices: List[int]) -> int:
    """
    Find maximum profit from single buy-sell transaction.
    
    Args:
        prices: Daily stock prices
    
    Returns:
        Maximum profit achievable, or 0 if no profit possible
    """
    if not prices:
        return 0
    
    min_price = float('inf')
    max_profit = 0
    
    for price in prices:
        # Update minimum price seen so far
        min_price = min(min_price, price)
        
        # Calculate profit if we sell today
        profit = price - min_price
        
        # Update maximum profit
        max_profit = max(max_profit, profit)
    
    return max_profit
````

## Validation

````python
cases = [
    ([7, 1, 5, 3, 6, 4], 5),  # Buy at 1, sell at 6
    ([7, 6, 4, 3, 1], 0),  # Prices always decreasing, no profit
    ([1, 2], 1),  # Buy at 1, sell at 2
    ([2, 1], 0),  # No profit possible
    ([1], 0),  # Single day, no transaction
    ([3, 3, 5, 0, 0, 3, 1, 4], 4),  # Buy at 0, sell at 4
    ([2, 4, 1], 2),  # Buy at 2, sell at 4
]

for prices, expected in cases:
    result = max_profit(prices)
    assert result == expected, f"Failed for prices={prices}: got {result}, expected {expected}"

print('All tests passed for LC 121.')
````

## Complexity Analysis
- **Time Complexity**: O(n) where n is the number of days
  - Single pass through the prices array
  - Constant time operations per element
- **Space Complexity**: O(1)
  - Only two variables maintained: `min_price` and `max_profit`
  - No additional data structures needed
- **Bottleneck**: The single pass is unavoidable.

## Edge Cases & Pitfalls
- **Empty array**: Return 0 (or handle based on constraints).
- **Single element**: Cannot perform transaction, return 0.
- **Decreasing prices**: No profitable transaction, return 0.
- **All same price**: No profit, return 0.
- **Maximum at the end**: Algorithm naturally handles this.
- **Minimum at the beginning**: Also handled correctly.
- **Integer overflow**: Not an issue in Python, but be careful in other languages.

## Takeaways
- One-pass algorithms are often optimal for sequential problems.
- Track running minimum/maximum to enable greedy decisions.
- For profit maximization: maximize (sell - buy) by minimizing buy and maximizing sell.
- This pattern generalizes to many optimization problems on sequences.
- Greedy works when local optimal choices lead to global optimum.

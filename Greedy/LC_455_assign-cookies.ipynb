## Problem Statement
Assume you are an awesome parent and want to give your children some cookies. But, you should give each child at most one cookie.

Each child `i` has a greed factor `g[i]`, which is the minimum size of a cookie that the child will be content with; and each cookie `j` has a size `s[j]`. If `s[j] >= g[i]`, we can assign the cookie `j` to the child `i`, and the child `i` will be content. Your goal is to maximize the number of your content children and output the maximum number.

**Constraints:**
- `1 <= g.length <= 3 * 10^4`
- `0 <= s.length <= 3 * 10^4`
- `1 <= g[i], s[j] <= 2^31 - 1`

## Progressive Hints
- Hint 1: Think about sorting both arrays. Why would this help?
- Hint 2: Use a greedy strategy: try to satisfy the child with the smallest greed factor using the smallest possible cookie that can satisfy them.
- Hint 3: Use two pointers to iterate through both sorted arrays, advancing the cookie pointer when you find a match.

## Solution Overview
The optimal approach uses a greedy strategy:
1. Sort both the greed factors and cookie sizes in ascending order.
2. Use two pointers to iterate through both arrays.
3. For each child, try to assign the smallest cookie that can satisfy them.
4. This ensures we maximize the number of satisfied children by not wasting larger cookies on children who could be satisfied with smaller ones.

**Why greedy works**: By satisfying children with minimal greed first using minimal cookies, we preserve larger cookies for children who need them, thus maximizing the total count.

## Detailed Explanation
1. **Sort both arrays**: Sort `g` (greed factors) and `s` (cookie sizes) in ascending order.
2. **Initialize two pointers**: `child = 0` for greed array, `cookie = 0` for size array.
3. **Iterate while both pointers are in bounds**:
   - If current cookie can satisfy current child (`s[cookie] >= g[child]`):
     - Increment both pointers (child is satisfied)
   - Else:
     - Increment only cookie pointer (try next larger cookie)
4. **Return the child pointer value**: It represents the number of satisfied children.

**Key insight**: By processing in sorted order, we ensure each child gets the minimal sufficient cookie, leaving larger cookies for greedier children.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy with sorting | O(n log n + m log m) | O(1) | n = len(g), m = len(s). Optimal solution. |
| Brute force matching | O(n * m) | O(1) | Try every cookie for every child. |
| Binary search per child | O(n * log m) | O(1) | After sorting cookies, binary search for each child. |

## Reference Implementation

```python
from typing import List


def find_content_children(g: List[int], s: List[int]) -> int:
    """
    Maximize the number of content children using greedy cookie assignment.
    
    Args:
        g: List of greed factors for children
        s: List of cookie sizes
    
    Returns:
        Maximum number of content children
    """
    # Sort both arrays to enable greedy matching
    g.sort()
    s.sort()
    
    child = 0  # Pointer for children
    cookie = 0  # Pointer for cookies
    
    # Try to satisfy each child with the smallest possible cookie
    while child < len(g) and cookie < len(s):
        if s[cookie] >= g[child]:
            # Current cookie satisfies current child
            child += 1
        # Move to next cookie regardless (either used or too small)
        cookie += 1
    
    return child
```

## Validation

```python
cases = [
    ([1, 2, 3], [1, 1], 1),
    ([1, 2], [1, 2, 3], 2),
    ([1, 2, 7, 10], [1, 3, 5, 9], 3),
    ([10, 9, 8, 7], [5, 6, 7, 8], 2),
    ([], [1, 2, 3], 0),
    ([1, 2, 3], [], 0),
]
for g, s, expected in cases:
    result = find_content_children(g.copy(), s.copy())
    assert result == expected, f"Failed for g={g}, s={s}: got {result}, expected {expected}"
print('All tests passed for LC 455.')
```

## Complexity Analysis
- **Time Complexity**: O(n log n + m log m) where n = len(g) and m = len(s)
  - Sorting g takes O(n log n)
  - Sorting s takes O(m log m)
  - Two-pointer traversal takes O(n + m)
  - Overall dominated by sorting
- **Space Complexity**: O(1) if sorting is done in-place, or O(n + m) if sorting creates copies
- **Bottleneck**: The sorting step dominates the time complexity.

## Edge Cases & Pitfalls
- **Empty arrays**: If either g or s is empty, return 0.
- **No satisfiable children**: All cookies too small for all children.
- **All children satisfied**: More than enough cookies for all children.
- **Sorting side effects**: The input arrays are modified by sorting; make copies if needed.
- **Integer overflow**: Not a concern in Python, but relevant in other languages.

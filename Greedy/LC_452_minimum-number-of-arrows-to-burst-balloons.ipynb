## Problem Statement
There are some spherical balloons taped onto a flat wall that represents the XY-plane. The balloons are represented as a 2D integer array `points` where `points[i] = [x_start, x_end]` denotes a balloon whose horizontal diameter stretches between `x_start` and `x_end`. You do not know the exact y-coordinates of the balloons.

Arrows can be shot up directly vertically (in the positive y-direction) from different points along the x-axis. A balloon with `x_start` and `x_end` is burst by an arrow shot at `x` if `x_start <= x <= x_end`. There is no limit to the number of arrows that can be shot. A shot arrow keeps traveling up infinitely, bursting any balloons in its path.

Given the array `points`, return the minimum number of arrows that must be shot to burst all balloons.

**Constraints:**
- `1 <= points.length <= 10^5`
- `points[i].length == 2`
- `-2^31 <= x_start < x_end <= 2^31 - 1`

## Progressive Hints
- Hint 1: Think about sorting the balloons by their end positions. Why would this help?
- Hint 2: Use a greedy strategy: shoot an arrow at the earliest possible position that covers the current balloon.
- Hint 3: Track the rightmost position of the current arrow. If a new balloon doesn't overlap with this position, you need a new arrow.
- Hint 4: This problem is similar to finding the minimum number of non-overlapping intervals.

## Solution Overview
The optimal approach uses a greedy interval scheduling strategy:
1. Sort balloons by their end position.
2. Shoot the first arrow at the end of the first balloon.
3. For each subsequent balloon, check if it overlaps with the current arrow position.
4. If it doesn't overlap, shoot a new arrow at the end of this balloon.

**Why greedy works**: By sorting by end position and shooting arrows at the rightmost valid position, we maximize the chance that subsequent balloons will be covered by the same arrow.

## Detailed Explanation
1. **Sort balloons by end position**: `points.sort(key=lambda x: x[1])`
2. **Initialize arrow count**: Start with 1 arrow (at least one is always needed).
3. **Track current arrow position**: Set to the end of the first balloon.
4. **Iterate through remaining balloons**:
   - If current balloon's start is after the arrow position, it's not covered.
   - Increment arrow count and update arrow position to the end of this balloon.
5. **Return arrow count**.

**Key insight**: Shooting at the end of each interval cluster maximizes coverage because any balloon that ends later but starts before this point will be covered.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy with sorting by end | O(n log n) | O(1) | Optimal solution. |
| Greedy with sorting by start | O(n log n) | O(1) | More complex tracking logic. |
| Interval merging | O(n log n) | O(n) | Merge overlapping intervals first. |
| Brute force | O(2^n) | O(n) | Try all possible arrow combinations. |

## Reference Implementation

```python
from typing import List


def find_min_arrow_shots(points: List[List[int]]) -> int:
    """
    Find minimum number of arrows needed to burst all balloons.
    
    Args:
        points: List of balloon intervals [x_start, x_end]
    
    Returns:
        Minimum number of arrows required
    """
    if not points:
        return 0
    
    # Sort balloons by their end position
    points.sort(key=lambda x: x[1])
    
    arrows = 1  # At least one arrow is needed
    current_arrow_pos = points[0][1]  # Shoot at the end of first balloon
    
    for i in range(1, len(points)):
        # If current balloon starts after the arrow position, need new arrow
        if points[i][0] > current_arrow_pos:
            arrows += 1
            current_arrow_pos = points[i][1]  # Shoot at end of this balloon
    
    return arrows
```

## Validation

```python
cases = [
    ([[10, 16], [2, 8], [1, 6], [7, 12]], 2),
    ([[1, 2], [3, 4], [5, 6], [7, 8]], 4),
    ([[1, 2], [2, 3], [3, 4], [4, 5]], 2),
    ([[1, 2]], 1),
    ([[1, 10], [2, 9], [3, 8], [4, 7]], 1),
    ([[-2147483646, -2147483645], [2147483646, 2147483647]], 2),
]
for points, expected in cases:
    result = find_min_arrow_shots([p.copy() for p in points])
    assert result == expected, f"Failed for points={points}: got {result}, expected {expected}"
print('All tests passed for LC 452.')
```

## Complexity Analysis
- **Time Complexity**: O(n log n) where n = len(points)
  - Sorting takes O(n log n)
  - Single pass through sorted array takes O(n)
  - Overall dominated by sorting
- **Space Complexity**: O(1) if sorting in-place, or O(n) for sorting overhead
  - No additional data structures needed
- **Bottleneck**: The sorting step dominates the time complexity.

## Edge Cases & Pitfalls
- **Empty array**: Should return 0 (no balloons to burst).
- **Single balloon**: Always requires exactly 1 arrow.
- **All overlapping**: All balloons can be burst with 1 arrow.
- **No overlaps**: Each balloon requires its own arrow.
- **Integer overflow**: Be careful with extreme values near INT_MAX/INT_MIN.
- **Inclusive endpoints**: The problem states x_start <= x <= x_end, so touching endpoints count as overlap.

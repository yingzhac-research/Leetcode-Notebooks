## Problem Statement
Given a characters array `tasks`, representing the tasks a CPU needs to do, where each letter represents a different task. Tasks could be done in any order. Each task is done in one unit of time. For each unit of time, the CPU could complete either one task or just be idle.

However, there is a non-negative integer `n` that represents the cooldown period between two same tasks (the same letter in the array), that is, there must be at least `n` units of time between any two same tasks.

Return the least number of units of time that the CPU will take to finish all the given tasks.

**Constraints:**
- `1 <= tasks.length <= 10^4`
- `tasks[i]` is an uppercase English letter.
- `0 <= n <= 100`

## Progressive Hints
- Hint 1: Focus on the task that appears most frequently. It will determine the minimum time needed.
- Hint 2: The most frequent task creates "frames" of size (n+1). Calculate how many such frames you need.
- Hint 3: If there are multiple tasks with the same maximum frequency, they all need space in the last frame. The answer is either the calculated frame time or the total number of tasks, whichever is larger.

## Solution Overview
The optimal greedy approach uses mathematical calculation based on task frequencies:
1. Count the frequency of each task.
2. Find the maximum frequency.
3. Calculate the number of idle slots needed based on the most frequent task.
4. Fill idle slots with other tasks.
5. Return max(calculated_time, total_tasks).

**Key insight**: The most frequent task(s) create a schedule framework. If we have task 'A' appearing max_freq times with cooldown n, we need (max_freq - 1) * (n + 1) + count_of_max_freq_tasks time slots.

## Detailed Explanation
**Mathematical Approach**:

1. **Count frequencies**: Use Counter to get frequency of each task.

2. **Find max_freq**: The highest frequency among all tasks.

3. **Count tasks with max_freq**: How many tasks have this maximum frequency.

4. **Calculate minimum time**:
   ```
   Frame structure: [Task1, ..., idle] repeated (max_freq - 1) times
   Frame size: n + 1
   Number of frames: max_freq - 1
   Last frame: Only tasks with max_freq
   
   Total time = (max_freq - 1) * (n + 1) + max_count
   ```

5. **Handle overflow**: If tasks fill all slots, no idle time needed:
   ```
   result = max(calculated_time, len(tasks))
   ```

**Example**: tasks = ["A","A","A","B","B","B"], n = 2
- max_freq = 3, max_count = 2 (both A and B)
- Time = (3-1) * (2+1) + 2 = 2*3 + 2 = 8
- Schedule: A -> B -> idle -> A -> B -> idle -> A -> B

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Math formula | O(n) | O(1) | Count frequencies (O(n)), then constant calculation. |
| Heap simulation | O(n log k) | O(k) | k = unique tasks. Simulate actual scheduling. |
| Greedy fill | O(n log k) | O(k) | Sort by frequency and fill slots. |

## Reference Implementation

````python
from typing import List
from collections import Counter


def least_interval(tasks: List[str], n: int) -> int:
    """
    Calculate minimum time needed to complete all tasks with cooldown.
    
    Args:
        tasks: List of tasks (characters)
        n: Cooldown period between same tasks
    
    Returns:
        Minimum units of time needed
    """
    # Count frequency of each task
    freq = Counter(tasks)
    
    # Find the maximum frequency
    max_freq = max(freq.values())
    
    # Count how many tasks have the maximum frequency
    max_count = sum(1 for count in freq.values() if count == max_freq)
    
    # Calculate minimum time
    # (max_freq - 1): number of complete frames
    # (n + 1): size of each frame
    # max_count: tasks in the last incomplete frame
    calculated_time = (max_freq - 1) * (n + 1) + max_count
    
    # If we have enough tasks to fill all slots, no idle time needed
    return max(calculated_time, len(tasks))
````

## Validation

````python
cases = [
    (["A","A","A","B","B","B"], 2, 8),  # A->B->idle->A->B->idle->A->B
    (["A","C","A","B","D","B"], 1, 6),  # No idle needed
    (["A","A","A","B","B","B"], 0, 6),  # No cooldown
    (["A","A","A","A","A","A","B","C","D","E","F","G"], 2, 16),
    (["A","A","A","B","B","B","C","C","C","D","D","E"], 2, 12),
    (["A"], 2, 1),  # Single task
]
for tasks, n, expected in cases:
    result = least_interval(tasks, n)
    assert result == expected, f"Failed for tasks={tasks}, n={n}: got {result}, expected {expected}"
print('All tests passed for LC 621.')
````

## Complexity Analysis
- **Time Complexity**: O(n) where n is the number of tasks
  - Counting frequencies: O(n)
  - Finding max frequency: O(26) = O(1) since at most 26 unique letters
  - Counting max_freq tasks: O(26) = O(1)
  - Overall: O(n)
- **Space Complexity**: O(1)
  - Counter stores at most 26 task types
  - Constant space regardless of input size
- **Bottleneck**: Single pass to count frequencies dominates.

## Edge Cases & Pitfalls
- **n = 0**: No cooldown, result is simply len(tasks).
- **Single task type**: Maximum idle time needed.
- **Many task types**: If enough diversity, no idle time needed (result = len(tasks)).
- **Multiple tasks with max frequency**: All need space in last frame.
- **Large n with few tasks**: Many idle slots.
- **Off-by-one errors**: Remember it's (max_freq - 1) frames, not max_freq.

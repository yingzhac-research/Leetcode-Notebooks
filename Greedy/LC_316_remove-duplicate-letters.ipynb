## Problem Statement
Given a string `s`, remove duplicate letters so that every letter appears once and only once. You must make sure your result is the smallest in lexicographical order among all possible results.

**Constraints:**
- `1 <= s.length <= 10^4`
- `s` consists of lowercase English letters.

## Progressive Hints
- Hint 1: Think about using a stack to build the result incrementally.
- Hint 2: Keep track of which characters are already in the result and which characters remain in the string.
- Hint 3: When deciding whether to add a character, check if removing the previous character would result in a lexicographically smaller string.
- Hint 4: You can only remove a character from the result if it appears later in the string.

## Solution Overview
The optimal approach uses a greedy monotonic stack strategy:
1. Count the frequency of each character.
2. Use a stack to build the result and a set to track included characters.
3. For each character:
   - Skip if already in result.
   - While stack top > current char AND stack top appears later, pop it.
   - Add current character to stack.
4. Convert stack to string.

**Why greedy works**: By maintaining a monotonically increasing stack and only removing characters that appear later, we ensure the lexicographically smallest result.

## Detailed Explanation
1. **Count character frequencies**: Use a dictionary to store remaining occurrences.
2. **Initialize stack and set**: Stack for building result, set for tracking included chars.
3. **Iterate through string**:
   - Decrement frequency for current character.
   - If character already in result, skip.
   - While stack is not empty AND:
     - Stack top > current char (lexicographically)
     - Stack top has remaining occurrences
     - Pop from stack and remove from set
   - Push current character to stack and add to set.
4. **Return stack as string**.

**Key insight**: We can safely remove a character if it appears later, allowing us to place a smaller character first.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy monotonic stack | O(n) | O(1) | Optimal, at most 26 chars in stack. |
| Recursive | O(n^2) | O(n) | Try removing each duplicate. |
| Brute force permutations | O(n!) | O(n) | Generate all orderings. |
| Sorting approach | O(n log n) | O(n) | Doesn't preserve order correctly. |

## Reference Implementation

```python
from collections import Counter


def remove_duplicate_letters(s: str) -> str:
    """
    Remove duplicate letters to get lexicographically smallest result.
    
    Args:
        s: Input string
    
    Returns:
        Lexicographically smallest string with each character appearing once
    """
    # Count remaining occurrences of each character
    char_count = Counter(s)
    
    # Track which characters are in the result
    in_result = set()
    
    # Stack to build the result
    stack = []
    
    for char in s:
        # Decrement count as we process this character
        char_count[char] -= 1
        
        # Skip if already in result
        if char in in_result:
            continue
        
        # Remove larger characters that appear later
        while stack and stack[-1] > char and char_count[stack[-1]] > 0:
            removed = stack.pop()
            in_result.remove(removed)
        
        # Add current character
        stack.append(char)
        in_result.add(char)
    
    return ''.join(stack)
```

## Validation

```python
cases = [
    ("bcabc", "abc"),
    ("cbacdcbc", "acdb"),
    ("ecbacba", "eacb"),
    ("abacb", "abc"),
    ("abcd", "abcd"),
    ("dcba", "dcba"),
    ("bbcaac", "bac"),
]
for s, expected in cases:
    result = remove_duplicate_letters(s)
    assert result == expected, f"Failed for s={s}: got {result}, expected {expected}"
print('All tests passed for LC 316.')
```

## Complexity Analysis
- **Time Complexity**: O(n) where n = len(s)
  - Each character is pushed and popped from stack at most once
  - Counting frequencies: O(n)
  - Overall linear time
- **Space Complexity**: O(1) or O(26) = O(1)
  - Stack contains at most 26 unique lowercase letters
  - Set also contains at most 26 characters
  - Counter contains at most 26 entries
- **Bottleneck**: Single pass through the string is optimal.

## Edge Cases & Pitfalls
- **No duplicates**: Return the original string.
- **All same character**: Return single character.
- **Already sorted**: No need to remove anything.
- **Reverse sorted**: Need to keep all if each appears only once.
- **Early vs late occurrence**: Must check if character appears later before removing.
- **Greedy vs optimal**: The greedy choice at each step leads to global optimum.

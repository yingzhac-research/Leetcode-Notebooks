# 826. Most Profit Assigning Work

- **Scenario**: Similar to workload distribution in cloud systems where tasks have different resource requirements and rewards.

## Problem Statement
You have `n` jobs and `m` workers. You are given three arrays: `difficulty`, `profit`, and `worker` where:
- `difficulty[i]` and `profit[i]` are the difficulty and the profit of the ith job, and
- `worker[j]` is the ability of jth worker (i.e., the jth worker can only complete a job with difficulty at most `worker[j]`).

Every worker can be assigned at most one job, but one job can be completed multiple times.

Return the maximum profit we can achieve after assigning the workers to the jobs.

**Constraints:**
- `n == difficulty.length`
- `n == profit.length`
- `m == worker.length`
- `1 <= n, m <= 10^4`
- `1 <= difficulty[i], profit[i], worker[i] <= 10^5`

## Progressive Hints
- Hint 1: For each worker, you want to find the job with maximum profit that they can do.
- Hint 2: Sort jobs by difficulty, and sort workers by ability.
- Hint 3: As you iterate through sorted workers, keep track of the maximum profit achievable so far.
- Hint 4: Use two pointers or binary search to efficiently find the best job for each worker.

## Solution Overview
The optimal approach uses greedy selection with sorting:
1. Combine difficulty and profit into job pairs and sort by difficulty.
2. Sort workers by ability.
3. For each worker, find all jobs they can do and track the maximum profit seen.
4. Assign the maximum profit job to each worker.

**Why greedy works**: By sorting both jobs and workers, we can efficiently track the best available job for each worker level without rechecking all jobs.

## Detailed Explanation
1. **Create job pairs**: Combine difficulty and profit into `[(difficulty[i], profit[i])]`
2. **Sort jobs by difficulty**: `jobs.sort()`
3. **Sort workers by ability**: `worker.sort()`
4. **Initialize variables**: `max_profit = 0`, `job_idx = 0`, `total_profit = 0`
5. **For each worker**:
   - While there are jobs the worker can do (difficulty <= worker ability):
     - Update max_profit to the best profit seen so far
     - Move to next job
   - Add max_profit to total_profit
6. **Return total_profit**.

**Key insight**: Once we know the best job for workers of ability X, we also know the best job for all workers with ability > X.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy two pointers | O(n log n + m log m) | O(n) | Optimal solution. |
| Binary search per worker | O(m * log n) | O(n) | After sorting, binary search for each worker. |
| Brute force | O(n * m) | O(1) | Check all jobs for each worker. |
| DP | O(n * max_difficulty) | O(max_difficulty) | Overkill for this problem. |

## Reference Implementation

```python
from typing import List


def max_profit_assignment(difficulty: List[int], profit: List[int], worker: List[int]) -> int:
    """
    Find maximum profit by optimally assigning workers to jobs.
    
    Args:
        difficulty: List of job difficulties
        profit: List of job profits
        worker: List of worker abilities
    
    Returns:
        Maximum total profit achievable
    """
    # Combine difficulty and profit, then sort by difficulty
    jobs = sorted(zip(difficulty, profit))
    
    # Sort workers by ability
    worker.sort()
    
    max_profit = 0  # Best profit seen so far for current ability level
    job_idx = 0  # Current job index
    total_profit = 0  # Total profit accumulated
    
    for ability in worker:
        # Update max_profit with all jobs this worker can do
        while job_idx < len(jobs) and jobs[job_idx][0] <= ability:
            max_profit = max(max_profit, jobs[job_idx][1])
            job_idx += 1
        
        # Assign the best job (highest profit) this worker can do
        total_profit += max_profit
    
    return total_profit
```

## Validation

```python
cases = [
    ([2, 4, 6, 8, 10], [10, 20, 30, 40, 50], [4, 5, 6, 7], 100),
    ([85, 47, 57], [24, 66, 99], [40, 25, 25], 0),
    ([68, 35, 52, 47, 86], [67, 17, 1, 81, 3], [92, 10, 85, 84, 82], 324),
    ([5, 10], [100, 200], [6, 11], 300),
    ([1], [100], [10, 20, 30], 300),
]
for difficulty, profit, worker, expected in cases:
    result = max_profit_assignment(difficulty.copy(), profit.copy(), worker.copy())
    assert result == expected, f"Failed for difficulty={difficulty}, profit={profit}, worker={worker}: got {result}, expected {expected}"
print('All tests passed for LC 826.')
```

## Complexity Analysis
- **Time Complexity**: O(n log n + m log m) where n = len(jobs), m = len(workers)
  - Sorting jobs: O(n log n)
  - Sorting workers: O(m log m)
  - Two-pointer traversal: O(n + m)
  - Overall dominated by sorting
- **Space Complexity**: O(n) for storing job pairs
- **Bottleneck**: Sorting both arrays dominates the time complexity.

## Edge Cases & Pitfalls
- **No worker can do any job**: All worker abilities < minimum difficulty, return 0.
- **All workers can do all jobs**: Assign the highest profit job to everyone.
- **Jobs with same difficulty**: May have different profits; take the maximum.
- **Workers with same ability**: They all get the same best job.
- **Job reuse**: One job can be assigned to multiple workers.
- **Sorting side effects**: Input arrays are modified; copy if needed.

## Problem Statement
The **numeric value** of a lowercase character is defined as its position `(1-indexed)` in the alphabet, so the numeric value of `a` is `1`, the numeric value of `b` is `2`, the numeric value of `c` is `3`, and so on.

The **numeric value** of a string consisting of lowercase characters is defined as the sum of its characters' numeric values. For example, the numeric value of the string `"abe"` is equal to `1 + 2 + 5 = 8`.

You are given two integers `n` and `k`. Return the **lexicographically smallest string** with **length** equal to `n` and **numeric value** equal to `k`.

Note that a string `x` is lexicographically smaller than string `y` if `x` comes before `y` in dictionary order, that is, either `x` is a prefix of `y`, or if `i` is the first position such that `x[i] != y[i]`, then `x[i]` comes before `y[i]` in alphabetic order.

**Constraints:**
- `1 <= n <= 10^5`
- `n <= k <= 26 * n`

## Progressive Hints
- Hint 1: To get the lexicographically smallest string, you want as many 'a's at the beginning as possible.
- Hint 2: Start with all 'a's and then adjust from the end to meet the numeric value requirement.
- Hint 3: Work backwards: for each position from the end, use the largest letter (up to 'z') that doesn't overshoot the target.
- Hint 4: Calculate how much value you need beyond the minimum (all 'a's = n), then distribute it greedily from right to left using 'z's.

## Solution Overview
The optimal greedy approach:
1. Start with a string of all 'a's (length n, value n).
2. Calculate the deficit: `deficit = k - n`.
3. Work from the rightmost position backwards.
4. For each position, add as much value as possible (up to 25, to make it 'z').
5. Continue until the deficit is zero.

**Why greedy works**: To be lexicographically smallest, we want small characters at the front. By filling from the right with large characters, we ensure the left side has the smallest possible characters.

## Detailed Explanation
1. **Initialize**: Create a list of 'a's with length n. This gives us a baseline value of n.
2. **Calculate deficit**: `deficit = k - n` (how much more value we need).
3. **Greedy fill from right**:
   - Start from the last position (index n-1)
   - For each position:
     - Calculate how much we can add: `min(deficit, 25)` (25 = 'z' - 'a')
     - Update the character: `result[i] = chr(ord('a') + add)`
     - Reduce deficit: `deficit -= add`
     - If deficit becomes 0, we're done
4. **Return**: Join the character list into a string.

**Algorithm flow**:
```
result = ['a'] * n
deficit = k - n
i = n - 1

while deficit > 0:
    add = min(deficit, 25)
    result[i] = chr(ord('a') + add)
    deficit -= add
    i -= 1

return ''.join(result)
```

**Example walkthrough** for `n = 5, k = 73`:
- Start: ['a', 'a', 'a', 'a', 'a'], value = 5
- Deficit: 73 - 5 = 68
- Position 4: add min(68, 25) = 25 → 'z', deficit = 43
- Position 3: add min(43, 25) = 25 → 'z', deficit = 18
- Position 2: add min(18, 25) = 18 → 's', deficit = 0
- Result: "aaszz"

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy from right | O(n) | O(n) | Optimal solution. |
| Math optimization | O(n) | O(n) | Calculate positions directly. Same complexity. |
| Backtracking | O(26^n) | O(n) | Try all combinations. Too slow. |

## Reference Implementation

````python
def get_smallest_string(n: int, k: int) -> str:
    """
    Generate lexicographically smallest string of length n with numeric value k.
    
    Args:
        n: Length of the string
        k: Target numeric value
    
    Returns:
        Lexicographically smallest string meeting the constraints
    """
    # Start with all 'a's
    result = ['a'] * n
    
    # Calculate how much more value we need
    deficit = k - n  # n 'a's give us value n
    
    # Fill from right to left with larger characters
    i = n - 1
    while deficit > 0:
        # Add as much as possible (max 25 to reach 'z')
        add = min(deficit, 25)
        result[i] = chr(ord('a') + add)
        deficit -= add
        i -= 1
    
    return ''.join(result)
````

## Validation

````python
def numeric_value(s: str) -> int:
    """Calculate numeric value of a string."""
    return sum(ord(c) - ord('a') + 1 for c in s)


cases = [
    (3, 27, "aay"),
    (5, 73, "aaszz"),
    (1, 1, "a"),
    (1, 26, "z"),
    (2, 2, "aa"),
    (2, 52, "zz"),
    (10, 100, "aaaaaamnzz"),
]

for n, k, expected in cases:
    result = get_smallest_string(n, k)
    # Check length and value
    assert len(result) == n, f"Wrong length for n={n}, k={k}: got {result}"
    assert numeric_value(result) == k, f"Wrong value for n={n}, k={k}: got {result} with value {numeric_value(result)}"
    # Check lexicographic order
    assert result == expected, f"Not lexicographically smallest for n={n}, k={k}: got {result}, expected {expected}"

print('All tests passed for LC 1663.')
````

## Complexity Analysis
- **Time Complexity**: O(n) where n is the length of the string
  - Initialize array: O(n)
  - Fill from right: O(n) worst case (if k is close to n)
  - Join array: O(n)
  - Overall: O(n)
- **Space Complexity**: O(n) for the result array
- **Bottleneck**: Linear scan is unavoidable as we must construct the entire string.

## Edge Cases & Pitfalls
- **Minimum case**: `n = 1, k = 1` → "a"
- **Maximum case**: `k = 26 * n` → all 'z's
- **All 'a's**: When `k = n`
- **Off-by-one**: Remember 'a' has value 1, not 0
- **Index bounds**: Ensure you don't go below index 0 when filling from right
- **Character arithmetic**: `chr(ord('a') + value)` where value is 0-25

## Takeaways
- For lexicographic ordering, think about which positions have the most impact (leftmost).
- Greedy fill from right allows you to keep left positions minimal.
- When dealing with character values, remember ASCII arithmetic: `ord()` and `chr()`.
- Deficit-based thinking: start with minimum and add what's needed.
- Working backwards is often useful for lexicographic optimization problems.

## Problem Statement
You are given a string `s`. We want to partition the string into as many parts as possible so that each letter appears in at most one part.

Note that the partition is done so that after concatenating all the parts in order, the resultant string should be `s`.

Return a list of integers representing the size of these parts.

**Constraints:**
- `1 <= s.length <= 500`
- `s` consists of lowercase English letters.

## Progressive Hints
- Hint 1: Track the last occurrence index of each character.
- Hint 2: Use a greedy approach: extend the current partition until you've included all occurrences of all characters seen so far.
- Hint 3: Keep track of the farthest index you need to reach to complete the current partition.
- Hint 4: When you reach the end of a partition, start a new one.

## Solution Overview
The optimal approach uses a greedy interval merging strategy:
1. Record the last occurrence index of each character.
2. Iterate through the string, tracking the farthest index needed for current partition.
3. When current index reaches the partition end, complete the partition.
4. Start a new partition and repeat.

**Why greedy works**: A character's last occurrence determines the minimum partition size. Once we reach it and have covered all previous characters' last occurrences, we can safely end the partition.

## Detailed Explanation
1. **Build last occurrence map**: `last = {char: idx for idx, char in enumerate(s)}`
2. **Initialize variables**: `start = 0`, `end = 0`, `result = []`
3. **Iterate through string with index**:
   - Update end to be max of current end and last occurrence of current char.
   - If current index equals end, we've completed a partition:
     - Add partition size to result.
     - Update start to next position.
4. **Return result**.

**Key insight**: A partition must extend to the last occurrence of every character it contains. By tracking the maximum last occurrence, we know when a partition can end.

## Complexity Trade-off Table
| Approach | Time Complexity | Space Complexity | Notes |
| --- | --- | --- | --- |
| Greedy with last occurrence | O(n) | O(1) | Optimal, O(26) space for map. |
| Interval merging | O(n) | O(n) | Build intervals then merge. |
| Two-pass approach | O(n) | O(n) | Similar to greedy. |
| Brute force | O(n^2) | O(n) | Check all possible partitions. |

## Reference Implementation

```python
from typing import List


def partition_labels(s: str) -> List[int]:
    """
    Partition string into maximum number of parts where each letter appears in at most one part.
    
    Args:
        s: Input string
    
    Returns:
        List of partition sizes
    """
    # Record the last occurrence of each character
    last_occurrence = {char: idx for idx, char in enumerate(s)}
    
    result = []
    start = 0  # Start of current partition
    end = 0    # End of current partition
    
    for idx, char in enumerate(s):
        # Extend partition to include last occurrence of current character
        end = max(end, last_occurrence[char])
        
        # If we've reached the end of the partition
        if idx == end:
            result.append(end - start + 1)
            start = idx + 1  # Start new partition
    
    return result
```

## Validation

```python
cases = [
    ("ababcbacadefegdehijhklij", [9, 7, 8]),
    ("eccbbbbdec", [10]),
    ("abcdef", [1, 1, 1, 1, 1, 1]),
    ("aa", [2]),
    ("abcabc", [6]),
    ("eababcbaca", [10]),
]
for s, expected in cases:
    result = partition_labels(s)
    assert result == expected, f"Failed for s={s}: got {result}, expected {expected}"
print('All tests passed for LC 763.')
```

## Complexity Analysis
- **Time Complexity**: O(n) where n = len(s)
  - Building last occurrence map: O(n)
  - Single pass through string: O(n)
  - Overall linear time
- **Space Complexity**: O(1) or O(26) = O(1)
  - Last occurrence map contains at most 26 lowercase letters
  - Result list size depends on number of partitions (worst case O(n))
- **Bottleneck**: Single pass through the string is optimal.

## Edge Cases & Pitfalls
- **All same character**: Single partition of length n.
- **All unique characters**: n partitions of size 1.
- **Two interleaved groups**: Two partitions.
- **Single character**: Return [1].
- **Overlapping character ranges**: Must extend partition to cover all.
- **Order preservation**: Partitions must maintain original string order.
